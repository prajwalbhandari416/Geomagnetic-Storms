```python
# ============================================================
# GEOMAGNETIC STORM PREDICTION USING DEEP LEARNING
# Based on Kp × 10 and Dst
#
# Author: Prajwal Bhandari
# Date: September 2026
#
# Models:
#   1. LSTM
#   2. BiLSTM
#   3. CNN-LSTM
#
#
# Targets:
#   - Dst
#   - Kp × 10
#
# Input features:
#   YEAR, DOY, HR, B, Bx, By, Bz, PD, FP, E, Kp &, Dst
# ============================================================


# ============================================================
# 0. INSTALL / IMPORT LIBRARIES
# ============================================================

# Google Colab:
# !pip install -q tensorflow scikit-learn scipy seaborn

import os
import re
import random
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_curve,
    auc,
    precision_recall_curve,
    average_precision_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (
    Input,
    LSTM,
    Dense,
    Dropout,
    Bidirectional,
    Conv1D,
    MaxPooling1D,
    GlobalAveragePooling1D,
    MultiHeadAttention,
    LayerNormalization,
    Add
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

from tensorflow.keras.optimizers import Adam


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)


# ============================================================
# 2. CONFIGURATION
# ============================================================

# ------------------------------------------------------------
# Change this path according to your CSV file
# ------------------------------------------------------------

DATA_PATH = "/content/D.csv"

# If your file is in Google Drive, for example:
# DATA_PATH = "/content/drive/MyDrive/D.csv"


# ------------------------------------------------------------
# Output directory
# ------------------------------------------------------------

OUTPUT_DIR = "/content/geomagnetic_results"

os.makedirs(OUTPUT_DIR, exist_ok=True)


# ------------------------------------------------------------
# Time-series configuration
# ------------------------------------------------------------

LOOKBACK = 48

TEST_SIZE = 0.20
VALIDATION_SIZE = 0.15

EPOCHS = 50
BATCH_SIZE = 32

LEARNING_RATE = 0.001


# ============================================================
# 3. COLUMN DEFINITIONS
# ============================================================

# Expected columns in your dataset

EXPECTED_COLUMNS = [
    "YEAR",
    "DOY",
    "HR",
    "B",
    "Bx",
    "By",
    "Bz",
    "PD",
    "FP",
    "E",
    "Kp",
    "Dst"
]


# ============================================================
# 4. DATA PREPROCESSOR
# ============================================================

class DataPreprocessor:
    """
    Loads, cleans and prepares geomagnetic data.
    """

    def __init__(self, filepath):

        self.filepath = filepath
        self.df = None

        # Missing/invalid values commonly used in
        # geomagnetic datasets
        self.missing_values_map = {
            "B": [999.99, 9999.99],
            "Bx": [999.99, 9999.99],
            "By": [999.99, 9999.99],
            "Bz": [999.99, 9999.99],
            "PD": [999.9, 999.99, 9999.9],
            "FP": [99.99, 999.99],
            "E": [999.99, 9999.99],
            "Kp": [9.99, 99.99],
            "Dst": [9999.9, 9999.99]
        }


    # --------------------------------------------------------
    # Normalize column names
    # --------------------------------------------------------

    def normalize_column_names(self):

        new_columns = []

        for col in self.df.columns:

            col = str(col).strip()

            # Remove special characters
            col_clean = re.sub(r"\s+", " ", col)

            # Kp column can appear as:
            # "Kp &"
            # "Kp"
            # "Kp*10"
            # "Kp × 10"

            if "kp" in col_clean.lower():
                col_clean = "Kp"

            elif col_clean.lower() == "dst":
                col_clean = "Dst"

            new_columns.append(col_clean)

        self.df.columns = new_columns

        return self


    # --------------------------------------------------------
    # Load data
    # --------------------------------------------------------

    def load_data(self):

        print("=" * 75)
        print("LOADING DATA")
        print("=" * 75)

        self.df = pd.read_csv(self.filepath)

        print("\nOriginal dataset shape:")
        print(self.df.shape)

        print("\nOriginal columns:")
        print(list(self.df.columns))

        self.normalize_column_names()

        print("\nNormalized columns:")
        print(list(self.df.columns))

        return self


    # --------------------------------------------------------
    # Check required columns
    # --------------------------------------------------------

    def check_columns(self):

        required = [
            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E",
            "Kp",
            "Dst"
        ]

        missing = [
            col for col in required
            if col not in self.df.columns
        ]

        if len(missing) > 0:

            raise ValueError(
                "\nMissing required columns:\n"
                + str(missing)
                + "\n\nAvailable columns:\n"
                + str(list(self.df.columns))
            )

        print("\nAll required columns found.")

        return self


    # --------------------------------------------------------
    # Handle missing values
    # --------------------------------------------------------

    def handle_missing_values(self):

        print("\n" + "=" * 75)
        print("HANDLING MISSING VALUES")
        print("=" * 75)

        for col, values in self.missing_values_map.items():

            if col in self.df.columns:

                for value in values:

                    count = (
                        self.df[col] == value
                    ).sum()

                    if count > 0:

                        self.df[col] = self.df[col].replace(
                            value,
                            np.nan
                        )

                        print(
                            f"{col}: replaced {count} "
                            f"values of {value}"
                        )

        return self


    # --------------------------------------------------------
    # Clean data
    # --------------------------------------------------------

    def clean_data(self):

        print("\n" + "=" * 75)
        print("CLEANING DATA")
        print("=" * 75)

        # Convert all columns to numeric

        for col in self.df.columns:

            self.df[col] = pd.to_numeric(
                self.df[col],
                errors="coerce"
            )

        # Sort chronologically

        self.df = self.df.sort_values(
            ["YEAR", "DOY", "HR"]
        ).reset_index(drop=True)


        # Display missing statistics

        print("\nMissing values before interpolation:")

        missing = self.df.isna().sum()

        print(
            missing[missing > 0]
        )


        # Interpolate numerical data

        numeric_columns = self.df.select_dtypes(
            include=[np.number]
        ).columns

        self.df[numeric_columns] = (
            self.df[numeric_columns]
            .interpolate(method="linear")
            .ffill()
            .bfill()
        )


        # Remove remaining NaN

        before = len(self.df)

        self.df = self.df.dropna()

        after = len(self.df)

        print(
            f"\nRows removed: {before - after}"
        )

        print(
            f"Final dataset shape: {self.df.shape}"
        )


        # ----------------------------------------------------
        # Create Kp × 10
        # ----------------------------------------------------

        # If the input Kp is already Kp × 10
        # values such as 0, 10, 20, ..., 90,
        # do not multiply again.
        #
        # If input values are 0–9,
        # multiply by 10.

        kp_max = self.df["Kp"].max()

        if kp_max <= 9.0:

            print(
                "\nKp values appear to be standard 0–9 scale."
            )

            print(
                "Creating Kp10 = Kp × 10."
            )

            self.df["Kp10"] = (
                self.df["Kp"] * 10.0
            )

        else:

            print(
                "\nKp values appear to already be Kp × 10."
            )

            self.df["Kp10"] = (
                self.df["Kp"]
            )


        # Remove original Kp column
        # because Kp10 will be used as target

        self.df.drop(
            columns=["Kp"],
            inplace=True
        )


        # Ensure correct order

        final_columns = [
            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E",
            "Kp10",
            "Dst"
        ]

        self.df = self.df[
            final_columns
        ]


        print("\nFinal columns:")

        print(
            list(self.df.columns)
        )


        print("\nKp10 statistics:")

        print(
            self.df["Kp10"].describe()
        )


        print("\nDst statistics:")

        print(
            self.df["Dst"].describe()
        )


        return self


    def get_processed_data(self):

        return self.df


# ============================================================
# 5. TIME SERIES SEQUENCE GENERATOR
# ============================================================

class SequenceGenerator:

    """
    Creates sequences for multi-output forecasting.

    Inputs:
        Historical solar-wind + geomagnetic parameters

    Outputs:
        Future Dst
        Future Kp × 10
    """

    def __init__(
        self,
        data,
        lookback=48
    ):

        self.data = data
        self.lookback = lookback

        self.scaler = MinMaxScaler()

        self.feature_columns = [
            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E"
        ]

        self.target_columns = [
            "Kp10",
            "Dst"
        ]

        self.all_columns = (
            self.feature_columns
            + self.target_columns
        )


    # --------------------------------------------------------
    # Create sequences
    # --------------------------------------------------------

    def create_sequences(self):

        print("\n" + "=" * 75)
        print("CREATING TIME-SERIES SEQUENCES")
        print("=" * 75)

        data_values = self.data[
            self.all_columns
        ].values


        # Scale data

        scaled_data = self.scaler.fit_transform(
            data_values
        )


        kp_index = self.all_columns.index(
            "Kp10"
        )

        dst_index = self.all_columns.index(
            "Dst"
        )


        X = []
        y_kp = []
        y_dst = []


        for i in range(
            len(scaled_data) - self.lookback
        ):

            X.append(
                scaled_data[
                    i:i + self.lookback
                ]
            )

            y_kp.append(
                scaled_data[
                    i + self.lookback,
                    kp_index
                ]
            )

            y_dst.append(
                scaled_data[
                    i + self.lookback,
                    dst_index
                ]
            )


        X = np.array(X)

        y_kp = np.array(y_kp)

        y_dst = np.array(y_dst)


        print(
            f"\nX shape: {X.shape}"
        )

        print(
            f"Kp10 target shape: {y_kp.shape}"
        )

        print(
            f"Dst target shape: {y_dst.shape}"
        )

        print(
            f"Number of input features: {X.shape[2]}"
        )

        print(
            f"Lookback: {self.lookback} hours"
        )


        # Combine targets

        y = np.column_stack(
            [y_kp, y_dst]
        )


        return X, y


    # --------------------------------------------------------
    # Chronological train-validation-test split
    # --------------------------------------------------------

    def split_data(
        self,
        X,
        y,
        test_size=0.20,
        val_size=0.15
    ):

        print("\n" + "=" * 75)
        print("CHRONOLOGICAL DATA SPLIT")
        print("=" * 75)


        n = len(X)

        test_start = int(
            n * (1 - test_size)
        )

        val_start = int(
            test_start *
            (1 - val_size)
        )


        X_train = X[:val_start]

        y_train = y[:val_start]


        X_val = X[val_start:test_start]

        y_val = y[val_start:test_start]


        X_test = X[test_start:]

        y_test = y[test_start:]


        print(
            f"Training samples:   {len(X_train):,}"
        )

        print(
            f"Validation samples: {len(X_val):,}"
        )

        print(
            f"Testing samples:    {len(X_test):,}"
        )


        return (
            X_train,
            X_val,
            X_test,
            y_train,
            y_val,
            y_test
        )


# ============================================================
# 6. MODEL FACTORY
# ============================================================

class ModelFactory:

    # --------------------------------------------------------
    # LSTM
    # --------------------------------------------------------

    @staticmethod
    def build_lstm(
        input_shape,
        units=64,
        dropout=0.2
    ):

        inputs = Input(
            shape=input_shape
        )

        x = LSTM(
            units,
            return_sequences=True
        )(inputs)

        x = Dropout(dropout)(x)

        x = LSTM(
            units // 2
        )(x)

        x = Dropout(dropout)(x)

        x = Dense(
            64,
            activation="relu"
        )(x)

        x = Dropout(dropout)(x)


        # Two outputs

        kp_output = Dense(
            1,
            name="Kp10"
        )(x)

        dst_output = Dense(
            1,
            name="Dst"
        )(x)


        model = Model(
            inputs=inputs,
            outputs=[
                kp_output,
                dst_output
            ]
        )

        return model


    # --------------------------------------------------------
    # Bidirectional LSTM
    # --------------------------------------------------------

    @staticmethod
    def build_bidirectional_lstm(
        input_shape,
        units=64,
        dropout=0.2
    ):

        inputs = Input(
            shape=input_shape
        )

        x = Bidirectional(
            LSTM(
                units,
                return_sequences=True
            )
        )(inputs)

        x = Dropout(dropout)(x)

        x = Bidirectional(
            LSTM(units // 2)
        )(x)

        x = Dropout(dropout)(x)

        x = Dense(
            64,
            activation="relu"
        )(x)

        x = Dropout(dropout)(x)


        kp_output = Dense(
            1,
            name="Kp10"
        )(x)

        dst_output = Dense(
            1,
            name="Dst"
        )(x)


        model = Model(
            inputs=inputs,
            outputs=[
                kp_output,
                dst_output
            ]
        )

        return model


    # --------------------------------------------------------
    # CNN-LSTM
    # --------------------------------------------------------

    @staticmethod
    def build_cnn_lstm(
        input_shape,
        filters=64,
        kernel_size=3,
        lstm_units=64,
        dropout=0.2
    ):

        inputs = Input(
            shape=input_shape
        )

        x = Conv1D(
            filters=filters,
            kernel_size=kernel_size,
            activation="relu",
            padding="same"
        )(inputs)

        x = MaxPooling1D(
            pool_size=2
        )(x)

        x = Dropout(dropout)(x)

        x = LSTM(
            lstm_units,
            return_sequences=True
        )(x)

        x = Dropout(dropout)(x)

        x = LSTM(
            lstm_units // 2
        )(x)

        x = Dropout(dropout)(x)

        x = Dense(
            64,
            activation="relu"
        )(x)

        x = Dropout(dropout)(x)


        kp_output = Dense(
            1,
            name="Kp10"
        )(x)

        dst_output = Dense(
            1,
            name="Dst"
        )(x)


        model = Model(
            inputs=inputs,
            outputs=[
                kp_output,
                dst_output
            ]
        )

        return model



# ============================================================
# 7. MODEL TRAINER
# ============================================================

class ModelTrainer:

    def __init__(
        self,
        model_name,
        model,
        scaler,
        all_columns
    ):

        self.model_name = model_name
        self.model = model
        self.scaler = scaler
        self.all_columns = all_columns

        self.history = None


    # --------------------------------------------------------
    # Compile
    # --------------------------------------------------------

    def compile_model(
        self,
        learning_rate=0.001
    ):

        self.model.compile(

            optimizer=Adam(
                learning_rate=learning_rate
            ),

            loss={
                "Kp10": "mse",
                "Dst": "mse"
            },

            metrics={
                "Kp10": ["mae"],
                "Dst": ["mae"]
            },

            loss_weights={
                "Kp10": 1.0,
                "Dst": 1.0
            }
        )


    # --------------------------------------------------------
    # Train
    # --------------------------------------------------------

    def train(
        self,
        X_train,
        y_train,
        X_val,
        y_val,
        epochs=100,
        batch_size=32
    ):

        model_path = os.path.join(
            OUTPUT_DIR,
            f"{self.model_name}_best.keras"
        )


        callbacks = [

            EarlyStopping(
                monitor="val_loss",
                patience=15,
                restore_best_weights=True,
                verbose=1
            ),

            ReduceLROnPlateau(
                monitor="val_loss",
                factor=0.5,
                patience=5,
                min_lr=1e-7,
                verbose=1
            ),

            ModelCheckpoint(
                model_path,
                monitor="val_loss",
                save_best_only=True,
                verbose=1
            )
        ]


        self.history = self.model.fit(

            X_train,

            {
                "Kp10": y_train[:, 0],
                "Dst": y_train[:, 1]
            },

            validation_data=(

                X_val,

                {
                    "Kp10": y_val[:, 0],
                    "Dst": y_val[:, 1]
                }
            ),

            epochs=epochs,

            batch_size=batch_size,

            callbacks=callbacks,

            verbose=1
        )


        return self.history


    # --------------------------------------------------------
    # Inverse transformation
    # --------------------------------------------------------

    def inverse_transform_targets(
        self,
        y_scaled
    ):

        kp_index = self.all_columns.index(
            "Kp10"
        )

        dst_index = self.all_columns.index(
            "Dst"
        )


        dummy = np.zeros(
            (
                len(y_scaled),
                len(self.all_columns)
            )
        )


        dummy[:, kp_index] = y_scaled[:, 0]

        dummy[:, dst_index] = y_scaled[:, 1]


        original = self.scaler.inverse_transform(
            dummy
        )


        kp = original[:, kp_index]

        dst = original[:, dst_index]


        return kp, dst


    # --------------------------------------------------------
    # Evaluation
    # --------------------------------------------------------

    def evaluate(
        self,
        X_test,
        y_test
    ):

        print("\n" + "=" * 75)

        print(
            f"EVALUATING {self.model_name}"
        )

        print("=" * 75)


        predictions = self.model.predict(
            X_test,
            verbose=0
        )


        # Kp10 prediction

        kp_pred_scaled = predictions[0].flatten()

        # Dst prediction

        dst_pred_scaled = predictions[1].flatten()


        y_pred_scaled = np.column_stack(
            [
                kp_pred_scaled,
                dst_pred_scaled
            ]
        )


        # Actual and predicted

        actual_kp, actual_dst = (
            self.inverse_transform_targets(
                y_test
            )
        )


        predicted_kp, predicted_dst = (
            self.inverse_transform_targets(
                y_pred_scaled
            )
        )


        # ----------------------------------------------------
        # Metrics
        # ----------------------------------------------------

        metrics = {}


        # Kp10

        metrics["Kp10"] = {

            "rmse":
                np.sqrt(
                    mean_squared_error(
                        actual_kp,
                        predicted_kp
                    )
                ),

            "mae":
                mean_absolute_error(
                    actual_kp,
                    predicted_kp
                ),

            "r2":
                r2_score(
                    actual_kp,
                    predicted_kp
                ),

            "correlation":
                pearsonr(
                    actual_kp,
                    predicted_kp
                )[0]
        }


        # Dst

        metrics["Dst"] = {

            "rmse":
                np.sqrt(
                    mean_squared_error(
                        actual_dst,
                        predicted_dst
                    )
                ),

            "mae":
                mean_absolute_error(
                    actual_dst,
                    predicted_dst
                ),

            "r2":
                r2_score(
                    actual_dst,
                    predicted_dst
                ),

            "correlation":
                pearsonr(
                    actual_dst,
                    predicted_dst
                )[0]
        }


        # Prediction efficiency

        metrics["Dst"]["pe"] = (
            1
            -
            np.sum(
                (
                    actual_dst
                    -
                    predicted_dst
                ) ** 2
            )
            /
            np.sum(
                actual_dst ** 2
            )
        )


        print("\nKp × 10 Performance:")

        print(
            f"RMSE: {metrics['Kp10']['rmse']:.4f}"
        )

        print(
            f"MAE: {metrics['Kp10']['mae']:.4f}"
        )

        print(
            f"R²: {metrics['Kp10']['r2']:.4f}"
        )

        print(
            f"Correlation: "
            f"{metrics['Kp10']['correlation']:.4f}"
        )


        print("\nDst Performance:")

        print(
            f"RMSE: {metrics['Dst']['rmse']:.4f} nT"
        )

        print(
            f"MAE: {metrics['Dst']['mae']:.4f} nT"
        )

        print(
            f"R²: {metrics['Dst']['r2']:.4f}"
        )

        print(
            f"Correlation: "
            f"{metrics['Dst']['correlation']:.4f}"
        )

        print(
            f"Prediction Efficiency: "
            f"{metrics['Dst']['pe']:.4f}"
        )


        return {

            "actual_kp": actual_kp,

            "predicted_kp": predicted_kp,

            "actual_dst": actual_dst,

            "predicted_dst": predicted_dst,

            "metrics": metrics
        }


# ============================================================
# 8. STORM CLASSIFICATION
# ============================================================

class StormClassifier:

    """
    Geomagnetic storm classification using both:

    1. Kp × 10
    2. Dst

    Kp storm scale:
        Kp < 5       -> Quiet
        Kp 5          -> G1 Minor
        Kp 6          -> G2 Moderate
        Kp 7          -> G3 Strong
        Kp 8          -> G4 Severe
        Kp 9          -> G5 Extreme
    """


    # --------------------------------------------------------
    # Kp classification
    # --------------------------------------------------------

    @staticmethod
    def classify_kp(kp10):

        categories = []


        for value in kp10:

            if value < 50:

                categories.append(
                    "Quiet"
                )

            elif value < 60:

                categories.append(
                    "G1 Minor"
                )

            elif value < 70:

                categories.append(
                    "G2 Moderate"
                )

            elif value < 80:

                categories.append(
                    "G3 Strong"
                )

            elif value < 90:

                categories.append(
                    "G4 Severe"
                )

            else:

                categories.append(
                    "G5 Extreme"
                )


        return np.array(
            categories
        )


    # --------------------------------------------------------
    # Dst classification
    # --------------------------------------------------------

    @staticmethod
    def classify_dst(dst):

        categories = []


        for value in dst:

            if value > -30:

                categories.append(
                    "Quiet"
                )

            elif value > -50:

                categories.append(
                    "Minor"
                )

            elif value > -100:

                categories.append(
                    "Moderate"
                )

            elif value > -200:

                categories.append(
                    "Intense"
                )

            else:

                categories.append(
                    "Super Storm"
                )


        return np.array(
            categories
        )


    # --------------------------------------------------------
    # Combined storm classification
    # --------------------------------------------------------

    @staticmethod
    def classify_combined(
        kp10,
        dst
    ):

        kp_classes = (
            StormClassifier.classify_kp(
                kp10
            )
        )

        dst_classes = (
            StormClassifier.classify_dst(
                dst
            )
        )


        combined = []


        for kp_class, dst_class in zip(
            kp_classes,
            dst_classes
        ):

            # ------------------------------------------------
            # Any storm indication from Kp or Dst
            # is retained.
            # ------------------------------------------------

            if (
                kp_class == "G5 Extreme"
                or
                dst_class == "Super Storm"
            ):

                combined.append(
                    "Extreme/Super Storm"
                )


            elif (
                kp_class == "G4 Severe"
                or
                dst_class == "Intense"
            ):

                combined.append(
                    "Severe/Intense"
                )


            elif (
                kp_class == "G3 Strong"
                or
                dst_class == "Moderate"
            ):

                combined.append(
                    "Strong/Moderate"
                )


            elif (
                kp_class == "G2 Moderate"
                or
                dst_class == "Minor"
            ):

                combined.append(
                    "Moderate/Minor"
                )


            elif (
                kp_class == "G1 Minor"
            ):

                combined.append(
                    "Minor"
                )


            else:

                combined.append(
                    "Quiet"
                )


        return np.array(
            combined
        )


    # --------------------------------------------------------
    # Binary storm classification
    # --------------------------------------------------------

    @staticmethod
    def binary_storm(
        kp10,
        dst
    ):

        """
        Storm = Kp >= 5 OR Dst <= -50 nT
        """

        return (
            (np.asarray(kp10) >= 50)
            |
            (np.asarray(dst) <= -50)
        ).astype(int)


    # --------------------------------------------------------
    # Classification evaluation
    # --------------------------------------------------------

    @staticmethod
    def evaluate(
        actual_kp,
        predicted_kp,
        actual_dst,
        predicted_dst,
        model_name
    ):

        print("\n" + "=" * 75)

        print(
            f"STORM CLASSIFICATION: {model_name}"
        )

        print("=" * 75)


        # ----------------------------------------------------
        # Kp classification
        # ----------------------------------------------------

        actual_kp_class = (
            StormClassifier.classify_kp(
                actual_kp
            )
        )

        predicted_kp_class = (
            StormClassifier.classify_kp(
                predicted_kp
            )
        )


        kp_accuracy = accuracy_score(
            actual_kp_class,
            predicted_kp_class
        )


        print(
            f"\nKp × 10 Classification Accuracy: "
            f"{kp_accuracy:.4f}"
        )


        kp_labels = [
            "Quiet",
            "G1 Minor",
            "G2 Moderate",
            "G3 Strong",
            "G4 Severe",
            "G5 Extreme"
        ]


        kp_cm = confusion_matrix(
            actual_kp_class,
            predicted_kp_class,
            labels=kp_labels
        )


        print(
            "\nKp × 10 Classification Report:"
        )

        print(
            classification_report(
                actual_kp_class,
                predicted_kp_class,
                labels=kp_labels,
                zero_division=0
            )
        )


        # ----------------------------------------------------
        # Dst classification
        # ----------------------------------------------------

        actual_dst_class = (
            StormClassifier.classify_dst(
                actual_dst
            )
        )

        predicted_dst_class = (
            StormClassifier.classify_dst(
                predicted_dst
            )
        )


        dst_accuracy = accuracy_score(
            actual_dst_class,
            predicted_dst_class
        )


        print(
            f"\nDst Classification Accuracy: "
            f"{dst_accuracy:.4f}"
        )


        dst_labels = [
            "Quiet",
            "Minor",
            "Moderate",
            "Intense",
            "Super Storm"
        ]


        dst_cm = confusion_matrix(
            actual_dst_class,
            predicted_dst_class,
            labels=dst_labels
        )


        print(
            "\nDst Classification Report:"
        )

        print(
            classification_report(
                actual_dst_class,
                predicted_dst_class,
                labels=dst_labels,
                zero_division=0
            )
        )


        # ----------------------------------------------------
        # Combined binary storm classification
        # ----------------------------------------------------

        actual_binary = (
            StormClassifier.binary_storm(
                actual_kp,
                actual_dst
            )
        )


        predicted_binary = (
            StormClassifier.binary_storm(
                predicted_kp,
                predicted_dst
            )
        )


        binary_accuracy = accuracy_score(
            actual_binary,
            predicted_binary
        )

        binary_precision = precision_score(
            actual_binary,
            predicted_binary,
            zero_division=0
        )

        binary_recall = recall_score(
            actual_binary,
            predicted_binary,
            zero_division=0
        )

        binary_f1 = f1_score(
            actual_binary,
            predicted_binary,
            zero_division=0
        )


        print(
            "\nCombined Kp × 10 + Dst Binary Classification:"
        )

        print(
            f"Accuracy:  {binary_accuracy:.4f}"
        )

        print(
            f"Precision: {binary_precision:.4f}"
        )

        print(
            f"Recall:    {binary_recall:.4f}"
        )

        print(
            f"F1 Score:  {binary_f1:.4f}"
        )


        binary_cm = confusion_matrix(
            actual_binary,
            predicted_binary
        )


        return {

            "actual_kp_class":
                actual_kp_class,

            "predicted_kp_class":
                predicted_kp_class,

            "kp_cm":
                kp_cm,

            "actual_dst_class":
                actual_dst_class,

            "predicted_dst_class":
                predicted_dst_class,

            "dst_cm":
                dst_cm,

            "actual_binary":
                actual_binary,

            "predicted_binary":
                predicted_binary,

            "binary_cm":
                binary_cm,

            "kp_accuracy":
                kp_accuracy,

            "dst_accuracy":
                dst_accuracy,

            "binary_accuracy":
                binary_accuracy,

            "binary_precision":
                binary_precision,

            "binary_recall":
                binary_recall,

            "binary_f1":
                binary_f1
        }


# ============================================================
# 9. VISUALIZATION
# ============================================================

class Visualizer:


    # --------------------------------------------------------
    # Training history
    # --------------------------------------------------------

    @staticmethod
    def plot_training_history(
        histories,
        model_names
    ):

        plt.figure(
            figsize=(14, 7)
        )


        for history, name in zip(
            histories,
            model_names
        ):

            plt.plot(
                history.history["loss"],
                label=f"{name} Train"
            )

            plt.plot(
                history.history["val_loss"],
                linestyle="--",
                label=f"{name} Validation"
            )


        plt.xlabel("Epoch")

        plt.ylabel("Total Loss")

        plt.title(
            "Training and Validation Loss"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(
            os.path.join(
                OUTPUT_DIR,
                "training_validation_loss.png"
            ),
            dpi=300
        )

        plt.show()


    # --------------------------------------------------------
    # Dst predictions
    # --------------------------------------------------------

    @staticmethod
    def plot_dst_predictions(
        predictions_dict,
        sample_size=2000
    ):

        for model_name, result in (
            predictions_dict.items()
        ):

            actual = (
                result["actual_dst"]
                [:sample_size]
            )

            predicted = (
                result["predicted_dst"]
                [:sample_size]
            )


            plt.figure(
                figsize=(15, 6)
            )


            plt.plot(
                actual,
                label="Actual Dst",
                linewidth=1
            )

            plt.plot(
                predicted,
                label="Predicted Dst",
                linewidth=1
            )


            plt.axhline(
                -30,
                linestyle="--",
                label="Minor: -30 nT"
            )

            plt.axhline(
                -50,
                linestyle="--",
                label="Moderate: -50 nT"
            )

            plt.axhline(
                -100,
                linestyle="--",
                label="Intense: -100 nT"
            )

            plt.axhline(
                -200,
                linestyle="--",
                label="Super Storm: -200 nT"
            )


            plt.xlabel(
                "Time Step"
            )

            plt.ylabel(
                "Dst (nT)"
            )

            plt.title(
                f"{model_name} - Dst Prediction"
            )

            plt.legend()

            plt.grid(
                True,
                alpha=0.3
            )

            plt.tight_layout()


            plt.savefig(
                os.path.join(
                    OUTPUT_DIR,
                    f"{model_name}_Dst_prediction.png"
                ),
                dpi=300
            )

            plt.show()


    # --------------------------------------------------------
    # Kp10 predictions
    # --------------------------------------------------------

    @staticmethod
    def plot_kp_predictions(
        predictions_dict,
        sample_size=2000
    ):

        for model_name, result in (
            predictions_dict.items()
        ):

            actual = (
                result["actual_kp"]
                [:sample_size]
            )

            predicted = (
                result["predicted_kp"]
                [:sample_size]
            )


            plt.figure(
                figsize=(15, 6)
            )


            plt.plot(
                actual,
                label="Actual Kp × 10",
                linewidth=1
            )

            plt.plot(
                predicted,
                label="Predicted Kp × 10",
                linewidth=1
            )


            # Kp storm thresholds

            plt.axhline(
                50,
                linestyle="--",
                label="Kp = 5 (G1)"
            )

            plt.axhline(
                60,
                linestyle="--",
                label="Kp = 6 (G2)"
            )

            plt.axhline(
                70,
                linestyle="--",
                label="Kp = 7 (G3)"
            )

            plt.axhline(
                80,
                linestyle="--",
                label="Kp = 8 (G4)"
            )

            plt.axhline(
                90,
                linestyle="--",
                label="Kp = 9 (G5)"
            )


            plt.xlabel(
                "Time Step"
            )

            plt.ylabel(
                "Kp × 10"
            )

            plt.title(
                f"{model_name} - Kp × 10 Prediction"
            )

            plt.legend()

            plt.grid(
                True,
                alpha=0.3
            )

            plt.tight_layout()


            plt.savefig(
                os.path.join(
                    OUTPUT_DIR,
                    f"{model_name}_Kp10_prediction.png"
                ),
                dpi=300
            )

            plt.show()


    # --------------------------------------------------------
    # Scatter plots
    # --------------------------------------------------------

    @staticmethod
    def plot_scatter(
        predictions_dict
    ):

        for model_name, result in (
            predictions_dict.items()
        ):

            # Dst

            plt.figure(
                figsize=(7, 7)
            )

            plt.scatter(
                result["actual_dst"],
                result["predicted_dst"],
                alpha=0.4,
                s=10
            )

            minimum = min(
                result["actual_dst"].min(),
                result["predicted_dst"].min()
            )

            maximum = max(
                result["actual_dst"].max(),
                result["predicted_dst"].max()
            )

            plt.plot(
                [minimum, maximum],
                [minimum, maximum],
                linestyle="--",
                linewidth=2
            )

            plt.xlabel(
                "Actual Dst (nT)"
            )

            plt.ylabel(
                "Predicted Dst (nT)"
            )

            plt.title(
                f"{model_name} - Dst\n"
                f"R² = "
                f"{result['metrics']['Dst']['r2']:.4f}"
            )

            plt.grid(
                True,
                alpha=0.3
            )

            plt.tight_layout()

            plt.savefig(
                os.path.join(
                    OUTPUT_DIR,
                    f"{model_name}_Dst_scatter.png"
                ),
                dpi=300
            )

            plt.show()


            # Kp10

            plt.figure(
                figsize=(7, 7)
            )

            plt.scatter(
                result["actual_kp"],
                result["predicted_kp"],
                alpha=0.4,
                s=10
            )

            minimum = min(
                result["actual_kp"].min(),
                result["predicted_kp"].min()
            )

            maximum = max(
                result["actual_kp"].max(),
                result["predicted_kp"].max()
            )

            plt.plot(
                [minimum, maximum],
                [minimum, maximum],
                linestyle="--",
                linewidth=2
            )

            plt.xlabel(
                "Actual Kp × 10"
            )

            plt.ylabel(
                "Predicted Kp × 10"
            )

            plt.title(
                f"{model_name} - Kp × 10\n"
                f"R² = "
                f"{result['metrics']['Kp10']['r2']:.4f}"
            )

            plt.grid(
                True,
                alpha=0.3
            )

            plt.tight_layout()

            plt.savefig(
                os.path.join(
                    OUTPUT_DIR,
                    f"{model_name}_Kp10_scatter.png"
                ),
                dpi=300
            )

            plt.show()


    # --------------------------------------------------------
    # Confusion matrix
    # --------------------------------------------------------

    @staticmethod
    def plot_confusion_matrix(
        cm,
        labels,
        title,
        filename
    ):

        plt.figure(
            figsize=(10, 8)
        )

        sns.heatmap(
            cm,
            annot=True,
            fmt="d",
            cmap="Blues",
            xticklabels=labels,
            yticklabels=labels
        )

        plt.xlabel(
            "Predicted"
        )

        plt.ylabel(
            "Actual"
        )

        plt.title(
            title
        )

        plt.tight_layout()


        plt.savefig(
            os.path.join(
                OUTPUT_DIR,
                filename
            ),
            dpi=300
        )

        plt.show()


    # --------------------------------------------------------
    # Residual analysis
    # --------------------------------------------------------

    @staticmethod
    def plot_residuals(
        predictions_dict
    ):

        for model_name, result in (
            predictions_dict.items()
        ):

            # Dst residuals

            residuals = (
                result["actual_dst"]
                -
                result["predicted_dst"]
            )


            plt.figure(
                figsize=(10, 6)
            )

            plt.scatter(
                result["predicted_dst"],
                residuals,
                alpha=0.4,
                s=10
            )

            plt.axhline(
                0,
                linestyle="--",
                linewidth=2
            )

            plt.xlabel(
                "Predicted Dst (nT)"
            )

            plt.ylabel(
                "Residual (nT)"
            )

            plt.title(
                f"{model_name} - Dst Residuals"
            )

            plt.grid(
                True,
                alpha=0.3
            )

            plt.tight_layout()


            plt.savefig(
                os.path.join(
                    OUTPUT_DIR,
                    f"{model_name}_Dst_residuals.png"
                ),
                dpi=300
            )

            plt.show()


    # --------------------------------------------------------
    # Model comparison
    # --------------------------------------------------------

    @staticmethod
    def create_comparison_table(
        predictions_dict
    ):

        rows = []


        for model_name, result in (
            predictions_dict.items()
        ):

            kp = result["metrics"]["Kp10"]

            dst = result["metrics"]["Dst"]


            rows.append({

                "Model":
                    model_name,

                "Kp10_RMSE":
                    kp["rmse"],

                "Kp10_MAE":
                    kp["mae"],

                "Kp10_R2":
                    kp["r2"],

                "Kp10_Correlation":
                    kp["correlation"],

                "Dst_RMSE_nT":
                    dst["rmse"],

                "Dst_MAE_nT":
                    dst["mae"],

                "Dst_R2":
                    dst["r2"],

                "Dst_Correlation":
                    dst["correlation"],

                "Dst_PE":
                    dst["pe"]
            })


        df_results = pd.DataFrame(
            rows
        )


        print("\n" + "=" * 100)

        print(
            "MODEL COMPARISON"
        )

        print("=" * 100)

        print(
            df_results.to_string(
                index=False
            )
        )


        df_results.to_csv(
            os.path.join(
                OUTPUT_DIR,
                "model_comparison.csv"
            ),
            index=False
        )


        return df_results


# ============================================================
# 10. ROC AND PRECISION-RECALL CURVES
# ============================================================

def plot_roc_pr_curves(
    predictions_dict
):

    # --------------------------------------------------------
    # ROC
    # --------------------------------------------------------

    plt.figure(
        figsize=(9, 7)
    )


    for model_name, result in (
        predictions_dict.items()
    ):

        actual = result[
            "actual_binary"
        ]

        predicted = result[
            "predicted_binary_score"
        ]


        fpr, tpr, _ = roc_curve(
            actual,
            predicted
        )

        roc_auc = auc(
            fpr,
            tpr
        )


        plt.plot(
            fpr,
            tpr,
            label=f"{model_name} AUC={roc_auc:.4f}"
        )


    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--"
    )


    plt.xlabel(
        "False Positive Rate"
    )

    plt.ylabel(
        "True Positive Rate"
    )

    plt.title(
        "ROC Curve - Geomagnetic Storm Classification"
    )

    plt.legend()

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()


    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "ROC_curve.png"
        ),
        dpi=300
    )

    plt.show()


    # --------------------------------------------------------
    # Precision-Recall
    # --------------------------------------------------------

    plt.figure(
        figsize=(9, 7)
    )


    for model_name, result in (
        predictions_dict.items()
    ):

        actual = result[
            "actual_binary"
        ]

        predicted = result[
            "predicted_binary_score"
        ]


        precision, recall, _ = (
            precision_recall_curve(
                actual,
                predicted
            )
        )


        ap = average_precision_score(
            actual,
            predicted
        )


        plt.plot(
            recall,
            precision,
            label=f"{model_name} AP={ap:.4f}"
        )


    plt.xlabel(
        "Recall"
    )

    plt.ylabel(
        "Precision"
    )

    plt.title(
        "Precision-Recall Curve - Geomagnetic Storm"
    )

    plt.legend()

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()


    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "PR_curve.png"
        ),
        dpi=300
    )

    plt.show()


# ============================================================
# 11. ADD BINARY CLASSIFICATION INFORMATION
# ============================================================

def add_binary_classification(
    predictions_dict
):

    for model_name, result in (
        predictions_dict.items()
    ):

        actual_binary = (
            StormClassifier.binary_storm(
                result["actual_kp"],
                result["actual_dst"]
            )
        )


        # ----------------------------------------------------
        # Create continuous storm score
        #
        # Kp contribution:
        #   Kp10 / 50
        #
        # Dst contribution:
        #   -Dst / 50
        #
        # This is used only for ROC/PR visualization.
        # ----------------------------------------------------

        kp_score = (
            result["predicted_kp"]
            / 50.0
        )

        dst_score = (
            -result["predicted_dst"]
            / 50.0
        )


        storm_score = np.maximum(
            kp_score,
            dst_score
        )


        result[
            "actual_binary"
        ] = actual_binary


        result[
            "predicted_binary_score"
        ] = storm_score


# ============================================================
# 12. SAVE ALL PREDICTIONS
# ============================================================

def save_prediction_results(
    predictions_dict
):

    for model_name, result in (
        predictions_dict.items()
    ):

        output_df = pd.DataFrame({

            "Actual_Kp10":
                result["actual_kp"],

            "Predicted_Kp10":
                result["predicted_kp"],

            "Actual_Dst":
                result["actual_dst"],

            "Predicted_Dst":
                result["predicted_dst"]
        })


        # Kp classification

        output_df[
            "Actual_Kp_Class"
        ] = StormClassifier.classify_kp(
            result["actual_kp"]
        )


        output_df[
            "Predicted_Kp_Class"
        ] = StormClassifier.classify_kp(
            result["predicted_kp"]
        )


        # Dst classification

        output_df[
            "Actual_Dst_Class"
        ] = StormClassifier.classify_dst(
            result["actual_dst"]
        )


        output_df[
            "Predicted_Dst_Class"
        ] = StormClassifier.classify_dst(
            result["predicted_dst"]
        )


        # Combined classification

        output_df[
            "Actual_Combined_Class"
        ] = StormClassifier.classify_combined(
            result["actual_kp"],
            result["actual_dst"]
        )


        output_df[
            "Predicted_Combined_Class"
        ] = StormClassifier.classify_combined(
            result["predicted_kp"],
            result["predicted_dst"]
        )


        # Binary storm

        output_df[
            "Actual_Storm"
        ] = StormClassifier.binary_storm(
            result["actual_kp"],
            result["actual_dst"]
        )


        output_df[
            "Predicted_Storm"
        ] = StormClassifier.binary_storm(
            result["predicted_kp"],
            result["predicted_dst"]
        )


        filename = os.path.join(
            OUTPUT_DIR,
            f"{model_name}_predictions.csv"
        )


        output_df.to_csv(
            filename,
            index=False
        )


        print(
            f"Saved: {filename}"
        )


# ============================================================
# 13. MAIN FUNCTION
# ============================================================

def main():

    print("\n")

    print("=" * 75)

    print(
        "GEOMAGNETIC STORM PREDICTION"
    )

    print(
        "Kp × 10 + Dst"
    )

    print("=" * 75)


    # ========================================================
    # STEP 1: LOAD AND CLEAN DATA
    # ========================================================

    preprocessor = (
        DataPreprocessor(
            DATA_PATH
        )
    )


    preprocessor.load_data()

    preprocessor.check_columns()

    preprocessor.handle_missing_values()

    preprocessor.clean_data()


    df = (
        preprocessor.get_processed_data()
    )


    # Save cleaned data

    cleaned_path = os.path.join(
        OUTPUT_DIR,
        "cleaned_geomagnetic_data.csv"
    )


    df.to_csv(
        cleaned_path,
        index=False
    )


    print(
        f"\nCleaned data saved to:\n"
        f"{cleaned_path}"
    )


    # ========================================================
    # STEP 2: CREATE SEQUENCES
    # ========================================================

    seq_gen = SequenceGenerator(
        df,
        lookback=LOOKBACK
    )


    X, y = (
        seq_gen.create_sequences()
    )


    (
        X_train,
        X_val,
        X_test,
        y_train,
        y_val,
        y_test
    ) = seq_gen.split_data(
        X,
        y,
        test_size=TEST_SIZE,
        val_size=VALIDATION_SIZE
    )


    # ========================================================
    # STEP 3: MODEL INPUT SHAPE
    # ========================================================

    input_shape = (
        X_train.shape[1],
        X_train.shape[2]
    )


    print(
        "\nModel input shape:",
        input_shape
    )


    # ========================================================
    # STEP 4: BUILD MODELS
    # ========================================================

    models_config = {

        "LSTM":
            ModelFactory.build_lstm(
                input_shape
            ),

        "BiLSTM":
            ModelFactory.build_bidirectional_lstm(
                input_shape
            ),

        "CNN-LSTM":
            ModelFactory.build_cnn_lstm(
                input_shape
            )
    }


    # ========================================================
    # STEP 5: TRAIN AND EVALUATE
    # ========================================================

    histories = []

    predictions_dict = {}


    for model_name, model in (
        models_config.items()
    ):

        print("\n")

        print("=" * 75)

        print(
            f"TRAINING {model_name}"
        )

        print("=" * 75)


        trainer = ModelTrainer(

            model_name,

            model,

            seq_gen.scaler,

            seq_gen.all_columns
        )


        trainer.compile_model(
            learning_rate=LEARNING_RATE
        )


        print(
            f"\n{model_name} architecture:"
        )

        model.summary()


        history = trainer.train(

            X_train,

            y_train,

            X_val,

            y_val,

            epochs=EPOCHS,

            batch_size=BATCH_SIZE
        )


        result = trainer.evaluate(

            X_test,

            y_test
        )


        histories.append(
            history
        )


        predictions_dict[
            model_name
        ] = result


    # ========================================================
    # STEP 6: CLASSIFICATION
    # ========================================================

    classification_results = {}


    for model_name, result in (
        predictions_dict.items()
    ):

        classification = (
            StormClassifier.evaluate(

                result["actual_kp"],

                result["predicted_kp"],

                result["actual_dst"],

                result["predicted_dst"],

                model_name
            )
        )


        classification_results[
            model_name
        ] = classification


    # ========================================================
    # STEP 7: ADD BINARY STORM SCORES
    # ========================================================

    add_binary_classification(
        predictions_dict
    )


    # ========================================================
    # STEP 8: VISUALIZATIONS
    # ========================================================

    print("\n")

    print("=" * 75)

    print(
        "GENERATING VISUALIZATIONS"
    )

    print("=" * 75)


    Visualizer.plot_training_history(
        histories,
        list(models_config.keys())
    )


    Visualizer.plot_dst_predictions(
        predictions_dict
    )


    Visualizer.plot_kp_predictions(
        predictions_dict
    )


    Visualizer.plot_scatter(
        predictions_dict
    )


    Visualizer.plot_residuals(
        predictions_dict
    )


    # ========================================================
    # STEP 9: CONFUSION MATRICES
    # ========================================================

    for model_name, result in (
        classification_results.items()
    ):

        Visualizer.plot_confusion_matrix(

            result["kp_cm"],

            [
                "Quiet",
                "G1 Minor",
                "G2 Moderate",
                "G3 Strong",
                "G4 Severe",
                "G5 Extreme"
            ],

            f"{model_name} - Kp × 10 Storm Classification",

            f"{model_name}_Kp10_confusion_matrix.png"
        )


        Visualizer.plot_confusion_matrix(

            result["dst_cm"],

            [
                "Quiet",
                "Minor",
                "Moderate",
                "Intense",
                "Super Storm"
            ],

            f"{model_name} - Dst Storm Classification",

            f"{model_name}_Dst_confusion_matrix.png"
        )


        Visualizer.plot_confusion_matrix(

            result["binary_cm"],

            [
                "No Storm",
                "Storm"
            ],

            f"{model_name} - Combined Kp × 10 + Dst",

            f"{model_name}_Combined_confusion_matrix.png"
        )


    # ========================================================
    # STEP 10: ROC AND PR
    # ========================================================

    plot_roc_pr_curves(
        predictions_dict
    )


    # ========================================================
    # STEP 11: MODEL COMPARISON
    # ========================================================

    comparison_df = (
        Visualizer.create_comparison_table(
            predictions_dict
        )
    )


    # ========================================================
    # STEP 12: CLASSIFICATION COMPARISON
    # ========================================================

    classification_table = []


    for model_name, result in (
        classification_results.items()
    ):

        classification_table.append({

            "Model":
                model_name,

            "Kp10_Accuracy":
                result["kp_accuracy"],

            "Dst_Accuracy":
                result["dst_accuracy"],

            "Combined_Accuracy":
                result["binary_accuracy"],

            "Combined_Precision":
                result["binary_precision"],

            "Combined_Recall":
                result["binary_recall"],

            "Combined_F1":
                result["binary_f1"]
        })


    classification_df = pd.DataFrame(
        classification_table
    )


    print("\n")

    print("=" * 100)

    print(
        "CLASSIFICATION MODEL COMPARISON"
    )

    print("=" * 100)

    print(
        classification_df.to_string(
            index=False
        )
    )


    classification_df.to_csv(

        os.path.join(
            OUTPUT_DIR,
            "classification_comparison.csv"
        ),

        index=False
    )


    # ========================================================
    # STEP 13: SAVE PREDICTIONS
    # ========================================================

    save_prediction_results(
        predictions_dict
    )


    # ========================================================
    # STEP 14: DETERMINE BEST MODELS
    # ========================================================

    best_dst_model = (
        comparison_df.loc[
            comparison_df["Dst_RMSE_nT"].idxmin(),
            "Model"
        ]
    )


    best_kp_model = (
        comparison_df.loc[
            comparison_df["Kp10_RMSE"].idxmin(),
            "Model"
        ]
    )


    best_classification_model = (
        classification_df.loc[
            classification_df[
                "Combined_F1"
            ].idxmax(),
            "Model"
        ]
    )


    print("\n")

    print("=" * 75)

    print(
        "FINAL RESULTS"
    )

    print("=" * 75)


    print(
        f"\nBest Dst model: "
        f"{best_dst_model}"
    )


    print(
        f"Best Kp × 10 model: "
        f"{best_kp_model}"
    )


    print(
        f"Best combined storm classification model: "
        f"{best_classification_model}"
    )


    print(
        f"\nResults directory:"
        f"\n{OUTPUT_DIR}"
    )


    return (
        predictions_dict,
        comparison_df,
        classification_df
    )


# ============================================================
# 14. RUN
# ============================================================

if __name__ == "__main__":

    (
        predictions_dict,
        comparison_df,
        classification_df
    ) = main()
```
