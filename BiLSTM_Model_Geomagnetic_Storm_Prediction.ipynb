```python
# ============================================================
# GEOMAGNETIC STORM PREDICTION USING BiLSTM
# ============================================================
# Model:
#   BiLSTM only
#
# Targets:
#   1. Kp × 10
#   2. Dst
#
# Input features:
#   YEAR, DOY, HR, B, Bx, By, Bz, PD, FP, E
#
# Data split:
#   Train      = 70%
#   Validation = 15%
#   Test       = 15%
#
# Lookback:
#   48 hours
#
# Optimizer:
#   Adam
#
# EarlyStopping:
#   patience = 5
#
# Author: Prajwal Bhandari
# Date: September 2026
# ============================================================


# ============================================================
# 0. INSTALL LIBRARIES - GOOGLE COLAB
# ============================================================

# Uncomment if required:
# !pip install -q tensorflow scikit-learn scipy seaborn


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import re
import random
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr

from sklearn.preprocessing import MinMaxScaler

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_curve,
    auc,
    precision_recall_curve,
    average_precision_score
)

import tensorflow as tf

from tensorflow.keras.models import Model

from tensorflow.keras.layers import (
    Input,
    LSTM,
    Dense,
    Dropout,
    Bidirectional
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

from tensorflow.keras.optimizers import Adam


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)


# ============================================================
# 3. CONFIGURATION
# ============================================================

# ------------------------------------------------------------
# DATA PATH
# ------------------------------------------------------------

DATA_PATH = "/content/Sun.csv"

# If using Google Drive:
# DATA_PATH = "/content/drive/MyDrive/D.csv"


# ------------------------------------------------------------
# OUTPUT DIRECTORY
# ------------------------------------------------------------

OUTPUT_DIR = "/content/geomagnetic_BiLSTM_results"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ------------------------------------------------------------
# TIME-SERIES SETTINGS
# ------------------------------------------------------------

LOOKBACK = 48

TRAIN_SIZE = 0.70
VALIDATION_SIZE = 0.15
TEST_SIZE = 0.15

EPOCHS = 90

BATCH_SIZE = 32

LEARNING_RATE = 0.001

EARLY_STOPPING_PATIENCE = 5

LR_PATIENCE = 3


# ============================================================
# 4. DATA PREPROCESSOR
# ============================================================

class DataPreprocessor:

    def __init__(self, filepath):

        self.filepath = filepath

        self.df = None

        self.missing_values_map = {

            "B": [
                999.99,
                9999.99
            ],

            "Bx": [
                999.99,
                9999.99
            ],

            "By": [
                999.99,
                9999.99
            ],

            "Bz": [
                999.99,
                9999.99
            ],

            "PD": [
                999.9,
                999.99,
                9999.9
            ],

            "FP": [
                99.99,
                999.99
            ],

            "E": [
                999.99,
                9999.99
            ],

            "Kp": [
                9.99,
                99.99
            ],

            "Dst": [
                9999.9,
                9999.99
            ]
        }


    # ========================================================
    # NORMALIZE COLUMN NAMES
    # ========================================================

    def normalize_column_names(self):

        new_columns = []

        for col in self.df.columns:

            col = str(col).strip()

            col_clean = re.sub(
                r"\s+",
                " ",
                col
            )

            lower_col = col_clean.lower()

            if "kp" in lower_col:

                col_clean = "Kp"

            elif lower_col == "dst":

                col_clean = "Dst"

            elif lower_col == "year":

                col_clean = "YEAR"

            elif lower_col == "doy":

                col_clean = "DOY"

            elif lower_col == "hr":

                col_clean = "HR"

            new_columns.append(
                col_clean
            )

        self.df.columns = new_columns

        return self


    # ========================================================
    # LOAD DATA
    # ========================================================

    def load_data(self):

        print("=" * 80)
        print("LOADING GEOMAGNETIC DATA")
        print("=" * 80)

        self.df = pd.read_csv(
            self.filepath
        )

        print(
            "\nOriginal dataset shape:",
            self.df.shape
        )

        print(
            "\nOriginal columns:"
        )

        print(
            list(self.df.columns)
        )

        self.normalize_column_names()

        print(
            "\nNormalized columns:"
        )

        print(
            list(self.df.columns)
        )

        return self


    # ========================================================
    # CHECK REQUIRED COLUMNS
    # ========================================================

    def check_columns(self):

        required_columns = [

            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E",
            "Kp",
            "Dst"
        ]

        missing = [

            col
            for col in required_columns
            if col not in self.df.columns
        ]

        if len(missing) > 0:

            raise ValueError(

                "\nMissing required columns:\n"
                + str(missing)
                + "\n\nAvailable columns:\n"
                + str(
                    list(
                        self.df.columns
                    )
                )
            )

        print(
            "\nAll required columns are available."
        )

        return self


    # ========================================================
    # HANDLE MISSING VALUES
    # ========================================================

    def handle_missing_values(self):

        print("\n" + "=" * 80)
        print("HANDLING MISSING VALUES")
        print("=" * 80)

        for col, values in (
            self.missing_values_map.items()
        ):

            if col in self.df.columns:

                for value in values:

                    count = (
                        self.df[col] == value
                    ).sum()

                    if count > 0:

                        self.df[col] = (
                            self.df[col]
                            .replace(
                                value,
                                np.nan
                            )
                        )

                        print(
                            f"{col}: replaced "
                            f"{count} values of {value}"
                        )

        return self


    # ========================================================
    # CLEAN DATA
    # ========================================================

    def clean_data(self):

        print("\n" + "=" * 80)
        print("CLEANING DATA")
        print("=" * 80)

        # ----------------------------------------------------
        # Convert columns to numeric
        # ----------------------------------------------------

        for col in self.df.columns:

            self.df[col] = pd.to_numeric(
                self.df[col],
                errors="coerce"
            )


        # ----------------------------------------------------
        # Sort chronologically
        # ----------------------------------------------------

        self.df = self.df.sort_values(
            [
                "YEAR",
                "DOY",
                "HR"
            ]
        ).reset_index(
            drop=True
        )


        # ----------------------------------------------------
        # Missing values
        # ----------------------------------------------------

        print(
            "\nMissing values before interpolation:"
        )

        missing = self.df.isna().sum()

        print(
            missing[
                missing > 0
            ]
        )


        # ----------------------------------------------------
        # Interpolation
        # ----------------------------------------------------

        numeric_columns = (
            self.df.select_dtypes(
                include=[np.number]
            ).columns
        )

        self.df[
            numeric_columns
        ] = (
            self.df[
                numeric_columns
            ]
            .interpolate(
                method="linear"
            )
            .ffill()
            .bfill()
        )


        # ----------------------------------------------------
        # Drop remaining NaN
        # ----------------------------------------------------

        before = len(
            self.df
        )

        self.df = self.df.dropna()

        after = len(
            self.df
        )

        print(
            f"\nRows removed: "
            f"{before - after}"
        )

        print(
            f"Final dataset shape: "
            f"{self.df.shape}"
        )


        # ====================================================
        # CREATE Kp × 10
        # ====================================================

        kp_max = self.df[
            "Kp"
        ].max()

        if kp_max <= 9.0:

            print(
                "\nKp is on standard 0–9 scale."
            )

            print(
                "Creating Kp10 = Kp × 10."
            )

            self.df[
                "Kp10"
            ] = (
                self.df["Kp"]
                * 10.0
            )

        else:

            print(
                "\nKp appears to already be Kp × 10."
            )

            self.df[
                "Kp10"
            ] = self.df["Kp"]


        # ----------------------------------------------------
        # Remove original Kp
        # ----------------------------------------------------

        self.df.drop(
            columns=["Kp"],
            inplace=True
        )


        # ----------------------------------------------------
        # Final columns
        # ----------------------------------------------------

        final_columns = [

            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E",
            "Kp10",
            "Dst"
        ]

        self.df = self.df[
            final_columns
        ]


        print(
            "\nFinal columns:"
        )

        print(
            list(
                self.df.columns
            )
        )


        print(
            "\nKp × 10 statistics:"
        )

        print(
            self.df[
                "Kp10"
            ].describe()
        )


        print(
            "\nDst statistics:"
        )

        print(
            self.df[
                "Dst"
            ].describe()
        )


        return self


    def get_processed_data(self):

        return self.df


# ============================================================
# 5. SEQUENCE GENERATOR
# ============================================================

class SequenceGenerator:

    def __init__(
        self,
        data,
        lookback=48
    ):

        self.data = data

        self.lookback = lookback

        self.scaler = MinMaxScaler()


        # ----------------------------------------------------
        # INPUT FEATURES
        # ----------------------------------------------------

        self.feature_columns = [

            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E"
        ]


        # ----------------------------------------------------
        # TARGETS
        # ----------------------------------------------------

        self.target_columns = [

            "Kp10",
            "Dst"
        ]


        self.all_columns = (
            self.feature_columns
            +
            self.target_columns
        )


    # ========================================================
    # CREATE SEQUENCES
    # ========================================================

    def create_sequences(self):

        print("\n" + "=" * 80)
        print("CREATING TIME-SERIES SEQUENCES")
        print("=" * 80)

        data_values = self.data[
            self.all_columns
        ].values


        # ----------------------------------------------------
        # Scale all variables
        # ----------------------------------------------------

        scaled_data = (
            self.scaler.fit_transform(
                data_values
            )
        )


        kp_index = (
            self.all_columns.index(
                "Kp10"
            )
        )

        dst_index = (
            self.all_columns.index(
                "Dst"
            )
        )


        X = []

        y_kp = []

        y_dst = []


        # ----------------------------------------------------
        # Sliding window
        # ----------------------------------------------------

        for i in range(
            len(scaled_data)
            -
            self.lookback
        ):

            X.append(

                scaled_data[
                    i:
                    i + self.lookback
                ]
            )

            y_kp.append(

                scaled_data[
                    i + self.lookback,
                    kp_index
                ]
            )

            y_dst.append(

                scaled_data[
                    i + self.lookback,
                    dst_index
                ]
            )


        X = np.array(
            X
        )

        y_kp = np.array(
            y_kp
        )

        y_dst = np.array(
            y_dst
        )


        y = np.column_stack(
            [
                y_kp,
                y_dst
            ]
        )


        print(
            "\nX shape:",
            X.shape
        )

        print(
            "y shape:",
            y.shape
        )

        print(
            "Lookback:",
            self.lookback,
            "hours"
        )

        print(
            "Number of input features:",
            X.shape[2]
        )

        print(
            "Number of target variables:",
            y.shape[1]
        )


        return X, y


    # ========================================================
    # 70 / 15 / 15 CHRONOLOGICAL SPLIT
    # ========================================================

    def split_data(
        self,
        X,
        y,
        train_size=0.70,
        validation_size=0.15,
        test_size=0.15
    ):

        print("\n" + "=" * 80)
        print(
            "70% TRAIN / 15% VALIDATION / 15% TEST"
        )
        print("=" * 80)


        if not np.isclose(
            train_size
            +
            validation_size
            +
            test_size,
            1.0
        ):

            raise ValueError(
                "Train + Validation + Test "
                "must equal 1.0"
            )


        n = len(X)


        train_end = int(
            n * train_size
        )

        val_end = int(
            n *
            (
                train_size
                +
                validation_size
            )
        )


        X_train = X[
            :train_end
        ]

        y_train = y[
            :train_end
        ]


        X_val = X[
            train_end:val_end
        ]

        y_val = y[
            train_end:val_end
        ]


        X_test = X[
            val_end:
        ]

        y_test = y[
            val_end:
        ]


        print(
            f"Training samples:   "
            f"{len(X_train):,}"
        )

        print(
            f"Validation samples: "
            f"{len(X_val):,}"
        )

        print(
            f"Testing samples:    "
            f"{len(X_test):,}"
        )


        print(
            "\nTraining percentage:",
            len(X_train) / n * 100
        )

        print(
            "Validation percentage:",
            len(X_val) / n * 100
        )

        print(
            "Testing percentage:",
            len(X_test) / n * 100
        )


        return (
            X_train,
            X_val,
            X_test,
            y_train,
            y_val,
            y_test
        )


# ============================================================
# 6. BiLSTM MODEL
# ============================================================

def build_bilstm(
    input_shape,
    units=64,
    dropout=0.20
):

    inputs = Input(
        shape=input_shape,
        name="Input"
    )


    # --------------------------------------------------------
    # BiLSTM Layer 1
    # --------------------------------------------------------

    x = Bidirectional(

        LSTM(
            units,
            return_sequences=True
        ),

        name="BiLSTM_1"

    )(inputs)


    x = Dropout(
        dropout,
        name="Dropout_1"
    )(x)


    # --------------------------------------------------------
    # BiLSTM Layer 2
    # --------------------------------------------------------

    x = Bidirectional(

        LSTM(
            units // 2,
            return_sequences=False
        ),

        name="BiLSTM_2"

    )(x)


    x = Dropout(
        dropout,
        name="Dropout_2"
    )(x)


    # --------------------------------------------------------
    # Dense layer
    # --------------------------------------------------------

    x = Dense(
        64,
        activation="relu",
        name="Dense_64"
    )(x)


    x = Dropout(
        dropout,
        name="Dropout_3"
    )(x)


    # ========================================================
    # OUTPUT 1: Kp × 10
    # ========================================================

    kp_output = Dense(
        1,
        name="Kp10"
    )(x)


    # ========================================================
    # OUTPUT 2: Dst
    # ========================================================

    dst_output = Dense(
        1,
        name="Dst"
    )(x)


    # --------------------------------------------------------
    # Build model
    # --------------------------------------------------------

    model = Model(

        inputs=inputs,

        outputs=[
            kp_output,
            dst_output
        ],

        name="BiLSTM_Geomagnetic_Storm_Model"
    )


    return model


# ============================================================
# 7. MODEL TRAINER
# ============================================================

class BiLSTMTrainer:

    def __init__(
        self,
        model,
        scaler,
        all_columns
    ):

        self.model = model

        self.scaler = scaler

        self.all_columns = all_columns

        self.history = None


    # ========================================================
    # COMPILE MODEL
    # ========================================================

    def compile_model(
        self,
        learning_rate=0.001
    ):

        self.model.compile(

            optimizer=Adam(
                learning_rate=
                learning_rate
            ),

            loss={

                "Kp10": "mse",

                "Dst": "mse"
            },

            metrics={

                "Kp10": [
                    "mae"
                ],

                "Dst": [
                    "mae"
                ]
            },

            loss_weights={

                "Kp10": 1.0,

                "Dst": 1.0
            }
        )


    # ========================================================
    # TRAIN MODEL
    # ========================================================

    def train(
        self,
        X_train,
        y_train,
        X_val,
        y_val,
        epochs=50,
        batch_size=32
    ):

        model_path = os.path.join(

            OUTPUT_DIR,

            "BiLSTM_best.keras"
        )


        callbacks = [

            EarlyStopping(

                monitor="val_loss",

                patience=
                EARLY_STOPPING_PATIENCE,

                restore_best_weights=True,

                verbose=1
            ),


            ReduceLROnPlateau(

                monitor="val_loss",

                factor=0.5,

                patience=LR_PATIENCE,

                min_lr=1e-7,

                verbose=1
            ),


            ModelCheckpoint(

                model_path,

                monitor="val_loss",

                save_best_only=True,

                verbose=1
            )
        ]


        self.history = self.model.fit(

            X_train,

            {
                "Kp10":
                    y_train[:, 0],

                "Dst":
                    y_train[:, 1]
            },

            validation_data=(

                X_val,

                {
                    "Kp10":
                        y_val[:, 0],

                    "Dst":
                        y_val[:, 1]
                }
            ),

            epochs=epochs,

            batch_size=batch_size,

            callbacks=callbacks,

            verbose=1
        )


        print(
            "\nBest BiLSTM model saved to:"
        )

        print(
            model_path
        )


        return self.history


    # ========================================================
    # INVERSE TRANSFORMATION
    # ========================================================

    def inverse_transform_targets(
        self,
        y_scaled
    ):

        kp_index = (
            self.all_columns.index(
                "Kp10"
            )
        )

        dst_index = (
            self.all_columns.index(
                "Dst"
            )
        )


        dummy = np.zeros(

            (
                len(y_scaled),

                len(self.all_columns)
            )
        )


        dummy[
            :,
            kp_index
        ] = y_scaled[
            :,
            0
        ]


        dummy[
            :,
            dst_index
        ] = y_scaled[
            :,
            1
        ]


        original = (
            self.scaler.inverse_transform(
                dummy
            )
        )


        kp = original[
            :,
            kp_index
        ]

        dst = original[
            :,
            dst_index
        ]


        return kp, dst


    # ========================================================
    # EVALUATION
    # ========================================================

    def evaluate(
        self,
        X_test,
        y_test
    ):

        print("\n" + "=" * 80)
        print(
            "BiLSTM MODEL EVALUATION"
        )
        print("=" * 80)


        predictions = (
            self.model.predict(
                X_test,
                verbose=0
            )
        )


        # ----------------------------------------------------
        # Scaled predictions
        # ----------------------------------------------------

        kp_pred_scaled = (
            predictions[0]
            .flatten()
        )

        dst_pred_scaled = (
            predictions[1]
            .flatten()
        )


        y_pred_scaled = np.column_stack(

            [
                kp_pred_scaled,
                dst_pred_scaled
            ]
        )


        # ----------------------------------------------------
        # Convert to original units
        # ----------------------------------------------------

        actual_kp, actual_dst = (
            self.inverse_transform_targets(
                y_test
            )
        )


        predicted_kp, predicted_dst = (
            self.inverse_transform_targets(
                y_pred_scaled
            )
        )


        # ====================================================
        # METRICS
        # ====================================================

        metrics = {}


        # ----------------------------------------------------
        # Kp × 10
        # ----------------------------------------------------

        kp_rmse = np.sqrt(

            mean_squared_error(
                actual_kp,
                predicted_kp
            )
        )

        kp_mae = (
            mean_absolute_error(
                actual_kp,
                predicted_kp
            )
        )

        kp_r2 = (
            r2_score(
                actual_kp,
                predicted_kp
            )
        )


        if (
            np.std(actual_kp) > 0
            and
            np.std(predicted_kp) > 0
        ):

            kp_corr = pearsonr(

                actual_kp,
                predicted_kp
            )[0]

        else:

            kp_corr = np.nan


        metrics["Kp10"] = {

            "rmse":
                kp_rmse,

            "mae":
                kp_mae,

            "r2":
                kp_r2,

            "correlation":
                kp_corr
        }


        # ----------------------------------------------------
        # Dst
        # ----------------------------------------------------

        dst_rmse = np.sqrt(

            mean_squared_error(
                actual_dst,
                predicted_dst
            )
        )

        dst_mae = (
            mean_absolute_error(
                actual_dst,
                predicted_dst
            )
        )

        dst_r2 = (
            r2_score(
                actual_dst,
                predicted_dst
            )
        )


        if (
            np.std(actual_dst) > 0
            and
            np.std(predicted_dst) > 0
        ):

            dst_corr = pearsonr(

                actual_dst,
                predicted_dst
            )[0]

        else:

            dst_corr = np.nan


        # ----------------------------------------------------
        # Prediction Efficiency
        # ----------------------------------------------------

        denominator = np.sum(
            actual_dst ** 2
        )


        if denominator != 0:

            dst_pe = (

                1
                -
                np.sum(
                    (
                        actual_dst
                        -
                        predicted_dst
                    ) ** 2
                )
                /
                denominator
            )

        else:

            dst_pe = np.nan


        metrics["Dst"] = {

            "rmse":
                dst_rmse,

            "mae":
                dst_mae,

            "r2":
                dst_r2,

            "correlation":
                dst_corr,

            "pe":
                dst_pe
        }


        # ====================================================
        # PRINT PERFORMANCE
        # ====================================================

        print(
            "\nKp × 10 Performance"
        )

        print(
            "-" * 50
        )

        print(
            f"RMSE        : {kp_rmse:.4f}"
        )

        print(
            f"MAE         : {kp_mae:.4f}"
        )

        print(
            f"R²          : {kp_r2:.4f}"
        )

        print(
            f"Correlation : {kp_corr:.4f}"
        )


        print(
            "\nDst Performance"
        )

        print(
            "-" * 50
        )

        print(
            f"RMSE        : {dst_rmse:.4f} nT"
        )

        print(
            f"MAE         : {dst_mae:.4f} nT"
        )

        print(
            f"R²          : {dst_r2:.4f}"
        )

        print(
            f"Correlation : {dst_corr:.4f}"
        )

        print(
            f"PE          : {dst_pe:.4f}"
        )


        return {

            "actual_kp":
                actual_kp,

            "predicted_kp":
                predicted_kp,

            "actual_dst":
                actual_dst,

            "predicted_dst":
                predicted_dst,

            "metrics":
                metrics
        }


# ============================================================
# 8. STORM CLASSIFICATION
# ============================================================

class StormClassifier:


    # ========================================================
    # Kp CLASSIFICATION
    # ========================================================

    @staticmethod
    def classify_kp(kp10):

        categories = []


        for value in kp10:

            if value < 50:

                categories.append(
                    "Quiet"
                )

            elif value < 60:

                categories.append(
                    "G1 Minor"
                )

            elif value < 70:

                categories.append(
                    "G2 Moderate"
                )

            elif value < 80:

                categories.append(
                    "G3 Strong"
                )

            elif value < 90:

                categories.append(
                    "G4 Severe"
                )

            else:

                categories.append(
                    "G5 Extreme"
                )


        return np.array(
            categories
        )


    # ========================================================
    # Dst CLASSIFICATION
    # ========================================================

    @staticmethod
    def classify_dst(dst):

        categories = []


        for value in dst:

            if value > -30:

                categories.append(
                    "Quiet"
                )

            elif value > -50:

                categories.append(
                    "Minor"
                )

            elif value > -100:

                categories.append(
                    "Moderate"
                )

            elif value > -200:

                categories.append(
                    "Intense"
                )

            else:

                categories.append(
                    "Super Storm"
                )


        return np.array(
            categories
        )


    # ========================================================
    # COMBINED CLASSIFICATION
    # ========================================================

    @staticmethod
    def classify_combined(
        kp10,
        dst
    ):

        kp_classes = (
            StormClassifier.classify_kp(
                kp10
            )
        )

        dst_classes = (
            StormClassifier.classify_dst(
                dst
            )
        )


        combined = []


        for kp_class, dst_class in zip(

            kp_classes,
            dst_classes
        ):

            if (

                kp_class == "G5 Extreme"
                or
                dst_class == "Super Storm"

            ):

                combined.append(
                    "Extreme/Super Storm"
                )


            elif (

                kp_class == "G4 Severe"
                or
                dst_class == "Intense"

            ):

                combined.append(
                    "Severe/Intense"
                )


            elif (

                kp_class == "G3 Strong"
                or
                dst_class == "Moderate"

            ):

                combined.append(
                    "Strong/Moderate"
                )


            elif (

                kp_class == "G2 Moderate"
                or
                dst_class == "Minor"

            ):

                combined.append(
                    "Moderate/Minor"
                )


            elif kp_class == "G1 Minor":

                combined.append(
                    "Minor"
                )


            else:

                combined.append(
                    "Quiet"
                )


        return np.array(
            combined
        )


    # ========================================================
    # BINARY STORM
    # ========================================================

    @staticmethod
    def binary_storm(
        kp10,
        dst
    ):

        # Storm:
        # Kp >= 5 OR Dst <= -50 nT

        return (

            (
                np.asarray(kp10)
                >=
                50
            )

            |

            (
                np.asarray(dst)
                <=
                -50
            )

        ).astype(int)


    # ========================================================
    # CLASSIFICATION EVALUATION
    # ========================================================

    @staticmethod
    def evaluate(

        actual_kp,
        predicted_kp,

        actual_dst,
        predicted_dst

    ):

        print("\n" + "=" * 80)
        print(
            "BiLSTM STORM CLASSIFICATION"
        )
        print("=" * 80)


        # ====================================================
        # Kp CLASSIFICATION
        # ====================================================

        actual_kp_class = (
            StormClassifier.classify_kp(
                actual_kp
            )
        )

        predicted_kp_class = (
            StormClassifier.classify_kp(
                predicted_kp
            )
        )


        kp_accuracy = (
            accuracy_score(
                actual_kp_class,
                predicted_kp_class
            )
        )


        kp_labels = [

            "Quiet",
            "G1 Minor",
            "G2 Moderate",
            "G3 Strong",
            "G4 Severe",
            "G5 Extreme"
        ]


        kp_cm = confusion_matrix(

            actual_kp_class,

            predicted_kp_class,

            labels=kp_labels
        )


        print(
            "\nKp × 10 Classification Accuracy:"
        )

        print(
            f"{kp_accuracy:.4f}"
        )


        print(
            "\nKp × 10 Classification Report:"
        )

        print(

            classification_report(

                actual_kp_class,

                predicted_kp_class,

                labels=kp_labels,

                zero_division=0
            )
        )


        # ====================================================
        # DST CLASSIFICATION
        # ====================================================

        actual_dst_class = (
            StormClassifier.classify_dst(
                actual_dst
            )
        )

        predicted_dst_class = (
            StormClassifier.classify_dst(
                predicted_dst
            )
        )


        dst_accuracy = (
            accuracy_score(
                actual_dst_class,
                predicted_dst_class
            )
        )


        dst_labels = [

            "Quiet",
            "Minor",
            "Moderate",
            "Intense",
            "Super Storm"
        ]


        dst_cm = confusion_matrix(

            actual_dst_class,

            predicted_dst_class,

            labels=dst_labels
        )


        print(
            "\nDst Classification Accuracy:"
        )

        print(
            f"{dst_accuracy:.4f}"
        )


        print(
            "\nDst Classification Report:"
        )

        print(

            classification_report(

                actual_dst_class,

                predicted_dst_class,

                labels=dst_labels,

                zero_division=0
            )
        )


        # ====================================================
        # BINARY CLASSIFICATION
        # ====================================================

        actual_binary = (
            StormClassifier.binary_storm(

                actual_kp,
                actual_dst
            )
        )


        predicted_binary = (
            StormClassifier.binary_storm(

                predicted_kp,
                predicted_dst
            )
        )


        binary_accuracy = (
            accuracy_score(
                actual_binary,
                predicted_binary
            )
        )


        binary_precision = (
            precision_score(

                actual_binary,

                predicted_binary,

                zero_division=0
            )
        )


        binary_recall = (
            recall_score(

                actual_binary,

                predicted_binary,

                zero_division=0
            )
        )


        binary_f1 = (
            f1_score(

                actual_binary,

                predicted_binary,

                zero_division=0
            )
        )


        binary_cm = confusion_matrix(

            actual_binary,

            predicted_binary
        )


        print(
            "\nCombined Kp × 10 + Dst Binary Classification"
        )

        print(
            "-" * 60
        )

        print(
            f"Accuracy  : {binary_accuracy:.4f}"
        )

        print(
            f"Precision : {binary_precision:.4f}"
        )

        print(
            f"Recall    : {binary_recall:.4f}"
        )

        print(
            f"F1 Score  : {binary_f1:.4f}"
        )


        return {

            "actual_kp_class":
                actual_kp_class,

            "predicted_kp_class":
                predicted_kp_class,

            "kp_cm":
                kp_cm,

            "actual_dst_class":
                actual_dst_class,

            "predicted_dst_class":
                predicted_dst_class,

            "dst_cm":
                dst_cm,

            "actual_binary":
                actual_binary,

            "predicted_binary":
                predicted_binary,

            "binary_cm":
                binary_cm,

            "kp_accuracy":
                kp_accuracy,

            "dst_accuracy":
                dst_accuracy,

            "binary_accuracy":
                binary_accuracy,

            "binary_precision":
                binary_precision,

            "binary_recall":
                binary_recall,

            "binary_f1":
                binary_f1
        }


# ============================================================
# 9. VISUALIZATION FUNCTIONS
# ============================================================

class Visualizer:


    # ========================================================
    # TRAINING HISTORY
    # ========================================================

    @staticmethod
    def plot_training_history(
        history
    ):

        # ----------------------------------------------------
        # Total loss
        # ----------------------------------------------------

        plt.figure(
            figsize=(12, 6)
        )

        plt.plot(

            history.history[
                "loss"
            ],

            label="Training Loss"
        )

        plt.plot(

            history.history[
                "val_loss"
            ],

            linestyle="--",

            label="Validation Loss"
        )

        plt.xlabel(
            "Epoch"
        )

        plt.ylabel(
            "Loss"
        )

        plt.title(
            "BiLSTM Training and Validation Loss"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_training_validation_loss.png"
            ),

            dpi=300
        )

        plt.show()


        # ----------------------------------------------------
        # Kp10 MAE
        # ----------------------------------------------------

        if "Kp10_mae" in history.history:

            plt.figure(
                figsize=(12, 6)
            )

            plt.plot(

                history.history[
                    "Kp10_mae"
                ],

                label="Training Kp × 10 MAE"
            )

            plt.plot(

                history.history[
                    "val_Kp10_mae"
                ],

                linestyle="--",

                label="Validation Kp × 10 MAE"
            )

            plt.xlabel(
                "Epoch"
            )

            plt.ylabel(
                "MAE"
            )

            plt.title(
                "BiLSTM Kp × 10 Training and Validation MAE"
            )

            plt.legend()

            plt.grid(
                True,
                alpha=0.3
            )

            plt.tight_layout()


            plt.savefig(

                os.path.join(

                    OUTPUT_DIR,

                    "BiLSTM_Kp10_MAE.png"
                ),

                dpi=300
            )

            plt.show()


        # ----------------------------------------------------
        # Dst MAE
        # ----------------------------------------------------

        if "Dst_mae" in history.history:

            plt.figure(
                figsize=(12, 6)
            )

            plt.plot(

                history.history[
                    "Dst_mae"
                ],

                label="Training Dst MAE"
            )

            plt.plot(

                history.history[
                    "val_Dst_mae"
                ],

                linestyle="--",

                label="Validation Dst MAE"
            )

            plt.xlabel(
                "Epoch"
            )

            plt.ylabel(
                "MAE (nT)"
            )

            plt.title(
                "BiLSTM Dst Training and Validation MAE"
            )

            plt.legend()

            plt.grid(
                True,
                alpha=0.3
            )

            plt.tight_layout()


            plt.savefig(

                os.path.join(

                    OUTPUT_DIR,

                    "BiLSTM_Dst_MAE.png"
                ),

                dpi=300
            )

            plt.show()


    # ========================================================
    # DST PREDICTION
    # ========================================================

    @staticmethod
    def plot_dst_prediction(
        result,
        sample_size=2000
    ):

        actual = (
            result[
                "actual_dst"
            ][:sample_size]
        )

        predicted = (
            result[
                "predicted_dst"
            ][:sample_size]
        )


        plt.figure(
            figsize=(15, 6)
        )


        plt.plot(

            actual,

            label="Actual Dst",

            linewidth=1
        )


        plt.plot(

            predicted,

            label="Predicted Dst",

            linewidth=1
        )


        plt.axhline(

            -30,

            linestyle="--",

            label="Minor: -30 nT"
        )


        plt.axhline(

            -50,

            linestyle="--",

            label="Moderate: -50 nT"
        )


        plt.axhline(

            -100,

            linestyle="--",

            label="Intense: -100 nT"
        )


        plt.axhline(

            -200,

            linestyle="--",

            label="Super Storm: -200 nT"
        )


        plt.xlabel(
            "Time Step"
        )

        plt.ylabel(
            "Dst (nT)"
        )

        plt.title(
            "BiLSTM - Dst Prediction"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_prediction.png"
            ),

            dpi=300
        )

        plt.show()


    # ========================================================
    # KP PREDICTION
    # ========================================================

    @staticmethod
    def plot_kp_prediction(
        result,
        sample_size=2000
    ):

        actual = (
            result[
                "actual_kp"
            ][:sample_size]
        )

        predicted = (
            result[
                "predicted_kp"
            ][:sample_size]
        )


        plt.figure(
            figsize=(15, 6)
        )


        plt.plot(

            actual,

            label="Actual Kp × 10",

            linewidth=1
        )


        plt.plot(

            predicted,

            label="Predicted Kp × 10",

            linewidth=1
        )


        # ----------------------------------------------------
        # Kp thresholds
        # ----------------------------------------------------

        plt.axhline(

            50,

            linestyle="--",

            label="Kp = 5 (G1)"
        )


        plt.axhline(

            60,

            linestyle="--",

            label="Kp = 6 (G2)"
        )


        plt.axhline(

            70,

            linestyle="--",

            label="Kp = 7 (G3)"
        )


        plt.axhline(

            80,

            linestyle="--",

            label="Kp = 8 (G4)"
        )


        plt.axhline(

            90,

            linestyle="--",

            label="Kp = 9 (G5)"
        )


        plt.xlabel(
            "Time Step"
        )

        plt.ylabel(
            "Kp × 10"
        )

        plt.title(
            "BiLSTM - Kp × 10 Prediction"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Kp10_prediction.png"
            ),

            dpi=300
        )

        plt.show()


    # ========================================================
    # SCATTER PLOTS
    # ========================================================

    @staticmethod
    def plot_scatter(
        result
    ):

        # ----------------------------------------------------
        # Dst scatter
        # ----------------------------------------------------

        plt.figure(
            figsize=(7, 7)
        )


        plt.scatter(

            result["actual_dst"],

            result["predicted_dst"],

            alpha=0.4,

            s=10
        )


        minimum = min(

            result[
                "actual_dst"
            ].min(),

            result[
                "predicted_dst"
            ].min()
        )


        maximum = max(

            result[
                "actual_dst"
            ].max(),

            result[
                "predicted_dst"
            ].max()
        )


        plt.plot(

            [
                minimum,
                maximum
            ],

            [
                minimum,
                maximum
            ],

            linestyle="--",

            linewidth=2
        )


        plt.xlabel(
            "Actual Dst (nT)"
        )

        plt.ylabel(
            "Predicted Dst (nT)"
        )

        plt.title(

            "BiLSTM - Dst Prediction\n"
            f"R² = "
            f"{result['metrics']['Dst']['r2']:.4f}"
        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_scatter.png"
            ),

            dpi=300
        )

        plt.show()


        # ----------------------------------------------------
        # Kp scatter
        # ----------------------------------------------------

        plt.figure(
            figsize=(7, 7)
        )


        plt.scatter(

            result["actual_kp"],

            result["predicted_kp"],

            alpha=0.4,

            s=10
        )


        minimum = min(

            result[
                "actual_kp"
            ].min(),

            result[
                "predicted_kp"
            ].min()
        )


        maximum = max(

            result[
                "actual_kp"
            ].max(),

            result[
                "predicted_kp"
            ].max()
        )


        plt.plot(

            [
                minimum,
                maximum
            ],

            [
                minimum,
                maximum
            ],

            linestyle="--",

            linewidth=2
        )


        plt.xlabel(
            "Actual Kp × 10"
        )

        plt.ylabel(
            "Predicted Kp × 10"
        )

        plt.title(

            "BiLSTM - Kp × 10 Prediction\n"
            f"R² = "
            f"{result['metrics']['Kp10']['r2']:.4f}"
        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Kp10_scatter.png"
            ),

            dpi=300
        )

        plt.show()


    # ========================================================
    # RESIDUALS
    # ========================================================

    @staticmethod
    def plot_residuals(
        result
    ):

        # ----------------------------------------------------
        # Dst residuals
        # ----------------------------------------------------

        residuals = (

            result["actual_dst"]

            -

            result["predicted_dst"]
        )


        plt.figure(
            figsize=(10, 6)
        )


        plt.scatter(

            result["predicted_dst"],

            residuals,

            alpha=0.4,

            s=10
        )


        plt.axhline(

            0,

            linestyle="--",

            linewidth=2
        )


        plt.xlabel(
            "Predicted Dst (nT)"
        )

        plt.ylabel(
            "Residual (nT)"
        )

        plt.title(
            "BiLSTM - Dst Residual Analysis"
        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_residuals.png"
            ),

            dpi=300
        )

        plt.show()


        # ----------------------------------------------------
        # Residual histogram
        # ----------------------------------------------------

        plt.figure(
            figsize=(10, 6)
        )


        plt.hist(

            residuals,

            bins=50
        )


        plt.xlabel(
            "Residual (nT)"
        )

        plt.ylabel(
            "Frequency"
        )

        plt.title(
            "BiLSTM - Dst Residual Distribution"
        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_residual_histogram.png"
            ),

            dpi=300
        )

        plt.show()


    # ========================================================
    # CONFUSION MATRIX
    # ========================================================

    @staticmethod
    def plot_confusion_matrix(

        cm,
        labels,
        title,
        filename
    ):

        plt.figure(
            figsize=(9, 7)
        )


        sns.heatmap(

            cm,

            annot=True,

            fmt="d",

            cmap="Blues",

            xticklabels=labels,

            yticklabels=labels
        )


        plt.xlabel(
            "Predicted"
        )

        plt.ylabel(
            "Actual"
        )

        plt.title(
            title
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                filename
            ),

            dpi=300
        )

        plt.show()


# ============================================================
# 10. ROC AND PRECISION-RECALL
# ============================================================

def add_binary_classification(
    result
):

    # --------------------------------------------------------
    # Actual binary storm
    # --------------------------------------------------------

    actual_binary = (

        StormClassifier.binary_storm(

            result["actual_kp"],

            result["actual_dst"]
        )
    )


    # --------------------------------------------------------
    # Continuous predicted storm score
    #
    # Kp score:
    #     predicted Kp10 / 50
    #
    # Dst score:
    #     -predicted Dst / 50
    #
    # Maximum of both is used as the storm score.
    # --------------------------------------------------------

    kp_score = (

        result["predicted_kp"]

        /

        50.0
    )


    dst_score = (

        -result["predicted_dst"]

        /

        50.0
    )


    storm_score = np.maximum(

        kp_score,

        dst_score
    )


    result[
        "actual_binary"
    ] = actual_binary


    result[
        "predicted_binary_score"
    ] = storm_score


# ============================================================
# ROC CURVE
# ============================================================

def plot_roc_curve(
    result
):

    actual = (
        result[
            "actual_binary"
        ]
    )

    score = (
        result[
            "predicted_binary_score"
        ]
    )


    fpr, tpr, thresholds = (
        roc_curve(
            actual,
            score
        )
    )


    roc_auc = auc(
        fpr,
        tpr
    )


    plt.figure(
        figsize=(9, 7)
    )


    plt.plot(

        fpr,

        tpr,

        linewidth=2,

        label=
        f"BiLSTM AUC = {roc_auc:.4f}"
    )


    plt.plot(

        [0, 1],

        [0, 1],

        linestyle="--",

        label="Random classifier"
    )


    plt.xlabel(
        "False Positive Rate"
    )

    plt.ylabel(
        "True Positive Rate"
    )

    plt.title(
        "ROC Curve - BiLSTM Geomagnetic Storm Prediction"
    )

    plt.legend()

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()


    plt.savefig(

        os.path.join(

            OUTPUT_DIR,

            "BiLSTM_ROC_curve.png"
        ),

        dpi=300
    )

    plt.show()


    print(
        f"\nBiLSTM ROC-AUC: {roc_auc:.4f}"
    )


    return roc_auc


# ============================================================
# PRECISION-RECALL CURVE
# ============================================================

def plot_precision_recall_curve_bilstm(
    result
):

    actual = (
        result[
            "actual_binary"
        ]
    )

    score = (
        result[
            "predicted_binary_score"
        ]
    )


    precision, recall, thresholds = (

        precision_recall_curve(

            actual,

            score
        )
    )


    average_precision = (

        average_precision_score(

            actual,

            score
        )
    )


    plt.figure(
        figsize=(9, 7)
    )


    plt.plot(

        recall,

        precision,

        linewidth=2,

        label=
        f"BiLSTM AP = "
        f"{average_precision:.4f}"
    )


    plt.xlabel(
        "Recall"
    )

    plt.ylabel(
        "Precision"
    )

    plt.title(
        "Precision-Recall Curve - BiLSTM"
    )

    plt.legend()

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()


    plt.savefig(

        os.path.join(

            OUTPUT_DIR,

            "BiLSTM_PR_curve.png"
        ),

        dpi=300
    )

    plt.show()


    print(
        f"BiLSTM Average Precision: "
        f"{average_precision:.4f}"
    )


    return average_precision


# ============================================================
# 11. SAVE ALL PREDICTIONS
# ============================================================

def save_prediction_results(
    result
):

    output_df = pd.DataFrame({

        "Actual_Kp10":
            result["actual_kp"],

        "Predicted_Kp10":
            result["predicted_kp"],

        "Actual_Dst":
            result["actual_dst"],

        "Predicted_Dst":
            result["predicted_dst"]
    })


    # --------------------------------------------------------
    # Kp classes
    # --------------------------------------------------------

    output_df[
        "Actual_Kp_Class"
    ] = StormClassifier.classify_kp(

        result["actual_kp"]
    )


    output_df[
        "Predicted_Kp_Class"
    ] = StormClassifier.classify_kp(

        result["predicted_kp"]
    )


    # --------------------------------------------------------
    # Dst classes
    # --------------------------------------------------------

    output_df[
        "Actual_Dst_Class"
    ] = StormClassifier.classify_dst(

        result["actual_dst"]
    )


    output_df[
        "Predicted_Dst_Class"
    ] = StormClassifier.classify_dst(

        result["predicted_dst"]
    )


    # --------------------------------------------------------
    # Combined classification
    # --------------------------------------------------------

    output_df[
        "Actual_Combined_Class"
    ] = StormClassifier.classify_combined(

        result["actual_kp"],

        result["actual_dst"]
    )


    output_df[
        "Predicted_Combined_Class"
    ] = StormClassifier.classify_combined(

        result["predicted_kp"],

        result["predicted_dst"]
    )


    # --------------------------------------------------------
    # Binary storm
    # --------------------------------------------------------

    output_df[
        "Actual_Storm"
    ] = StormClassifier.binary_storm(

        result["actual_kp"],

        result["actual_dst"]
    )


    output_df[
        "Predicted_Storm"
    ] = StormClassifier.binary_storm(

        result["predicted_kp"],

        result["predicted_dst"]
    )


    # --------------------------------------------------------
    # Continuous storm score
    # --------------------------------------------------------

    output_df[
        "Predicted_Storm_Score"
    ] = result[
        "predicted_binary_score"
    ]


    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    filepath = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_predictions.csv"
    )


    output_df.to_csv(

        filepath,

        index=False
    )


    print(
        "\nPrediction results saved:"
    )

    print(
        filepath
    )


    return output_df


# ============================================================
# 12. SAVE PERFORMANCE METRICS
# ============================================================

def save_metrics(
    result,
    classification_result,
    roc_auc,
    average_precision
):

    kp = (
        result[
            "metrics"
        ]["Kp10"]
    )

    dst = (
        result[
            "metrics"
        ]["Dst"]
    )


    metrics_df = pd.DataFrame({

        "Model": [
            "BiLSTM"
        ],

        "Kp10_RMSE": [
            kp["rmse"]
        ],

        "Kp10_MAE": [
            kp["mae"]
        ],

        "Kp10_R2": [
            kp["r2"]
        ],

        "Kp10_Correlation": [
            kp["correlation"]
        ],

        "Dst_RMSE_nT": [
            dst["rmse"]
        ],

        "Dst_MAE_nT": [
            dst["mae"]
        ],

        "Dst_R2": [
            dst["r2"]
        ],

        "Dst_Correlation": [
            dst["correlation"]
        ],

        "Dst_PE": [
            dst["pe"]
        ],

        "Kp10_Accuracy": [
            classification_result[
                "kp_accuracy"
            ]
        ],

        "Dst_Accuracy": [
            classification_result[
                "dst_accuracy"
            ]
        ],

        "Combined_Accuracy": [
            classification_result[
                "binary_accuracy"
            ]
        ],

        "Combined_Precision": [
            classification_result[
                "binary_precision"
            ]
        ],

        "Combined_Recall": [
            classification_result[
                "binary_recall"
            ]
        ],

        "Combined_F1": [
            classification_result[
                "binary_f1"
            ]
        ],

        "ROC_AUC": [
            roc_auc
        ],

        "Average_Precision": [
            average_precision
        ]
    })


    filepath = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_performance_metrics.csv"
    )


    metrics_df.to_csv(

        filepath,

        index=False
    )


    print(
        "\nPerformance metrics saved:"
    )

    print(
        filepath
    )


    return metrics_df


# ============================================================
# 13. MAIN FUNCTION
# ============================================================

def main():

    print("\n")

    print("=" * 80)

    print(
        "GEOMAGNETIC STORM PREDICTION USING BiLSTM"
    )

    print(
        "Targets: Kp × 10 + Dst"
    )

    print("=" * 80)


    # ========================================================
    # STEP 1
    # LOAD DATA
    # ========================================================

    preprocessor = DataPreprocessor(

        DATA_PATH
    )


    preprocessor.load_data()

    preprocessor.check_columns()

    preprocessor.handle_missing_values()

    preprocessor.clean_data()


    df = (
        preprocessor.get_processed_data()
    )


    # --------------------------------------------------------
    # Save cleaned data
    # --------------------------------------------------------

    cleaned_path = os.path.join(

        OUTPUT_DIR,

        "cleaned_geomagnetic_data.csv"
    )


    df.to_csv(

        cleaned_path,

        index=False
    )


    print(
        "\nCleaned data saved:"
    )

    print(
        cleaned_path
    )


    # ========================================================
    # STEP 2
    # CREATE SEQUENCES
    # ========================================================

    seq_gen = SequenceGenerator(

        df,

        lookback=LOOKBACK
    )


    X, y = (
        seq_gen.create_sequences()
    )


    # ========================================================
    # STEP 3
    # SPLIT DATA
    # ========================================================

    (

        X_train,
        X_val,
        X_test,

        y_train,
        y_val,
        y_test

    ) = seq_gen.split_data(

        X,

        y,

        train_size=TRAIN_SIZE,

        validation_size=VALIDATION_SIZE,

        test_size=TEST_SIZE
    )


    # ========================================================
    # STEP 4
    # INPUT SHAPE
    # ========================================================

    input_shape = (

        X_train.shape[1],

        X_train.shape[2]
    )


    print(
        "\nBiLSTM input shape:"
    )

    print(
        input_shape
    )


    # ========================================================
    # STEP 5
    # BUILD BiLSTM
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "BUILDING BiLSTM MODEL"
    )

    print("=" * 80)


    model = build_bilstm(

        input_shape=input_shape,

        units=64,

        dropout=0.20
    )


    # ========================================================
    # STEP 6
    # COMPILE
    # ========================================================

    trainer = BiLSTMTrainer(

        model,

        seq_gen.scaler,

        seq_gen.all_columns
    )


    trainer.compile_model(

        learning_rate=LEARNING_RATE
    )


    # ========================================================
    # STEP 7
    # MODEL SUMMARY
    # ========================================================

    print("\n")

    model.summary()


    # ========================================================
    # SAVE MODEL SUMMARY
    # ========================================================

    summary_path = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_model_summary.txt"
    )


    with open(

        summary_path,

        "w"
    ) as f:

        model.summary(
            print_fn=f.write
        )


    print(
        "\nModel summary saved:"
    )

    print(
        summary_path
    )


    # ========================================================
    # STEP 8
    # TRAIN
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "TRAINING BiLSTM"
    )

    print("=" * 80)


    history = trainer.train(

        X_train,

        y_train,

        X_val,

        y_val,

        epochs=EPOCHS,

        batch_size=BATCH_SIZE
    )


    # ========================================================
    # STEP 9
    # EVALUATE REGRESSION
    # ========================================================

    result = trainer.evaluate(

        X_test,

        y_test
    )


    # ========================================================
    # STEP 10
    # STORM CLASSIFICATION
    # ========================================================

    classification_result = (

        StormClassifier.evaluate(

            result["actual_kp"],

            result["predicted_kp"],

            result["actual_dst"],

            result["predicted_dst"]
        )
    )


    # ========================================================
    # STEP 11
    # ADD BINARY STORM SCORE
    # ========================================================

    add_binary_classification(

        result
    )


    # ========================================================
    # STEP 12
    # TRAINING PLOTS
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "GENERATING TRAINING OUTPUTS"
    )

    print("=" * 80)


    Visualizer.plot_training_history(

        history
    )


    # ========================================================
    # STEP 13
    # DST PREDICTION
    # ========================================================

    Visualizer.plot_dst_prediction(

        result,

        sample_size=2000
    )


    # ========================================================
    # STEP 14
    # KP PREDICTION
    # ========================================================

    Visualizer.plot_kp_prediction(

        result,

        sample_size=2000
    )


    # ========================================================
    # STEP 15
    # SCATTER PLOTS
    # ========================================================

    Visualizer.plot_scatter(

        result
    )


    # ========================================================
    # STEP 16
    # RESIDUAL ANALYSIS
    # ========================================================

    Visualizer.plot_residuals(

        result
    )


    # ========================================================
    # STEP 17
    # CONFUSION MATRICES
    # ========================================================

    Visualizer.plot_confusion_matrix(

        classification_result[
            "kp_cm"
        ],

        [

            "Quiet",
            "G1 Minor",
            "G2 Moderate",
            "G3 Strong",
            "G4 Severe",
            "G5 Extreme"

        ],

        "BiLSTM - Kp × 10 Storm Classification",

        "BiLSTM_Kp10_confusion_matrix.png"
    )


    Visualizer.plot_confusion_matrix(

        classification_result[
            "dst_cm"
        ],

        [

            "Quiet",
            "Minor",
            "Moderate",
            "Intense",
            "Super Storm"

        ],

        "BiLSTM - Dst Storm Classification",

        "BiLSTM_Dst_confusion_matrix.png"
    )


    Visualizer.plot_confusion_matrix(

        classification_result[
            "binary_cm"
        ],

        [

            "No Storm",
            "Storm"

        ],

        "BiLSTM - Combined Kp × 10 + Dst",

        "BiLSTM_Combined_confusion_matrix.png"
    )


    # ========================================================
    # STEP 18
    # ROC CURVE
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "GENERATING ROC CURVE"
    )

    print("=" * 80)


    roc_auc = plot_roc_curve(

        result
    )


    # ========================================================
    # STEP 19
    # PRECISION-RECALL CURVE
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "GENERATING PRECISION-RECALL CURVE"
    )

    print("=" * 80)


    average_precision = (
        plot_precision_recall_curve_bilstm(

            result
        )
    )


    # ========================================================
    # STEP 20
    # SAVE PREDICTIONS
    # ========================================================

    prediction_df = (
        save_prediction_results(

            result
        )
    )


    # ========================================================
    # STEP 21
    # SAVE METRICS
    # ========================================================

    metrics_df = save_metrics(

        result,

        classification_result,

        roc_auc,

        average_precision
    )


    # ========================================================
    # STEP 22
    # PRINT FINAL RESULT TABLE
    # ========================================================

    print("\n")

    print("=" * 110)

    print(
        "BiLSTM FINAL PERFORMANCE"
    )

    print("=" * 110)


    print(
        metrics_df.to_string(
            index=False
        )
    )


    # ========================================================
    # STEP 23
    # FINAL SUMMARY
    # ========================================================

    print("\n")

    print("=" * 80)

    print(
        "FINAL BiLSTM RESULTS"
    )

    print("=" * 80)


    print(
        "\nRegression Results:"
    )

    print(
        f"Kp × 10 RMSE = "
        f"{result['metrics']['Kp10']['rmse']:.4f}"
    )

    print(
        f"Kp × 10 MAE  = "
        f"{result['metrics']['Kp10']['mae']:.4f}"
    )

    print(
        f"Kp × 10 R²   = "
        f"{result['metrics']['Kp10']['r2']:.4f}"
    )

    print(
        f"Kp × 10 r    = "
        f"{result['metrics']['Kp10']['correlation']:.4f}"
    )


    print(
        f"\nDst RMSE      = "
        f"{result['metrics']['Dst']['rmse']:.4f} nT"
    )

    print(
        f"Dst MAE       = "
        f"{result['metrics']['Dst']['mae']:.4f} nT"
    )

    print(
        f"Dst R²        = "
        f"{result['metrics']['Dst']['r2']:.4f}"
    )

    print(
        f"Dst r         = "
        f"{result['metrics']['Dst']['correlation']:.4f}"
    )

    print(
        f"Dst PE        = "
        f"{result['metrics']['Dst']['pe']:.4f}"
    )


    print(
        "\nClassification Results:"
    )

    print(
        f"Kp × 10 Accuracy = "
        f"{classification_result['kp_accuracy']:.4f}"
    )

    print(
        f"Dst Accuracy     = "
        f"{classification_result['dst_accuracy']:.4f}"
    )

    print(
        f"Combined Accuracy = "
        f"{classification_result['binary_accuracy']:.4f}"
    )

    print(
        f"Combined Precision = "
        f"{classification_result['binary_precision']:.4f}"
    )

    print(
        f"Combined Recall = "
        f"{classification_result['binary_recall']:.4f}"
    )

    print(
        f"Combined F1 = "
        f"{classification_result['binary_f1']:.4f}"
    )


    print(
        "\nROC/PR Results:"
    )

    print(
        f"ROC-AUC = {roc_auc:.4f}"
    )

    print(
        f"Average Precision = "
        f"{average_precision:.4f}"
    )


    print(
        "\nOutput directory:"
    )

    print(
        OUTPUT_DIR
    )


    print("\n")

    print("=" * 80)

    print(
        "BiLSTM ANALYSIS COMPLETED SUCCESSFULLY"
    )

    print("=" * 80)


    return (

        model,

        history,

        result,

        classification_result,

        metrics_df,

        prediction_df
    )


# ============================================================
# 14. RUN PROGRAM
# ============================================================

if __name__ == "__main__":

    (

        model,

        history,

        result,

        classification_result,

        metrics_df,

        prediction_df

    ) = main()
```

```python
# ============================================================
# GEOMAGNETIC STORM PREDICTION USING BiLSTM
# ============================================================
# Model:
#   BiLSTM only
#
# Targets:
#   1. Kp × 10
#   2. Dst
#
# Input features:
#   YEAR, DOY, HR, B, Bx, By, Bz, PD, FP, E
#
# Data split:
#   Train      = 70%
#   Validation = 15%
#   Test       = 15%
#
# Lookback:
#   48 hours
#
# Optimizer:
#   Adam
#
# EarlyStopping:
#   patience = 5
#
# Author: Prajwal Bhandari
# Date: September 2026
# ============================================================


# ============================================================
# 0. INSTALL LIBRARIES - GOOGLE COLAB
# ============================================================

# Uncomment if required:
# !pip install -q tensorflow scikit-learn scipy seaborn


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import re
import random
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr

from sklearn.preprocessing import MinMaxScaler

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_curve,
    auc,
    precision_recall_curve,
    average_precision_score
)

import tensorflow as tf

from tensorflow.keras.models import Model

from tensorflow.keras.layers import (
    Input,
    LSTM,
    Dense,
    Dropout,
    Bidirectional
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

from tensorflow.keras.optimizers import Adam


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)


# ============================================================
# 3. CONFIGURATION
# ============================================================

# ------------------------------------------------------------
# DATA PATH
# ------------------------------------------------------------

DATA_PATH = "/content/Sun.csv"

# If using Google Drive:
# DATA_PATH = "/content/drive/MyDrive/D.csv"


# ------------------------------------------------------------
# OUTPUT DIRECTORY
# ------------------------------------------------------------

OUTPUT_DIR = "/content/geomagnetic_BiLSTM_results"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ------------------------------------------------------------
# TIME-SERIES SETTINGS
# ------------------------------------------------------------

LOOKBACK = 48

TRAIN_SIZE = 0.70
VALIDATION_SIZE = 0.15
TEST_SIZE = 0.15

EPOCHS = 90

BATCH_SIZE = 32

LEARNING_RATE = 0.001

EARLY_STOPPING_PATIENCE = 7

LR_PATIENCE = 3


# ============================================================
# 4. DATA PREPROCESSOR
# ============================================================

class DataPreprocessor:

    def __init__(self, filepath):

        self.filepath = filepath

        self.df = None

        self.missing_values_map = {

            "B": [
                999.99,
                9999.99
            ],

            "Bx": [
                999.99,
                9999.99
            ],

            "By": [
                999.99,
                9999.99
            ],

            "Bz": [
                999.99,
                9999.99
            ],

            "PD": [
                999.9,
                999.99,
                9999.9
            ],

            "FP": [
                99.99,
                999.99
            ],

            "E": [
                999.99,
                9999.99
            ],

            "Kp": [
                9.99,
                99.99
            ],

            "Dst": [
                9999.9,
                9999.99
            ]
        }


    # ========================================================
    # NORMALIZE COLUMN NAMES
    # ========================================================

    def normalize_column_names(self):

        new_columns = []

        for col in self.df.columns:

            col = str(col).strip()

            col_clean = re.sub(
                r"\s+",
                " ",
                col
            )

            lower_col = col_clean.lower()

            if "kp" in lower_col:

                col_clean = "Kp"

            elif lower_col == "dst":

                col_clean = "Dst"

            elif lower_col == "year":

                col_clean = "YEAR"

            elif lower_col == "doy":

                col_clean = "DOY"

            elif lower_col == "hr":

                col_clean = "HR"

            new_columns.append(
                col_clean
            )

        self.df.columns = new_columns

        return self


    # ========================================================
    # LOAD DATA
    # ========================================================

    def load_data(self):

        print("=" * 80)
        print("LOADING GEOMAGNETIC DATA")
        print("=" * 80)

        self.df = pd.read_csv(
            self.filepath
        )

        print(
            "\nOriginal dataset shape:",
            self.df.shape
        )

        print(
            "\nOriginal columns:"
        )

        print(
            list(self.df.columns)
        )

        self.normalize_column_names()

        print(
            "\nNormalized columns:"
        )

        print(
            list(self.df.columns)
        )

        return self


    # ========================================================
    # CHECK REQUIRED COLUMNS
    # ========================================================

    def check_columns(self):

        required_columns = [

            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E",
            "Kp",
            "Dst"
        ]

        missing = [

            col
            for col in required_columns
            if col not in self.df.columns
        ]

        if len(missing) > 0:

            raise ValueError(

                "\nMissing required columns:\n"
                + str(missing)
                + "\n\nAvailable columns:\n"
                + str(
                    list(
                        self.df.columns
                    )
                )
            )

        print(
            "\nAll required columns are available."
        )

        return self


    # ========================================================
    # HANDLE MISSING VALUES
    # ========================================================

    def handle_missing_values(self):

        print("\n" + "=" * 80)
        print("HANDLING MISSING VALUES")
        print("=" * 80)

        for col, values in (
            self.missing_values_map.items()
        ):

            if col in self.df.columns:

                for value in values:

                    count = (
                        self.df[col] == value
                    ).sum()

                    if count > 0:

                        self.df[col] = (
                            self.df[col]
                            .replace(
                                value,
                                np.nan
                            )
                        )

                        print(
                            f"{col}: replaced "
                            f"{count} values of {value}"
                        )

        return self


    # ========================================================
    # CLEAN DATA
    # ========================================================

    def clean_data(self):

        print("\n" + "=" * 80)
        print("CLEANING DATA")
        print("=" * 80)

        # ----------------------------------------------------
        # Convert columns to numeric
        # ----------------------------------------------------

        for col in self.df.columns:

            self.df[col] = pd.to_numeric(
                self.df[col],
                errors="coerce"
            )


        # ----------------------------------------------------
        # Sort chronologically
        # ----------------------------------------------------

        self.df = self.df.sort_values(
            [
                "YEAR",
                "DOY",
                "HR"
            ]
        ).reset_index(
            drop=True
        )


        # ----------------------------------------------------
        # Missing values
        # ----------------------------------------------------

        print(
            "\nMissing values before interpolation:"
        )

        missing = self.df.isna().sum()

        print(
            missing[
                missing > 0
            ]
        )


        # ----------------------------------------------------
        # Interpolation
        # ----------------------------------------------------

        numeric_columns = (
            self.df.select_dtypes(
                include=[np.number]
            ).columns
        )

        self.df[
            numeric_columns
        ] = (
            self.df[
                numeric_columns
            ]
            .interpolate(
                method="linear"
            )
            .ffill()
            .bfill()
        )


        # ----------------------------------------------------
        # Drop remaining NaN
        # ----------------------------------------------------

        before = len(
            self.df
        )

        self.df = self.df.dropna()

        after = len(
            self.df
        )

        print(
            f"\nRows removed: "
            f"{before - after}"
        )

        print(
            f"Final dataset shape: "
            f"{self.df.shape}"
        )


        # ====================================================
        # CREATE Kp × 10
        # ====================================================

        kp_max = self.df[
            "Kp"
        ].max()

        if kp_max <= 9.0:

            print(
                "\nKp is on standard 0–9 scale."
            )

            print(
                "Creating Kp10 = Kp × 10."
            )

            self.df[
                "Kp10"
            ] = (
                self.df["Kp"]
                * 10.0
            )

        else:

            print(
                "\nKp appears to already be Kp × 10."
            )

            self.df[
                "Kp10"
            ] = self.df["Kp"]


        # ----------------------------------------------------
        # Remove original Kp
        # ----------------------------------------------------

        self.df.drop(
            columns=["Kp"],
            inplace=True
        )


        # ----------------------------------------------------
        # Final columns
        # ----------------------------------------------------

        final_columns = [

            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E",
            "Kp10",
            "Dst"
        ]

        self.df = self.df[
            final_columns
        ]


        print(
            "\nFinal columns:"
        )

        print(
            list(
                self.df.columns
            )
        )


        print(
            "\nKp × 10 statistics:"
        )

        print(
            self.df[
                "Kp10"
            ].describe()
        )


        print(
            "\nDst statistics:"
        )

        print(
            self.df[
                "Dst"
            ].describe()
        )


        return self


    def get_processed_data(self):

        return self.df


# ============================================================
# 5. SEQUENCE GENERATOR
# ============================================================

class SequenceGenerator:

    def __init__(
        self,
        data,
        lookback=48
    ):

        self.data = data

        self.lookback = lookback

        self.scaler = MinMaxScaler()


        # ----------------------------------------------------
        # INPUT FEATURES
        # ----------------------------------------------------

        self.feature_columns = [

            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E"
        ]


        # ----------------------------------------------------
        # TARGETS
        # ----------------------------------------------------

        self.target_columns = [

            "Kp10",
            "Dst"
        ]


        self.all_columns = (
            self.feature_columns
            +
            self.target_columns
        )


    # ========================================================
    # CREATE SEQUENCES
    # ========================================================

    def create_sequences(self):

        print("\n" + "=" * 80)
        print("CREATING TIME-SERIES SEQUENCES")
        print("=" * 80)

        data_values = self.data[
            self.all_columns
        ].values


        # ----------------------------------------------------
        # Scale all variables
        # ----------------------------------------------------

        scaled_data = (
            self.scaler.fit_transform(
                data_values
            )
        )


        kp_index = (
            self.all_columns.index(
                "Kp10"
            )
        )

        dst_index = (
            self.all_columns.index(
                "Dst"
            )
        )


        X = []

        y_kp = []

        y_dst = []


        # ----------------------------------------------------
        # Sliding window
        # ----------------------------------------------------

        for i in range(
            len(scaled_data)
            -
            self.lookback
        ):

            X.append(

                scaled_data[
                    i:
                    i + self.lookback
                ]
            )

            y_kp.append(

                scaled_data[
                    i + self.lookback,
                    kp_index
                ]
            )

            y_dst.append(

                scaled_data[
                    i + self.lookback,
                    dst_index
                ]
            )


        X = np.array(
            X
        )

        y_kp = np.array(
            y_kp
        )

        y_dst = np.array(
            y_dst
        )


        y = np.column_stack(
            [
                y_kp,
                y_dst
            ]
        )


        print(
            "\nX shape:",
            X.shape
        )

        print(
            "y shape:",
            y.shape
        )

        print(
            "Lookback:",
            self.lookback,
            "hours"
        )

        print(
            "Number of input features:",
            X.shape[2]
        )

        print(
            "Number of target variables:",
            y.shape[1]
        )


        return X, y


    # ========================================================
    # 70 / 15 / 15 CHRONOLOGICAL SPLIT
    # ========================================================

    def split_data(
        self,
        X,
        y,
        train_size=0.70,
        validation_size=0.15,
        test_size=0.15
    ):

        print("\n" + "=" * 80)
        print(
            "70% TRAIN / 15% VALIDATION / 15% TEST"
        )
        print("=" * 80)


        if not np.isclose(
            train_size
            +
            validation_size
            +
            test_size,
            1.0
        ):

            raise ValueError(
                "Train + Validation + Test "
                "must equal 1.0"
            )


        n = len(X)


        train_end = int(
            n * train_size
        )

        val_end = int(
            n *
            (
                train_size
                +
                validation_size
            )
        )


        X_train = X[
            :train_end
        ]

        y_train = y[
            :train_end
        ]


        X_val = X[
            train_end:val_end
        ]

        y_val = y[
            train_end:val_end
        ]


        X_test = X[
            val_end:
        ]

        y_test = y[
            val_end:
        ]


        print(
            f"Training samples:   "
            f"{len(X_train):,}"
        )

        print(
            f"Validation samples: "
            f"{len(X_val):,}"
        )

        print(
            f"Testing samples:    "
            f"{len(X_test):,}"
        )


        print(
            "\nTraining percentage:",
            len(X_train) / n * 100
        )

        print(
            "Validation percentage:",
            len(X_val) / n * 100
        )

        print(
            "Testing percentage:",
            len(X_test) / n * 100
        )


        return (
            X_train,
            X_val,
            X_test,
            y_train,
            y_val,
            y_test
        )


# ============================================================
# 6. BiLSTM MODEL
# ============================================================

def build_bilstm(
    input_shape,
    units=64,
    dropout=0.20
):

    inputs = Input(
        shape=input_shape,
        name="Input"
    )


    # --------------------------------------------------------
    # BiLSTM Layer 1
    # --------------------------------------------------------

    x = Bidirectional(

        LSTM(
            units,
            return_sequences=True
        ),

        name="BiLSTM_1"

    )(inputs)


    x = Dropout(
        dropout,
        name="Dropout_1"
    )(x)


    # --------------------------------------------------------
    # BiLSTM Layer 2
    # --------------------------------------------------------

    x = Bidirectional(

        LSTM(
            units // 2,
            return_sequences=False
        ),

        name="BiLSTM_2"

    )(x)


    x = Dropout(
        dropout,
        name="Dropout_2"
    )(x)


    # --------------------------------------------------------
    # Dense layer
    # --------------------------------------------------------

    x = Dense(
        64,
        activation="relu",
        name="Dense_64"
    )(x)


    x = Dropout(
        dropout,
        name="Dropout_3"
    )(x)


    # ========================================================
    # OUTPUT 1: Kp × 10
    # ========================================================

    kp_output = Dense(
        1,
        name="Kp10"
    )(x)


    # ========================================================
    # OUTPUT 2: Dst
    # ========================================================

    dst_output = Dense(
        1,
        name="Dst"
    )(x)


    # --------------------------------------------------------
    # Build model
    # --------------------------------------------------------

    model = Model(

        inputs=inputs,

        outputs=[
            kp_output,
            dst_output
        ],

        name="BiLSTM_Geomagnetic_Storm_Model"
    )


    return model


# ============================================================
# 7. MODEL TRAINER
# ============================================================

class BiLSTMTrainer:

    def __init__(
        self,
        model,
        scaler,
        all_columns
    ):

        self.model = model

        self.scaler = scaler

        self.all_columns = all_columns

        self.history = None


    # ========================================================
    # COMPILE MODEL
    # ========================================================

    def compile_model(
        self,
        learning_rate=0.001
    ):

        self.model.compile(

            optimizer=Adam(
                learning_rate=
                learning_rate
            ),

            loss={

                "Kp10": "mse",

                "Dst": "mse"
            },

            metrics={

                "Kp10": [
                    "mae"
                ],

                "Dst": [
                    "mae"
                ]
            },

            loss_weights={

                "Kp10": 1.0,

                "Dst": 1.0
            }
        )


    # ========================================================
    # TRAIN MODEL
    # ========================================================

    def train(
        self,
        X_train,
        y_train,
        X_val,
        y_val,
        epochs=50,
        batch_size=32
    ):

        model_path = os.path.join(

            OUTPUT_DIR,

            "BiLSTM_best.keras"
        )


        callbacks = [

            EarlyStopping(

                monitor="val_loss",

                patience=
                EARLY_STOPPING_PATIENCE,

                restore_best_weights=True,

                verbose=1
            ),


            ReduceLROnPlateau(

                monitor="val_loss",

                factor=0.5,

                patience=LR_PATIENCE,

                min_lr=1e-7,

                verbose=1
            ),


            ModelCheckpoint(

                model_path,

                monitor="val_loss",

                save_best_only=True,

                verbose=1
            )
        ]


        self.history = self.model.fit(

            X_train,

            {
                "Kp10":
                    y_train[:, 0],

                "Dst":
                    y_train[:, 1]
            },

            validation_data=(

                X_val,

                {
                    "Kp10":
                        y_val[:, 0],

                    "Dst":
                        y_val[:, 1]
                }
            ),

            epochs=epochs,

            batch_size=batch_size,

            callbacks=callbacks,

            verbose=1
        )


        print(
            "\nBest BiLSTM model saved to:"
        )

        print(
            model_path
        )


        return self.history


    # ========================================================
    # INVERSE TRANSFORMATION
    # ========================================================

    def inverse_transform_targets(
        self,
        y_scaled
    ):

        kp_index = (
            self.all_columns.index(
                "Kp10"
            )
        )

        dst_index = (
            self.all_columns.index(
                "Dst"
            )
        )


        dummy = np.zeros(

            (
                len(y_scaled),

                len(self.all_columns)
            )
        )


        dummy[
            :,
            kp_index
        ] = y_scaled[
            :,
            0
        ]


        dummy[
            :,
            dst_index
        ] = y_scaled[
            :,
            1
        ]


        original = (
            self.scaler.inverse_transform(
                dummy
            )
        )


        kp = original[
            :,
            kp_index
        ]

        dst = original[
            :,
            dst_index
        ]


        return kp, dst


    # ========================================================
    # EVALUATION
    # ========================================================

    def evaluate(
        self,
        X_test,
        y_test
    ):

        print("\n" + "=" * 80)
        print(
            "BiLSTM MODEL EVALUATION"
        )
        print("=" * 80)


        predictions = (
            self.model.predict(
                X_test,
                verbose=0
            )
        )


        # ----------------------------------------------------
        # Scaled predictions
        # ----------------------------------------------------

        kp_pred_scaled = (
            predictions[0]
            .flatten()
        )

        dst_pred_scaled = (
            predictions[1]
            .flatten()
        )


        y_pred_scaled = np.column_stack(

            [
                kp_pred_scaled,
                dst_pred_scaled
            ]
        )


        # ----------------------------------------------------
        # Convert to original units
        # ----------------------------------------------------

        actual_kp, actual_dst = (
            self.inverse_transform_targets(
                y_test
            )
        )


        predicted_kp, predicted_dst = (
            self.inverse_transform_targets(
                y_pred_scaled
            )
        )


        # ====================================================
        # METRICS
        # ====================================================

        metrics = {}


        # ----------------------------------------------------
        # Kp × 10
        # ----------------------------------------------------

        kp_rmse = np.sqrt(

            mean_squared_error(
                actual_kp,
                predicted_kp
            )
        )

        kp_mae = (
            mean_absolute_error(
                actual_kp,
                predicted_kp
            )
        )

        kp_r2 = (
            r2_score(
                actual_kp,
                predicted_kp
            )
        )


        if (
            np.std(actual_kp) > 0
            and
            np.std(predicted_kp) > 0
        ):

            kp_corr = pearsonr(

                actual_kp,
                predicted_kp
            )[0]

        else:

            kp_corr = np.nan


        metrics["Kp10"] = {

            "rmse":
                kp_rmse,

            "mae":
                kp_mae,

            "r2":
                kp_r2,

            "correlation":
                kp_corr
        }


        # ----------------------------------------------------
        # Dst
        # ----------------------------------------------------

        dst_rmse = np.sqrt(

            mean_squared_error(
                actual_dst,
                predicted_dst
            )
        )

        dst_mae = (
            mean_absolute_error(
                actual_dst,
                predicted_dst
            )
        )

        dst_r2 = (
            r2_score(
                actual_dst,
                predicted_dst
            )
        )


        if (
            np.std(actual_dst) > 0
            and
            np.std(predicted_dst) > 0
        ):

            dst_corr = pearsonr(

                actual_dst,
                predicted_dst
            )[0]

        else:

            dst_corr = np.nan


        # ----------------------------------------------------
        # Prediction Efficiency
        # ----------------------------------------------------

        denominator = np.sum(
            actual_dst ** 2
        )


        if denominator != 0:

            dst_pe = (

                1
                -
                np.sum(
                    (
                        actual_dst
                        -
                        predicted_dst
                    ) ** 2
                )
                /
                denominator
            )

        else:

            dst_pe = np.nan


        metrics["Dst"] = {

            "rmse":
                dst_rmse,

            "mae":
                dst_mae,

            "r2":
                dst_r2,

            "correlation":
                dst_corr,

            "pe":
                dst_pe
        }


        # ====================================================
        # PRINT PERFORMANCE
        # ====================================================

        print(
            "\nKp × 10 Performance"
        )

        print(
            "-" * 50
        )

        print(
            f"RMSE        : {kp_rmse:.4f}"
        )

        print(
            f"MAE         : {kp_mae:.4f}"
        )

        print(
            f"R²          : {kp_r2:.4f}"
        )

        print(
            f"Correlation : {kp_corr:.4f}"
        )


        print(
            "\nDst Performance"
        )

        print(
            "-" * 50
        )

        print(
            f"RMSE        : {dst_rmse:.4f} nT"
        )

        print(
            f"MAE         : {dst_mae:.4f} nT"
        )

        print(
            f"R²          : {dst_r2:.4f}"
        )

        print(
            f"Correlation : {dst_corr:.4f}"
        )

        print(
            f"PE          : {dst_pe:.4f}"
        )


        return {

            "actual_kp":
                actual_kp,

            "predicted_kp":
                predicted_kp,

            "actual_dst":
                actual_dst,

            "predicted_dst":
                predicted_dst,

            "metrics":
                metrics
        }


# ============================================================
# 8. STORM CLASSIFICATION
# ============================================================

class StormClassifier:


    # ========================================================
    # Kp CLASSIFICATION
    # ========================================================

    @staticmethod
    def classify_kp(kp10):

        categories = []


        for value in kp10:

            if value < 50:

                categories.append(
                    "Quiet"
                )

            elif value < 60:

                categories.append(
                    "G1 Minor"
                )

            elif value < 70:

                categories.append(
                    "G2 Moderate"
                )

            elif value < 80:

                categories.append(
                    "G3 Strong"
                )

            elif value < 90:

                categories.append(
                    "G4 Severe"
                )

            else:

                categories.append(
                    "G5 Extreme"
                )


        return np.array(
            categories
        )


    # ========================================================
    # Dst CLASSIFICATION
    # ========================================================

    @staticmethod
    def classify_dst(dst):

        categories = []


        for value in dst:

            if value > -30:

                categories.append(
                    "Quiet"
                )

            elif value > -50:

                categories.append(
                    "Minor"
                )

            elif value > -100:

                categories.append(
                    "Moderate"
                )

            elif value > -200:

                categories.append(
                    "Intense"
                )

            else:

                categories.append(
                    "Super Storm"
                )


        return np.array(
            categories
        )


    # ========================================================
    # COMBINED CLASSIFICATION
    # ========================================================

    @staticmethod
    def classify_combined(
        kp10,
        dst
    ):

        kp_classes = (
            StormClassifier.classify_kp(
                kp10
            )
        )

        dst_classes = (
            StormClassifier.classify_dst(
                dst
            )
        )


        combined = []


        for kp_class, dst_class in zip(

            kp_classes,
            dst_classes
        ):

            if (

                kp_class == "G5 Extreme"
                or
                dst_class == "Super Storm"

            ):

                combined.append(
                    "Extreme/Super Storm"
                )


            elif (

                kp_class == "G4 Severe"
                or
                dst_class == "Intense"

            ):

                combined.append(
                    "Severe/Intense"
                )


            elif (

                kp_class == "G3 Strong"
                or
                dst_class == "Moderate"

            ):

                combined.append(
                    "Strong/Moderate"
                )


            elif (

                kp_class == "G2 Moderate"
                or
                dst_class == "Minor"

            ):

                combined.append(
                    "Moderate/Minor"
                )


            elif kp_class == "G1 Minor":

                combined.append(
                    "Minor"
                )


            else:

                combined.append(
                    "Quiet"
                )


        return np.array(
            combined
        )


    # ========================================================
    # BINARY STORM
    # ========================================================

    @staticmethod
    def binary_storm(
        kp10,
        dst
    ):

        # Storm:
        # Kp >= 5 OR Dst <= -50 nT

        return (

            (
                np.asarray(kp10)
                >=
                50
            )

            |

            (
                np.asarray(dst)
                <=
                -50
            )

        ).astype(int)


    # ========================================================
    # CLASSIFICATION EVALUATION
    # ========================================================

    @staticmethod
    def evaluate(

        actual_kp,
        predicted_kp,

        actual_dst,
        predicted_dst

    ):

        print("\n" + "=" * 80)
        print(
            "BiLSTM STORM CLASSIFICATION"
        )
        print("=" * 80)


        # ====================================================
        # Kp CLASSIFICATION
        # ====================================================

        actual_kp_class = (
            StormClassifier.classify_kp(
                actual_kp
            )
        )

        predicted_kp_class = (
            StormClassifier.classify_kp(
                predicted_kp
            )
        )


        kp_accuracy = (
            accuracy_score(
                actual_kp_class,
                predicted_kp_class
            )
        )


        kp_labels = [

            "Quiet",
            "G1 Minor",
            "G2 Moderate",
            "G3 Strong",
            "G4 Severe",
            "G5 Extreme"
        ]


        kp_cm = confusion_matrix(

            actual_kp_class,

            predicted_kp_class,

            labels=kp_labels
        )


        print(
            "\nKp × 10 Classification Accuracy:"
        )

        print(
            f"{kp_accuracy:.4f}"
        )


        print(
            "\nKp × 10 Classification Report:"
        )

        print(

            classification_report(

                actual_kp_class,

                predicted_kp_class,

                labels=kp_labels,

                zero_division=0
            )
        )


        # ====================================================
        # DST CLASSIFICATION
        # ====================================================

        actual_dst_class = (
            StormClassifier.classify_dst(
                actual_dst
            )
        )

        predicted_dst_class = (
            StormClassifier.classify_dst(
                predicted_dst
            )
        )


        dst_accuracy = (
            accuracy_score(
                actual_dst_class,
                predicted_dst_class
            )
        )


        dst_labels = [

            "Quiet",
            "Minor",
            "Moderate",
            "Intense",
            "Super Storm"
        ]


        dst_cm = confusion_matrix(

            actual_dst_class,

            predicted_dst_class,

            labels=dst_labels
        )


        print(
            "\nDst Classification Accuracy:"
        )

        print(
            f"{dst_accuracy:.4f}"
        )


        print(
            "\nDst Classification Report:"
        )

        print(

            classification_report(

                actual_dst_class,

                predicted_dst_class,

                labels=dst_labels,

                zero_division=0
            )
        )


        # ====================================================
        # BINARY CLASSIFICATION
        # ====================================================

        actual_binary = (
            StormClassifier.binary_storm(

                actual_kp,
                actual_dst
            )
        )


        predicted_binary = (
            StormClassifier.binary_storm(

                predicted_kp,
                predicted_dst
            )
        )


        binary_accuracy = (
            accuracy_score(
                actual_binary,
                predicted_binary
            )
        )


        binary_precision = (
            precision_score(

                actual_binary,

                predicted_binary,

                zero_division=0
            )
        )


        binary_recall = (
            recall_score(

                actual_binary,

                predicted_binary,

                zero_division=0
            )
        )


        binary_f1 = (
            f1_score(

                actual_binary,

                predicted_binary,

                zero_division=0
            )
        )


        binary_cm = confusion_matrix(

            actual_binary,

            predicted_binary
        )


        print(
            "\nCombined Kp × 10 + Dst Binary Classification"
        )

        print(
            "-" * 60
        )

        print(
            f"Accuracy  : {binary_accuracy:.4f}"
        )

        print(
            f"Precision : {binary_precision:.4f}"
        )

        print(
            f"Recall    : {binary_recall:.4f}"
        )

        print(
            f"F1 Score  : {binary_f1:.4f}"
        )


        return {

            "actual_kp_class":
                actual_kp_class,

            "predicted_kp_class":
                predicted_kp_class,

            "kp_cm":
                kp_cm,

            "actual_dst_class":
                actual_dst_class,

            "predicted_dst_class":
                predicted_dst_class,

            "dst_cm":
                dst_cm,

            "actual_binary":
                actual_binary,

            "predicted_binary":
                predicted_binary,

            "binary_cm":
                binary_cm,

            "kp_accuracy":
                kp_accuracy,

            "dst_accuracy":
                dst_accuracy,

            "binary_accuracy":
                binary_accuracy,

            "binary_precision":
                binary_precision,

            "binary_recall":
                binary_recall,

            "binary_f1":
                binary_f1
        }


# ============================================================
# 9. VISUALIZATION FUNCTIONS
# ============================================================

class Visualizer:


    # ========================================================
    # TRAINING HISTORY
    # ========================================================

    @staticmethod
    def plot_training_history(
        history
    ):

        # ----------------------------------------------------
        # Total loss
        # ----------------------------------------------------

        plt.figure(
            figsize=(12, 6)
        )

        plt.plot(

            history.history[
                "loss"
            ],

            label="Training Loss"
        )

        plt.plot(

            history.history[
                "val_loss"
            ],

            linestyle="--",

            label="Validation Loss"
        )

        plt.xlabel(
            "Epoch"
        )

        plt.ylabel(
            "Loss"
        )

        plt.title(
            "BiLSTM Training and Validation Loss"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_training_validation_loss.png"
            ),

            dpi=300
        )

        plt.show()


        # ----------------------------------------------------
        # Kp10 MAE
        # ----------------------------------------------------

        if "Kp10_mae" in history.history:

            plt.figure(
                figsize=(12, 6)
            )

            plt.plot(

                history.history[
                    "Kp10_mae"
                ],

                label="Training Kp × 10 MAE"
            )

            plt.plot(

                history.history[
                    "val_Kp10_mae"
                ],

                linestyle="--",

                label="Validation Kp × 10 MAE"
            )

            plt.xlabel(
                "Epoch"
            )

            plt.ylabel(
                "MAE"
            )

            plt.title(
                "BiLSTM Kp × 10 Training and Validation MAE"
            )

            plt.legend()

            plt.grid(
                True,
                alpha=0.3
            )

            plt.tight_layout()


            plt.savefig(

                os.path.join(

                    OUTPUT_DIR,

                    "BiLSTM_Kp10_MAE.png"
                ),

                dpi=300
            )

            plt.show()


        # ----------------------------------------------------
        # Dst MAE
        # ----------------------------------------------------

        if "Dst_mae" in history.history:

            plt.figure(
                figsize=(12, 6)
            )

            plt.plot(

                history.history[
                    "Dst_mae"
                ],

                label="Training Dst MAE"
            )

            plt.plot(

                history.history[
                    "val_Dst_mae"
                ],

                linestyle="--",

                label="Validation Dst MAE"
            )

            plt.xlabel(
                "Epoch"
            )

            plt.ylabel(
                "MAE (nT)"
            )

            plt.title(
                "BiLSTM Dst Training and Validation MAE"
            )

            plt.legend()

            plt.grid(
                True,
                alpha=0.3
            )

            plt.tight_layout()


            plt.savefig(

                os.path.join(

                    OUTPUT_DIR,

                    "BiLSTM_Dst_MAE.png"
                ),

                dpi=300
            )

            plt.show()


    # ========================================================
    # DST PREDICTION
    # ========================================================

    @staticmethod
    def plot_dst_prediction(
        result,
        sample_size=2000
    ):

        actual = (
            result[
                "actual_dst"
            ][:sample_size]
        )

        predicted = (
            result[
                "predicted_dst"
            ][:sample_size]
        )


        plt.figure(
            figsize=(15, 6)
        )


        plt.plot(

            actual,

            label="Actual Dst",

            linewidth=1
        )


        plt.plot(

            predicted,

            label="Predicted Dst",

            linewidth=1
        )


        plt.axhline(

            -30,

            linestyle="--",

            label="Minor: -30 nT"
        )


        plt.axhline(

            -50,

            linestyle="--",

            label="Moderate: -50 nT"
        )


        plt.axhline(

            -100,

            linestyle="--",

            label="Intense: -100 nT"
        )


        plt.axhline(

            -200,

            linestyle="--",

            label="Super Storm: -200 nT"
        )


        plt.xlabel(
            "Time Step"
        )

        plt.ylabel(
            "Dst (nT)"
        )

        plt.title(
            "BiLSTM - Dst Prediction"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_prediction.png"
            ),

            dpi=300
        )

        plt.show()


    # ========================================================
    # KP PREDICTION
    # ========================================================

    @staticmethod
    def plot_kp_prediction(
        result,
        sample_size=2000
    ):

        actual = (
            result[
                "actual_kp"
            ][:sample_size]
        )

        predicted = (
            result[
                "predicted_kp"
            ][:sample_size]
        )


        plt.figure(
            figsize=(15, 6)
        )


        plt.plot(

            actual,

            label="Actual Kp × 10",

            linewidth=1
        )


        plt.plot(

            predicted,

            label="Predicted Kp × 10",

            linewidth=1
        )


        # ----------------------------------------------------
        # Kp thresholds
        # ----------------------------------------------------

        plt.axhline(

            50,

            linestyle="--",

            label="Kp = 5 (G1)"
        )


        plt.axhline(

            60,

            linestyle="--",

            label="Kp = 6 (G2)"
        )


        plt.axhline(

            70,

            linestyle="--",

            label="Kp = 7 (G3)"
        )


        plt.axhline(

            80,

            linestyle="--",

            label="Kp = 8 (G4)"
        )


        plt.axhline(

            90,

            linestyle="--",

            label="Kp = 9 (G5)"
        )


        plt.xlabel(
            "Time Step"
        )

        plt.ylabel(
            "Kp × 10"
        )

        plt.title(
            "BiLSTM - Kp × 10 Prediction"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Kp10_prediction.png"
            ),

            dpi=300
        )

        plt.show()


    # ========================================================
    # SCATTER PLOTS
    # ========================================================

    @staticmethod
    def plot_scatter(
        result
    ):

        # ----------------------------------------------------
        # Dst scatter
        # ----------------------------------------------------

        plt.figure(
            figsize=(7, 7)
        )


        plt.scatter(

            result["actual_dst"],

            result["predicted_dst"],

            alpha=0.4,

            s=10
        )


        minimum = min(

            result[
                "actual_dst"
            ].min(),

            result[
                "predicted_dst"
            ].min()
        )


        maximum = max(

            result[
                "actual_dst"
            ].max(),

            result[
                "predicted_dst"
            ].max()
        )


        plt.plot(

            [
                minimum,
                maximum
            ],

            [
                minimum,
                maximum
            ],

            linestyle="--",

            linewidth=2
        )


        plt.xlabel(
            "Actual Dst (nT)"
        )

        plt.ylabel(
            "Predicted Dst (nT)"
        )

        plt.title(

            "BiLSTM - Dst Prediction\n"
            f"R² = "
            f"{result['metrics']['Dst']['r2']:.4f}"
        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_scatter.png"
            ),

            dpi=300
        )

        plt.show()


        # ----------------------------------------------------
        # Kp scatter
        # ----------------------------------------------------

        plt.figure(
            figsize=(7, 7)
        )


        plt.scatter(

            result["actual_kp"],

            result["predicted_kp"],

            alpha=0.4,

            s=10
        )


        minimum = min(

            result[
                "actual_kp"
            ].min(),

            result[
                "predicted_kp"
            ].min()
        )


        maximum = max(

            result[
                "actual_kp"
            ].max(),

            result[
                "predicted_kp"
            ].max()
        )


        plt.plot(

            [
                minimum,
                maximum
            ],

            [
                minimum,
                maximum
            ],

            linestyle="--",

            linewidth=2
        )


        plt.xlabel(
            "Actual Kp × 10"
        )

        plt.ylabel(
            "Predicted Kp × 10"
        )

        plt.title(

            "BiLSTM - Kp × 10 Prediction\n"
            f"R² = "
            f"{result['metrics']['Kp10']['r2']:.4f}"
        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Kp10_scatter.png"
            ),

            dpi=300
        )

        plt.show()


    # ========================================================
    # RESIDUALS
    # ========================================================

    @staticmethod
    def plot_residuals(
        result
    ):

        # ----------------------------------------------------
        # Dst residuals
        # ----------------------------------------------------

        residuals = (

            result["actual_dst"]

            -

            result["predicted_dst"]
        )


        plt.figure(
            figsize=(10, 6)
        )


        plt.scatter(

            result["predicted_dst"],

            residuals,

            alpha=0.4,

            s=10
        )


        plt.axhline(

            0,

            linestyle="--",

            linewidth=2
        )


        plt.xlabel(
            "Predicted Dst (nT)"
        )

        plt.ylabel(
            "Residual (nT)"
        )

        plt.title(
            "BiLSTM - Dst Residual Analysis"
        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_residuals.png"
            ),

            dpi=300
        )

        plt.show()


        # ----------------------------------------------------
        # Residual histogram
        # ----------------------------------------------------

        plt.figure(
            figsize=(10, 6)
        )


        plt.hist(

            residuals,

            bins=50
        )


        plt.xlabel(
            "Residual (nT)"
        )

        plt.ylabel(
            "Frequency"
        )

        plt.title(
            "BiLSTM - Dst Residual Distribution"
        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_residual_histogram.png"
            ),

            dpi=300
        )

        plt.show()


    # ========================================================
    # CONFUSION MATRIX
    # ========================================================

    @staticmethod
    def plot_confusion_matrix(

        cm,
        labels,
        title,
        filename
    ):

        plt.figure(
            figsize=(9, 7)
        )


        sns.heatmap(

            cm,

            annot=True,

            fmt="d",

            cmap="Blues",

            xticklabels=labels,

            yticklabels=labels
        )


        plt.xlabel(
            "Predicted"
        )

        plt.ylabel(
            "Actual"
        )

        plt.title(
            title
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                filename
            ),

            dpi=300
        )

        plt.show()


# ============================================================
# 10. ROC AND PRECISION-RECALL
# ============================================================

def add_binary_classification(
    result
):

    # --------------------------------------------------------
    # Actual binary storm
    # --------------------------------------------------------

    actual_binary = (

        StormClassifier.binary_storm(

            result["actual_kp"],

            result["actual_dst"]
        )
    )


    # --------------------------------------------------------
    # Continuous predicted storm score
    #
    # Kp score:
    #     predicted Kp10 / 50
    #
    # Dst score:
    #     -predicted Dst / 50
    #
    # Maximum of both is used as the storm score.
    # --------------------------------------------------------

    kp_score = (

        result["predicted_kp"]

        /

        50.0
    )


    dst_score = (

        -result["predicted_dst"]

        /

        50.0
    )


    storm_score = np.maximum(

        kp_score,

        dst_score
    )


    result[
        "actual_binary"
    ] = actual_binary


    result[
        "predicted_binary_score"
    ] = storm_score


# ============================================================
# ROC CURVE
# ============================================================

def plot_roc_curve(
    result
):

    actual = (
        result[
            "actual_binary"
        ]
    )

    score = (
        result[
            "predicted_binary_score"
        ]
    )


    fpr, tpr, thresholds = (
        roc_curve(
            actual,
            score
        )
    )


    roc_auc = auc(
        fpr,
        tpr
    )


    plt.figure(
        figsize=(9, 7)
    )


    plt.plot(

        fpr,

        tpr,

        linewidth=2,

        label=
        f"BiLSTM AUC = {roc_auc:.4f}"
    )


    plt.plot(

        [0, 1],

        [0, 1],

        linestyle="--",

        label="Random classifier"
    )


    plt.xlabel(
        "False Positive Rate"
    )

    plt.ylabel(
        "True Positive Rate"
    )

    plt.title(
        "ROC Curve - BiLSTM Geomagnetic Storm Prediction"
    )

    plt.legend()

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()


    plt.savefig(

        os.path.join(

            OUTPUT_DIR,

            "BiLSTM_ROC_curve.png"
        ),

        dpi=300
    )

    plt.show()


    print(
        f"\nBiLSTM ROC-AUC: {roc_auc:.4f}"
    )


    return roc_auc


# ============================================================
# PRECISION-RECALL CURVE
# ============================================================

def plot_precision_recall_curve_bilstm(
    result
):

    actual = (
        result[
            "actual_binary"
        ]
    )

    score = (
        result[
            "predicted_binary_score"
        ]
    )


    precision, recall, thresholds = (

        precision_recall_curve(

            actual,

            score
        )
    )


    average_precision = (

        average_precision_score(

            actual,

            score
        )
    )


    plt.figure(
        figsize=(9, 7)
    )


    plt.plot(

        recall,

        precision,

        linewidth=2,

        label=
        f"BiLSTM AP = "
        f"{average_precision:.4f}"
    )


    plt.xlabel(
        "Recall"
    )

    plt.ylabel(
        "Precision"
    )

    plt.title(
        "Precision-Recall Curve - BiLSTM"
    )

    plt.legend()

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()


    plt.savefig(

        os.path.join(

            OUTPUT_DIR,

            "BiLSTM_PR_curve.png"
        ),

        dpi=300
    )

    plt.show()


    print(
        f"BiLSTM Average Precision: "
        f"{average_precision:.4f}"
    )


    return average_precision


# ============================================================
# 11. SAVE ALL PREDICTIONS
# ============================================================

def save_prediction_results(
    result
):

    output_df = pd.DataFrame({

        "Actual_Kp10":
            result["actual_kp"],

        "Predicted_Kp10":
            result["predicted_kp"],

        "Actual_Dst":
            result["actual_dst"],

        "Predicted_Dst":
            result["predicted_dst"]
    })


    # --------------------------------------------------------
    # Kp classes
    # --------------------------------------------------------

    output_df[
        "Actual_Kp_Class"
    ] = StormClassifier.classify_kp(

        result["actual_kp"]
    )


    output_df[
        "Predicted_Kp_Class"
    ] = StormClassifier.classify_kp(

        result["predicted_kp"]
    )


    # --------------------------------------------------------
    # Dst classes
    # --------------------------------------------------------

    output_df[
        "Actual_Dst_Class"
    ] = StormClassifier.classify_dst(

        result["actual_dst"]
    )


    output_df[
        "Predicted_Dst_Class"
    ] = StormClassifier.classify_dst(

        result["predicted_dst"]
    )


    # --------------------------------------------------------
    # Combined classification
    # --------------------------------------------------------

    output_df[
        "Actual_Combined_Class"
    ] = StormClassifier.classify_combined(

        result["actual_kp"],

        result["actual_dst"]
    )


    output_df[
        "Predicted_Combined_Class"
    ] = StormClassifier.classify_combined(

        result["predicted_kp"],

        result["predicted_dst"]
    )


    # --------------------------------------------------------
    # Binary storm
    # --------------------------------------------------------

    output_df[
        "Actual_Storm"
    ] = StormClassifier.binary_storm(

        result["actual_kp"],

        result["actual_dst"]
    )


    output_df[
        "Predicted_Storm"
    ] = StormClassifier.binary_storm(

        result["predicted_kp"],

        result["predicted_dst"]
    )


    # --------------------------------------------------------
    # Continuous storm score
    # --------------------------------------------------------

    output_df[
        "Predicted_Storm_Score"
    ] = result[
        "predicted_binary_score"
    ]


    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    filepath = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_predictions.csv"
    )


    output_df.to_csv(

        filepath,

        index=False
    )


    print(
        "\nPrediction results saved:"
    )

    print(
        filepath
    )


    return output_df


# ============================================================
# 12. SAVE PERFORMANCE METRICS
# ============================================================

def save_metrics(
    result,
    classification_result,
    roc_auc,
    average_precision
):

    kp = (
        result[
            "metrics"
        ]["Kp10"]
    )

    dst = (
        result[
            "metrics"
        ]["Dst"]
    )


    metrics_df = pd.DataFrame({

        "Model": [
            "BiLSTM"
        ],

        "Kp10_RMSE": [
            kp["rmse"]
        ],

        "Kp10_MAE": [
            kp["mae"]
        ],

        "Kp10_R2": [
            kp["r2"]
        ],

        "Kp10_Correlation": [
            kp["correlation"]
        ],

        "Dst_RMSE_nT": [
            dst["rmse"]
        ],

        "Dst_MAE_nT": [
            dst["mae"]
        ],

        "Dst_R2": [
            dst["r2"]
        ],

        "Dst_Correlation": [
            dst["correlation"]
        ],

        "Dst_PE": [
            dst["pe"]
        ],

        "Kp10_Accuracy": [
            classification_result[
                "kp_accuracy"
            ]
        ],

        "Dst_Accuracy": [
            classification_result[
                "dst_accuracy"
            ]
        ],

        "Combined_Accuracy": [
            classification_result[
                "binary_accuracy"
            ]
        ],

        "Combined_Precision": [
            classification_result[
                "binary_precision"
            ]
        ],

        "Combined_Recall": [
            classification_result[
                "binary_recall"
            ]
        ],

        "Combined_F1": [
            classification_result[
                "binary_f1"
            ]
        ],

        "ROC_AUC": [
            roc_auc
        ],

        "Average_Precision": [
            average_precision
        ]
    })


    filepath = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_performance_metrics.csv"
    )


    metrics_df.to_csv(

        filepath,

        index=False
    )


    print(
        "\nPerformance metrics saved:"
    )

    print(
        filepath
    )


    return metrics_df


# ============================================================
# 13. MAIN FUNCTION
# ============================================================

def main():

    print("\n")

    print("=" * 80)

    print(
        "GEOMAGNETIC STORM PREDICTION USING BiLSTM"
    )

    print(
        "Targets: Kp × 10 + Dst"
    )

    print("=" * 80)


    # ========================================================
    # STEP 1
    # LOAD DATA
    # ========================================================

    preprocessor = DataPreprocessor(

        DATA_PATH
    )


    preprocessor.load_data()

    preprocessor.check_columns()

    preprocessor.handle_missing_values()

    preprocessor.clean_data()


    df = (
        preprocessor.get_processed_data()
    )


    # --------------------------------------------------------
    # Save cleaned data
    # --------------------------------------------------------

    cleaned_path = os.path.join(

        OUTPUT_DIR,

        "cleaned_geomagnetic_data.csv"
    )


    df.to_csv(

        cleaned_path,

        index=False
    )


    print(
        "\nCleaned data saved:"
    )

    print(
        cleaned_path
    )


    # ========================================================
    # STEP 2
    # CREATE SEQUENCES
    # ========================================================

    seq_gen = SequenceGenerator(

        df,

        lookback=LOOKBACK
    )


    X, y = (
        seq_gen.create_sequences()
    )


    # ========================================================
    # STEP 3
    # SPLIT DATA
    # ========================================================

    (

        X_train,
        X_val,
        X_test,

        y_train,
        y_val,
        y_test

    ) = seq_gen.split_data(

        X,

        y,

        train_size=TRAIN_SIZE,

        validation_size=VALIDATION_SIZE,

        test_size=TEST_SIZE
    )


    # ========================================================
    # STEP 4
    # INPUT SHAPE
    # ========================================================

    input_shape = (

        X_train.shape[1],

        X_train.shape[2]
    )


    print(
        "\nBiLSTM input shape:"
    )

    print(
        input_shape
    )


    # ========================================================
    # STEP 5
    # BUILD BiLSTM
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "BUILDING BiLSTM MODEL"
    )

    print("=" * 80)


    model = build_bilstm(

        input_shape=input_shape,

        units=64,

        dropout=0.20
    )


    # ========================================================
    # STEP 6
    # COMPILE
    # ========================================================

    trainer = BiLSTMTrainer(

        model,

        seq_gen.scaler,

        seq_gen.all_columns
    )


    trainer.compile_model(

        learning_rate=LEARNING_RATE
    )


    # ========================================================
    # STEP 7
    # MODEL SUMMARY
    # ========================================================

    print("\n")

    model.summary()


    # ========================================================
    # SAVE MODEL SUMMARY
    # ========================================================

    summary_path = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_model_summary.txt"
    )


    with open(

        summary_path,

        "w"
    ) as f:

        model.summary(
            print_fn=f.write
        )


    print(
        "\nModel summary saved:"
    )

    print(
        summary_path
    )


    # ========================================================
    # STEP 8
    # TRAIN
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "TRAINING BiLSTM"
    )

    print("=" * 80)


    history = trainer.train(

        X_train,

        y_train,

        X_val,

        y_val,

        epochs=EPOCHS,

        batch_size=BATCH_SIZE
    )


    # ========================================================
    # STEP 9
    # EVALUATE REGRESSION
    # ========================================================

    result = trainer.evaluate(

        X_test,

        y_test
    )


    # ========================================================
    # STEP 10
    # STORM CLASSIFICATION
    # ========================================================

    classification_result = (

        StormClassifier.evaluate(

            result["actual_kp"],

            result["predicted_kp"],

            result["actual_dst"],

            result["predicted_dst"]
        )
    )


    # ========================================================
    # STEP 11
    # ADD BINARY STORM SCORE
    # ========================================================

    add_binary_classification(

        result
    )


    # ========================================================
    # STEP 12
    # TRAINING PLOTS
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "GENERATING TRAINING OUTPUTS"
    )

    print("=" * 80)


    Visualizer.plot_training_history(

        history
    )


    # ========================================================
    # STEP 13
    # DST PREDICTION
    # ========================================================

    Visualizer.plot_dst_prediction(

        result,

        sample_size=2000
    )


    # ========================================================
    # STEP 14
    # KP PREDICTION
    # ========================================================

    Visualizer.plot_kp_prediction(

        result,

        sample_size=2000
    )


    # ========================================================
    # STEP 15
    # SCATTER PLOTS
    # ========================================================

    Visualizer.plot_scatter(

        result
    )


    # ========================================================
    # STEP 16
    # RESIDUAL ANALYSIS
    # ========================================================

    Visualizer.plot_residuals(

        result
    )


    # ========================================================
    # STEP 17
    # CONFUSION MATRICES
    # ========================================================

    Visualizer.plot_confusion_matrix(

        classification_result[
            "kp_cm"
        ],

        [

            "Quiet",
            "G1 Minor",
            "G2 Moderate",
            "G3 Strong",
            "G4 Severe",
            "G5 Extreme"

        ],

        "BiLSTM - Kp × 10 Storm Classification",

        "BiLSTM_Kp10_confusion_matrix.png"
    )


    Visualizer.plot_confusion_matrix(

        classification_result[
            "dst_cm"
        ],

        [

            "Quiet",
            "Minor",
            "Moderate",
            "Intense",
            "Super Storm"

        ],

        "BiLSTM - Dst Storm Classification",

        "BiLSTM_Dst_confusion_matrix.png"
    )


    Visualizer.plot_confusion_matrix(

        classification_result[
            "binary_cm"
        ],

        [

            "No Storm",
            "Storm"

        ],

        "BiLSTM - Combined Kp × 10 + Dst",

        "BiLSTM_Combined_confusion_matrix.png"
    )


    # ========================================================
    # STEP 18
    # ROC CURVE
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "GENERATING ROC CURVE"
    )

    print("=" * 80)


    roc_auc = plot_roc_curve(

        result
    )


    # ========================================================
    # STEP 19
    # PRECISION-RECALL CURVE
    # ========================================================

    print("\n" + "=" * 80)

    print(
        "GENERATING PRECISION-RECALL CURVE"
    )

    print("=" * 80)


    average_precision = (
        plot_precision_recall_curve_bilstm(

            result
        )
    )


    # ========================================================
    # STEP 20
    # SAVE PREDICTIONS
    # ========================================================

    prediction_df = (
        save_prediction_results(

            result
        )
    )


    # ========================================================
    # STEP 21
    # SAVE METRICS
    # ========================================================

    metrics_df = save_metrics(

        result,

        classification_result,

        roc_auc,

        average_precision
    )


    # ========================================================
    # STEP 22
    # PRINT FINAL RESULT TABLE
    # ========================================================

    print("\n")

    print("=" * 110)

    print(
        "BiLSTM FINAL PERFORMANCE"
    )

    print("=" * 110)


    print(
        metrics_df.to_string(
            index=False
        )
    )


    # ========================================================
    # STEP 23
    # FINAL SUMMARY
    # ========================================================

    print("\n")

    print("=" * 80)

    print(
        "FINAL BiLSTM RESULTS"
    )

    print("=" * 80)


    print(
        "\nRegression Results:"
    )

    print(
        f"Kp × 10 RMSE = "
        f"{result['metrics']['Kp10']['rmse']:.4f}"
    )

    print(
        f"Kp × 10 MAE  = "
        f"{result['metrics']['Kp10']['mae']:.4f}"
    )

    print(
        f"Kp × 10 R²   = "
        f"{result['metrics']['Kp10']['r2']:.4f}"
    )

    print(
        f"Kp × 10 r    = "
        f"{result['metrics']['Kp10']['correlation']:.4f}"
    )


    print(
        f"\nDst RMSE      = "
        f"{result['metrics']['Dst']['rmse']:.4f} nT"
    )

    print(
        f"Dst MAE       = "
        f"{result['metrics']['Dst']['mae']:.4f} nT"
    )

    print(
        f"Dst R²        = "
        f"{result['metrics']['Dst']['r2']:.4f}"
    )

    print(
        f"Dst r         = "
        f"{result['metrics']['Dst']['correlation']:.4f}"
    )

    print(
        f"Dst PE        = "
        f"{result['metrics']['Dst']['pe']:.4f}"
    )


    print(
        "\nClassification Results:"
    )

    print(
        f"Kp × 10 Accuracy = "
        f"{classification_result['kp_accuracy']:.4f}"
    )

    print(
        f"Dst Accuracy     = "
        f"{classification_result['dst_accuracy']:.4f}"
    )

    print(
        f"Combined Accuracy = "
        f"{classification_result['binary_accuracy']:.4f}"
    )

    print(
        f"Combined Precision = "
        f"{classification_result['binary_precision']:.4f}"
    )

    print(
        f"Combined Recall = "
        f"{classification_result['binary_recall']:.4f}"
    )

    print(
        f"Combined F1 = "
        f"{classification_result['binary_f1']:.4f}"
    )


    print(
        "\nROC/PR Results:"
    )

    print(
        f"ROC-AUC = {roc_auc:.4f}"
    )

    print(
        f"Average Precision = "
        f"{average_precision:.4f}"
    )


    print(
        "\nOutput directory:"
    )

    print(
        OUTPUT_DIR
    )


    print("\n")

    print("=" * 80)

    print(
        "BiLSTM ANALYSIS COMPLETED SUCCESSFULLY"
    )

    print("=" * 80)


    return (

        model,

        history,

        result,

        classification_result,

        metrics_df,

        prediction_df
    )


# ============================================================
# 14. RUN PROGRAM
# ============================================================

if __name__ == "__main__":

    (

        model,

        history,

        result,

        classification_result,

        metrics_df,

        prediction_df

    ) = main()
```

```python
# ============================================================
# GEOMAGNETIC STORM PREDICTION USING BiLSTM
# ============================================================
# Author: Prajwal Bhandari
# Date: September 2026
#
# MODEL:
#   Bidirectional LSTM (BiLSTM)
#
# TARGETS:
#   1. Dst
#   2. Kp × 10
#
# INPUT FEATURES:
#   YEAR, DOY, HR, B, Bx, By, Bz, PD, FP, E
#
# DATA SPLIT:
#   Training   = 70%
#   Validation = 15%
#   Testing    = 15%
#
# LOOKBACK:
#   48 hours
#
# OPTIMIZER:
#   Adam
#
# UNCERTAINTY:
#   Monte Carlo Dropout
#   95% Prediction Interval
#
# OUTPUTS:
#   1. Model summary
#   2. Training/validation loss
#   3. Dst prediction
#   4. Kp × 10 prediction
#   5. Dst 95% interval
#   6. Kp × 10 95% interval
#   7. Scatter plots
#   8. Residual plots
#   9. Confusion matrices
#  10. ROC curve
#  11. Precision-Recall curve
#  12. Classification report
#  13. Regression metrics
#  14. Confidence interval coverage
#  15. Prediction CSV
# ============================================================


# ============================================================
# 0. INSTALL / IMPORT LIBRARIES
# ============================================================

# Google Colab:
# !pip install -q tensorflow scikit-learn scipy seaborn

import os
import re
import random
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr

from sklearn.preprocessing import MinMaxScaler

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_curve,
    auc,
    precision_recall_curve,
    average_precision_score
)

import tensorflow as tf

from tensorflow.keras.models import Model

from tensorflow.keras.layers import (
    Input,
    LSTM,
    Dense,
    Dropout,
    Bidirectional
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

from tensorflow.keras.optimizers import Adam


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)


# ============================================================
# 2. CONFIGURATION
# ============================================================

# ------------------------------------------------------------
# DATA PATH
# ------------------------------------------------------------

DATA_PATH = "/content/Sun.csv"

# If using Google Drive:
# DATA_PATH = "/content/drive/MyDrive/D.csv"


# ------------------------------------------------------------
# OUTPUT DIRECTORY
# ------------------------------------------------------------

OUTPUT_DIR = "/content/BiLSTM_geomagnetic_results"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ------------------------------------------------------------
# TIME-SERIES CONFIGURATION
# ------------------------------------------------------------

LOOKBACK = 72

TEST_SIZE = 0.15

VALIDATION_SIZE = 0.15

TRAIN_SIZE = 0.70


# ------------------------------------------------------------
# TRAINING CONFIGURATION
# ------------------------------------------------------------

EPOCHS = 100

BATCH_SIZE = 32

LEARNING_RATE = 0.001


# ------------------------------------------------------------
# EARLY STOPPING
# ------------------------------------------------------------

PATIENCE = 10


# ------------------------------------------------------------
# MONTE CARLO DROPOUT
# ------------------------------------------------------------
# Number of stochastic forward passes
#
# Higher values give more stable uncertainty estimates.
#
# 100 = good for research/testing
# 200–500 = more stable but slower
# ------------------------------------------------------------

MC_SAMPLES = 100

CONFIDENCE_LEVEL = 0.95


# ------------------------------------------------------------
# PREDICTION PLOT SIZE
# ------------------------------------------------------------

PLOT_SAMPLES = 2000


# ============================================================
# 3. DATA PREPROCESSOR
# ============================================================

class DataPreprocessor:

    def __init__(self, filepath):

        self.filepath = filepath

        self.df = None

        self.missing_values_map = {

            "B": [
                999.99,
                9999.99
            ],

            "Bx": [
                999.99,
                9999.99
            ],

            "By": [
                999.99,
                9999.99
            ],

            "Bz": [
                999.99,
                9999.99
            ],

            "PD": [
                999.9,
                999.99,
                9999.9
            ],

            "FP": [
                99.99,
                999.99
            ],

            "E": [
                999.99,
                9999.99
            ],

            "Kp": [
                9.99,
                99.99
            ],

            "Dst": [
                9999.9,
                9999.99
            ]
        }


    # ========================================================
    # NORMALIZE COLUMN NAMES
    # ========================================================

    def normalize_column_names(self):

        new_columns = []

        for col in self.df.columns:

            col_clean = str(col).strip()

            col_clean = re.sub(
                r"\s+",
                " ",
                col_clean
            )

            if "kp" in col_clean.lower():

                col_clean = "Kp"

            elif col_clean.lower() == "dst":

                col_clean = "Dst"

            new_columns.append(
                col_clean
            )

        self.df.columns = new_columns

        return self


    # ========================================================
    # LOAD DATA
    # ========================================================

    def load_data(self):

        print("\n" + "=" * 80)

        print(
            "LOADING GEOMAGNETIC DATA"
        )

        print("=" * 80)

        self.df = pd.read_csv(
            self.filepath
        )

        print(
            "\nOriginal dataset shape:",
            self.df.shape
        )

        print(
            "\nOriginal columns:"
        )

        print(
            list(self.df.columns)
        )

        self.normalize_column_names()

        print(
            "\nNormalized columns:"
        )

        print(
            list(self.df.columns)
        )

        return self


    # ========================================================
    # CHECK REQUIRED COLUMNS
    # ========================================================

    def check_columns(self):

        required_columns = [

            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E",
            "Kp",
            "Dst"
        ]

        missing = [

            col
            for col in required_columns
            if col not in self.df.columns

        ]

        if len(missing) > 0:

            raise ValueError(

                "\nMissing required columns:\n"
                + str(missing)
                + "\n\nAvailable columns:\n"
                + str(list(self.df.columns))
            )

        print(
            "\nAll required columns found."
        )

        return self


    # ========================================================
    # HANDLE MISSING VALUES
    # ========================================================

    def handle_missing_values(self):

        print(
            "\n" + "=" * 80
        )

        print(
            "HANDLING MISSING VALUES"
        )

        print(
            "=" * 80
        )

        for col, values in (
            self.missing_values_map.items()
        ):

            if col in self.df.columns:

                for value in values:

                    count = (
                        self.df[col] == value
                    ).sum()

                    if count > 0:

                        self.df[col] = (
                            self.df[col]
                            .replace(
                                value,
                                np.nan
                            )
                        )

                        print(
                            f"{col}: replaced "
                            f"{count} values of {value}"
                        )

        return self


    # ========================================================
    # CLEAN DATA
    # ========================================================

    def clean_data(self):

        print(
            "\n" + "=" * 80
        )

        print(
            "CLEANING DATA"
        )

        print(
            "=" * 80
        )


        # Convert columns to numeric

        for col in self.df.columns:

            self.df[col] = pd.to_numeric(
                self.df[col],
                errors="coerce"
            )


        # Chronological ordering

        self.df = self.df.sort_values(

            [
                "YEAR",
                "DOY",
                "HR"
            ]

        ).reset_index(
            drop=True
        )


        # Missing values

        print(
            "\nMissing values before interpolation:"
        )

        missing = self.df.isna().sum()

        print(
            missing[missing > 0]
        )


        # Linear interpolation

        numeric_columns = (
            self.df.select_dtypes(
                include=[np.number]
            ).columns
        )


        self.df[numeric_columns] = (

            self.df[numeric_columns]

            .interpolate(
                method="linear"
            )

            .ffill()

            .bfill()
        )


        # Remove remaining NaN

        before = len(
            self.df
        )

        self.df = (
            self.df.dropna()
        )

        after = len(
            self.df
        )


        print(
            f"\nRows removed: "
            f"{before - after}"
        )

        print(
            f"Final dataset shape: "
            f"{self.df.shape}"
        )


        # ====================================================
        # CREATE Kp × 10
        # ====================================================

        kp_max = self.df["Kp"].max()

        if kp_max <= 9.0:

            print(
                "\nKp appears to be on 0–9 scale."
            )

            print(
                "Creating Kp10 = Kp × 10."
            )

            self.df["Kp10"] = (
                self.df["Kp"] * 10.0
            )

        else:

            print(
                "\nKp appears to already be Kp × 10."
            )

            self.df["Kp10"] = (
                self.df["Kp"]
            )


        # Remove original Kp

        self.df.drop(
            columns=["Kp"],
            inplace=True
        )


        # ====================================================
        # FINAL COLUMN ORDER
        # ====================================================

        final_columns = [

            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E",
            "Kp10",
            "Dst"
        ]


        self.df = self.df[
            final_columns
        ]


        print(
            "\nFinal columns:"
        )

        print(
            list(self.df.columns)
        )


        print(
            "\nKp × 10 statistics:"
        )

        print(
            self.df["Kp10"].describe()
        )


        print(
            "\nDst statistics:"
        )

        print(
            self.df["Dst"].describe()
        )


        return self


    def get_processed_data(self):

        return self.df


# ============================================================
# 4. SEQUENCE GENERATOR
# ============================================================

class SequenceGenerator:

    def __init__(
        self,
        data,
        lookback=48
    ):

        self.data = data

        self.lookback = lookback

        self.scaler = MinMaxScaler()


        self.feature_columns = [

            "YEAR",
            "DOY",
            "HR",
            "B",
            "Bx",
            "By",
            "Bz",
            "PD",
            "FP",
            "E"
        ]


        self.target_columns = [

            "Kp10",
            "Dst"
        ]


        self.all_columns = (

            self.feature_columns
            +
            self.target_columns

        )


    # ========================================================
    # CREATE SEQUENCES
    # ========================================================

    def create_sequences(self):

        print(
            "\n" + "=" * 80
        )

        print(
            "CREATING TIME-SERIES SEQUENCES"
        )

        print(
            "=" * 80
        )


        data_values = (
            self.data[
                self.all_columns
            ].values
        )


        # Scale

        scaled_data = (
            self.scaler.fit_transform(
                data_values
            )
        )


        kp_index = (
            self.all_columns.index(
                "Kp10"
            )
        )


        dst_index = (
            self.all_columns.index(
                "Dst"
            )
        )


        X = []

        y_kp = []

        y_dst = []


        for i in range(

            len(scaled_data)
            -
            self.lookback

        ):

            X.append(

                scaled_data[
                    i:
                    i + self.lookback
                ]

            )


            y_kp.append(

                scaled_data[
                    i + self.lookback,
                    kp_index
                ]

            )


            y_dst.append(

                scaled_data[
                    i + self.lookback,
                    dst_index
                ]

            )


        X = np.array(X)

        y_kp = np.array(y_kp)

        y_dst = np.array(y_dst)


        y = np.column_stack(

            [
                y_kp,
                y_dst
            ]

        )


        print(
            "\nX shape:",
            X.shape
        )

        print(
            "Y shape:",
            y.shape
        )

        print(
            "Lookback:",
            self.lookback,
            "hours"
        )

        print(
            "Input features:",
            X.shape[2]
        )


        return X, y


    # ========================================================
    # CHRONOLOGICAL SPLIT
    # ========================================================

    def split_data(

        self,
        X,
        y,
        test_size=0.15,
        val_size=0.15

    ):

        print(
            "\n" + "=" * 80
        )

        print(
            "CHRONOLOGICAL DATA SPLIT"
        )

        print(
            "=" * 80
        )


        n = len(X)


        train_end = int(
            n * 0.70
        )


        val_end = int(
            n * 0.85
        )


        X_train = X[
            :train_end
        ]

        y_train = y[
            :train_end
        ]


        X_val = X[
            train_end:val_end
        ]

        y_val = y[
            train_end:val_end
        ]


        X_test = X[
            val_end:
        ]

        y_test = y[
            val_end:
        ]


        print(
            f"Training samples:   "
            f"{len(X_train):,}"
        )

        print(
            f"Validation samples: "
            f"{len(X_val):,}"
        )

        print(
            f"Testing samples:    "
            f"{len(X_test):,}"
        )


        print(
            "\nSplit:"
        )

        print(
            "Train = 70%"
        )

        print(
            "Validation = 15%"
        )

        print(
            "Test = 15%"
        )


        return (

            X_train,
            X_val,
            X_test,

            y_train,
            y_val,
            y_test

        )


# ============================================================
# 5. BiLSTM MODEL
# ============================================================

class BiLSTMModel:

    @staticmethod
    def build(
        input_shape,
        units=64,
        dropout=0.20
    ):

        inputs = Input(
            shape=input_shape,
            name="Input"
        )


        # ====================================================
        # BiLSTM Layer 1
        # ====================================================

        x = Bidirectional(

            LSTM(

                units,

                return_sequences=True

            ),

            name="BiLSTM_1"

        )(inputs)


        x = Dropout(
            dropout,
            name="Dropout_1"
        )(x)


        # ====================================================
        # BiLSTM Layer 2
        # ====================================================

        x = Bidirectional(

            LSTM(
                units // 2
            ),

            name="BiLSTM_2"

        )(x)


        x = Dropout(
            dropout,
            name="Dropout_2"
        )(x)


        # ====================================================
        # Dense Layer
        # ====================================================

        x = Dense(

            64,

            activation="relu",

            name="Dense_64"

        )(x)


        x = Dropout(

            dropout,

            name="Dropout_3"

        )(x)


        # ====================================================
        # Kp × 10 OUTPUT
        # ====================================================

        kp_output = Dense(

            1,

            name="Kp10"

        )(x)


        # ====================================================
        # Dst OUTPUT
        # ====================================================

        dst_output = Dense(

            1,

            name="Dst"

        )(x)


        model = Model(

            inputs=inputs,

            outputs=[
                kp_output,
                dst_output
            ],

            name="BiLSTM_Geomagnetic"

        )


        return model


# ============================================================
# 6. TRAINER
# ============================================================

class BiLSTMTrainer:

    def __init__(
        self,
        model,
        scaler,
        all_columns
    ):

        self.model = model

        self.scaler = scaler

        self.all_columns = all_columns

        self.history = None


    # ========================================================
    # COMPILE
    # ========================================================

    def compile_model(
        self,
        learning_rate=0.001
    ):

        self.model.compile(

            optimizer=Adam(

                learning_rate=
                learning_rate

            ),

            loss={

                "Kp10": "mse",

                "Dst": "mse"

            },

            metrics={

                "Kp10": ["mae"],

                "Dst": ["mae"]

            },

            loss_weights={

                "Kp10": 1.0,

                "Dst": 1.0

            }

        )


    # ========================================================
    # TRAIN
    # ========================================================

    def train(

        self,
        X_train,
        y_train,
        X_val,
        y_val,
        epochs=50,
        batch_size=32

    ):

        model_path = os.path.join(

            OUTPUT_DIR,

            "BiLSTM_best.keras"

        )


        callbacks = [

            EarlyStopping(

                monitor="val_loss",

                patience=PATIENCE,

                restore_best_weights=True,

                verbose=1

            ),


            ReduceLROnPlateau(

                monitor="val_loss",

                factor=0.5,

                patience=3,

                min_lr=1e-7,

                verbose=1

            ),


            ModelCheckpoint(

                model_path,

                monitor="val_loss",

                save_best_only=True,

                verbose=1

            )

        ]


        self.history = self.model.fit(

            X_train,

            {

                "Kp10":
                    y_train[:, 0],

                "Dst":
                    y_train[:, 1]

            },


            validation_data=(

                X_val,

                {

                    "Kp10":
                        y_val[:, 0],

                    "Dst":
                        y_val[:, 1]

                }

            ),


            epochs=epochs,

            batch_size=batch_size,

            callbacks=callbacks,

            verbose=1

        )


        return self.history


    # ========================================================
    # INVERSE TRANSFORMATION
    # ========================================================

    def inverse_transform_targets(

        self,
        y_scaled

    ):

        kp_index = (
            self.all_columns.index(
                "Kp10"
            )
        )


        dst_index = (
            self.all_columns.index(
                "Dst"
            )
        )


        dummy = np.zeros(

            (
                len(y_scaled),
                len(self.all_columns)
            )

        )


        dummy[
            :,
            kp_index
        ] = y_scaled[:, 0]


        dummy[
            :,
            dst_index
        ] = y_scaled[:, 1]


        original = (
            self.scaler.inverse_transform(
                dummy
            )
        )


        kp = original[
            :,
            kp_index
        ]


        dst = original[
            :,
            dst_index
        ]


        return kp, dst


    # ========================================================
    # STANDARD EVALUATION
    # ========================================================

    def evaluate(

        self,
        X_test,
        y_test

    ):

        print(
            "\n" + "=" * 80
        )

        print(
            "BiLSTM MODEL EVALUATION"
        )

        print(
            "=" * 80
        )


        predictions = (
            self.model.predict(
                X_test,
                verbose=0
            )
        )


        kp_pred_scaled = (
            predictions[0].flatten()
        )


        dst_pred_scaled = (
            predictions[1].flatten()
        )


        y_pred_scaled = np.column_stack(

            [
                kp_pred_scaled,
                dst_pred_scaled
            ]

        )


        actual_kp, actual_dst = (

            self.inverse_transform_targets(
                y_test
            )

        )


        predicted_kp, predicted_dst = (

            self.inverse_transform_targets(
                y_pred_scaled
            )

        )


        # ====================================================
        # Kp METRICS
        # ====================================================

        kp_rmse = np.sqrt(

            mean_squared_error(

                actual_kp,
                predicted_kp

            )

        )


        kp_mae = mean_absolute_error(

            actual_kp,
            predicted_kp

        )


        kp_r2 = r2_score(

            actual_kp,
            predicted_kp

        )


        kp_corr = pearsonr(

            actual_kp,
            predicted_kp

        )[0]


        # ====================================================
        # Dst METRICS
        # ====================================================

        dst_rmse = np.sqrt(

            mean_squared_error(

                actual_dst,
                predicted_dst

            )

        )


        dst_mae = mean_absolute_error(

            actual_dst,
            predicted_dst

        )


        dst_r2 = r2_score(

            actual_dst,
            predicted_dst

        )


        dst_corr = pearsonr(

            actual_dst,
            predicted_dst

        )[0]


        dst_pe = (

            1

            -

            np.sum(

                (
                    actual_dst
                    -
                    predicted_dst
                ) ** 2

            )

            /

            np.sum(

                actual_dst ** 2

            )

        )


        metrics = {

            "Kp10": {

                "RMSE": kp_rmse,

                "MAE": kp_mae,

                "R2": kp_r2,

                "Correlation": kp_corr

            },

            "Dst": {

                "RMSE": dst_rmse,

                "MAE": dst_mae,

                "R2": dst_r2,

                "Correlation": dst_corr,

                "PE": dst_pe

            }

        }


        print(
            "\nKp × 10 Performance"
        )

        print(
            "-" * 50
        )

        print(
            f"RMSE        : {kp_rmse:.4f}"
        )

        print(
            f"MAE         : {kp_mae:.4f}"
        )

        print(
            f"R²          : {kp_r2:.4f}"
        )

        print(
            f"Correlation : {kp_corr:.4f}"
        )


        print(
            "\nDst Performance"
        )

        print(
            "-" * 50
        )

        print(
            f"RMSE        : {dst_rmse:.4f} nT"
        )

        print(
            f"MAE         : {dst_mae:.4f} nT"
        )

        print(
            f"R²          : {dst_r2:.4f}"
        )

        print(
            f"Correlation : {dst_corr:.4f}"
        )

        print(
            f"PE          : {dst_pe:.4f}"
        )


        return {

            "actual_kp":
                actual_kp,

            "predicted_kp":
                predicted_kp,

            "actual_dst":
                actual_dst,

            "predicted_dst":
                predicted_dst,

            "metrics":
                metrics

        }


    # ========================================================
    # MONTE CARLO DROPOUT
    # ========================================================

    def monte_carlo_predictions(

        self,
        X_test,
        n_samples=100

    ):

        print(
            "\n" + "=" * 80
        )

        print(
            "MONTE CARLO DROPOUT UNCERTAINTY"
        )

        print(
            "=" * 80
        )

        print(
            f"MC samples: {n_samples}"
        )


        kp_predictions = []

        dst_predictions = []


        # ====================================================
        # STOCHASTIC FORWARD PASSES
        # ====================================================

        for i in range(
            n_samples
        ):

            predictions = (
                self.model(
                    X_test,
                    training=True
                )
            )


            kp_scaled = (
                predictions[0]
                .numpy()
                .flatten()
            )


            dst_scaled = (
                predictions[1]
                .numpy()
                .flatten()
            )


            scaled_predictions = (
                np.column_stack(
                    [
                        kp_scaled,
                        dst_scaled
                    ]
                )
            )


            kp_original, dst_original = (

                self.inverse_transform_targets(
                    scaled_predictions
                )

            )


            kp_predictions.append(
                kp_original
            )


            dst_predictions.append(
                dst_original
            )


            if (
                (i + 1) % 20 == 0
                or i == 0
                or i == n_samples - 1
            ):

                print(
                    f"MC pass "
                    f"{i + 1}/{n_samples}"
                )


        kp_predictions = np.array(
            kp_predictions
        )


        dst_predictions = np.array(
            dst_predictions
        )


        # ====================================================
        # MEAN PREDICTION
        # ====================================================

        kp_mean = np.mean(

            kp_predictions,

            axis=0

        )


        dst_mean = np.mean(

            dst_predictions,

            axis=0

        )


        # ====================================================
        # 95% INTERVAL
        # ====================================================

        alpha = (
            1 - CONFIDENCE_LEVEL
        )


        lower_percentile = (
            100 * alpha / 2
        )


        upper_percentile = (
            100 * (
                1 - alpha / 2
            )
        )


        kp_lower = np.percentile(

            kp_predictions,

            lower_percentile,

            axis=0

        )


        kp_upper = np.percentile(

            kp_predictions,

            upper_percentile,

            axis=0

        )


        dst_lower = np.percentile(

            dst_predictions,

            lower_percentile,

            axis=0

        )


        dst_upper = np.percentile(

            dst_predictions,

            upper_percentile,

            axis=0

        )


        # ====================================================
        # STANDARD DEVIATION
        # ====================================================

        kp_std = np.std(

            kp_predictions,

            axis=0

        )


        dst_std = np.std(

            dst_predictions,

            axis=0

        )


        print(
            "\n95% Monte Carlo prediction intervals calculated."
        )


        return {

            "kp_mean":
                kp_mean,

            "kp_lower":
                kp_lower,

            "kp_upper":
                kp_upper,

            "kp_std":
                kp_std,

            "dst_mean":
                dst_mean,

            "dst_lower":
                dst_lower,

            "dst_upper":
                dst_upper,

            "dst_std":
                dst_std,

            "kp_mc_samples":
                kp_predictions,

            "dst_mc_samples":
                dst_predictions

        }


# ============================================================
# 7. STORM CLASSIFICATION
# ============================================================

class StormClassifier:


    # ========================================================
    # Kp CLASSIFICATION
    # ========================================================

    @staticmethod
    def classify_kp(kp10):

        categories = []


        for value in kp10:

            if value < 50:

                categories.append(
                    "Quiet"
                )

            elif value < 60:

                categories.append(
                    "G1 Minor"
                )

            elif value < 70:

                categories.append(
                    "G2 Moderate"
                )

            elif value < 80:

                categories.append(
                    "G3 Strong"
                )

            elif value < 90:

                categories.append(
                    "G4 Severe"
                )

            else:

                categories.append(
                    "G5 Extreme"
                )


        return np.array(
            categories
        )


    # ========================================================
    # DST CLASSIFICATION
    # ========================================================

    @staticmethod
    def classify_dst(dst):

        categories = []


        for value in dst:

            if value > -30:

                categories.append(
                    "Quiet"
                )

            elif value > -50:

                categories.append(
                    "Minor"
                )

            elif value > -100:

                categories.append(
                    "Moderate"
                )

            elif value > -200:

                categories.append(
                    "Intense"
                )

            else:

                categories.append(
                    "Super Storm"
                )


        return np.array(
            categories
        )


    # ========================================================
    # COMBINED CLASSIFICATION
    # ========================================================

    @staticmethod
    def classify_combined(

        kp10,
        dst

    ):

        kp_classes = (
            StormClassifier.classify_kp(
                kp10
            )
        )


        dst_classes = (
            StormClassifier.classify_dst(
                dst
            )
        )


        combined = []


        for kp_class, dst_class in zip(

            kp_classes,
            dst_classes

        ):

            if (

                kp_class == "G5 Extreme"
                or
                dst_class == "Super Storm"

            ):

                combined.append(
                    "Extreme/Super Storm"
                )


            elif (

                kp_class == "G4 Severe"
                or
                dst_class == "Intense"

            ):

                combined.append(
                    "Severe/Intense"
                )


            elif (

                kp_class == "G3 Strong"
                or
                dst_class == "Moderate"

            ):

                combined.append(
                    "Strong/Moderate"
                )


            elif (

                kp_class == "G2 Moderate"
                or
                dst_class == "Minor"

            ):

                combined.append(
                    "Moderate/Minor"
                )


            elif (

                kp_class == "G1 Minor"

            ):

                combined.append(
                    "Minor"
                )


            else:

                combined.append(
                    "Quiet"
                )


        return np.array(
            combined
        )


    # ========================================================
    # BINARY STORM
    # ========================================================

    @staticmethod
    def binary_storm(

        kp10,
        dst

    ):

        return (

            (
                np.asarray(kp10)
                >= 50
            )

            |

            (
                np.asarray(dst)
                <= -50
            )

        ).astype(int)


    # ========================================================
    # CLASSIFICATION EVALUATION
    # ========================================================

    @staticmethod
    def evaluate(

        actual_kp,
        predicted_kp,
        actual_dst,
        predicted_dst

    ):

        print(
            "\n" + "=" * 80
        )

        print(
            "BiLSTM STORM CLASSIFICATION"
        )

        print(
            "=" * 80
        )


        # ====================================================
        # Kp
        # ====================================================

        actual_kp_class = (
            StormClassifier.classify_kp(
                actual_kp
            )
        )


        predicted_kp_class = (
            StormClassifier.classify_kp(
                predicted_kp
            )
        )


        kp_accuracy = accuracy_score(

            actual_kp_class,
            predicted_kp_class

        )


        kp_labels = [

            "Quiet",
            "G1 Minor",
            "G2 Moderate",
            "G3 Strong",
            "G4 Severe",
            "G5 Extreme"

        ]


        kp_cm = confusion_matrix(

            actual_kp_class,
            predicted_kp_class,

            labels=kp_labels

        )


        print(
            "\nKp × 10 Classification Accuracy:"
        )

        print(
            f"{kp_accuracy:.4f}"
        )


        print(
            "\nKp × 10 Classification Report:"
        )


        print(

            classification_report(

                actual_kp_class,
                predicted_kp_class,

                labels=kp_labels,

                zero_division=0

            )

        )


        # ====================================================
        # DST
        # ====================================================

        actual_dst_class = (
            StormClassifier.classify_dst(
                actual_dst
            )
        )


        predicted_dst_class = (
            StormClassifier.classify_dst(
                predicted_dst
            )
        )


        dst_accuracy = accuracy_score(

            actual_dst_class,
            predicted_dst_class

        )


        dst_labels = [

            "Quiet",
            "Minor",
            "Moderate",
            "Intense",
            "Super Storm"

        ]


        dst_cm = confusion_matrix(

            actual_dst_class,
            predicted_dst_class,

            labels=dst_labels

        )


        print(
            "\nDst Classification Accuracy:"
        )

        print(
            f"{dst_accuracy:.4f}"
        )


        print(
            "\nDst Classification Report:"
        )


        print(

            classification_report(

                actual_dst_class,
                predicted_dst_class,

                labels=dst_labels,

                zero_division=0

            )

        )


        # ====================================================
        # BINARY
        # ====================================================

        actual_binary = (
            StormClassifier.binary_storm(

                actual_kp,
                actual_dst

            )
        )


        predicted_binary = (
            StormClassifier.binary_storm(

                predicted_kp,
                predicted_dst

            )
        )


        binary_accuracy = accuracy_score(

            actual_binary,
            predicted_binary

        )


        binary_precision = precision_score(

            actual_binary,
            predicted_binary,

            zero_division=0

        )


        binary_recall = recall_score(

            actual_binary,
            predicted_binary,

            zero_division=0

        )


        binary_f1 = f1_score(

            actual_binary,
            predicted_binary,

            zero_division=0

        )


        binary_cm = confusion_matrix(

            actual_binary,
            predicted_binary

        )


        print(
            "\nCombined Binary Storm Classification:"
        )

        print(
            f"Accuracy  : {binary_accuracy:.4f}"
        )

        print(
            f"Precision : {binary_precision:.4f}"
        )

        print(
            f"Recall    : {binary_recall:.4f}"
        )

        print(
            f"F1 Score  : {binary_f1:.4f}"
        )


        return {

            "actual_kp_class":
                actual_kp_class,

            "predicted_kp_class":
                predicted_kp_class,

            "kp_cm":
                kp_cm,

            "actual_dst_class":
                actual_dst_class,

            "predicted_dst_class":
                predicted_dst_class,

            "dst_cm":
                dst_cm,

            "actual_binary":
                actual_binary,

            "predicted_binary":
                predicted_binary,

            "binary_cm":
                binary_cm,

            "kp_accuracy":
                kp_accuracy,

            "dst_accuracy":
                dst_accuracy,

            "binary_accuracy":
                binary_accuracy,

            "binary_precision":
                binary_precision,

            "binary_recall":
                binary_recall,

            "binary_f1":
                binary_f1
        }


# ============================================================
# 8. CONFIDENCE INTERVAL ANALYSIS
# ============================================================

def calculate_interval_statistics(

    actual,
    lower,
    upper

):

    inside = (

        (actual >= lower)
        &
        (actual <= upper)

    )


    coverage = (
        np.mean(inside) * 100
    )


    width = (
        upper - lower
    )


    mean_width = (
        np.mean(width)
    )


    median_width = (
        np.median(width)
    )


    return {

        "Coverage_Percent":
            coverage,

        "Mean_Interval_Width":
            mean_width,

        "Median_Interval_Width":
            median_width

    }


# ============================================================
# 9. VISUALIZATION
# ============================================================

class Visualizer:


    # ========================================================
    # TRAINING HISTORY
    # ========================================================

    @staticmethod
    def plot_training_history(history):

        plt.figure(
            figsize=(12, 6)
        )


        plt.plot(

            history.history["loss"],

            label="Training Loss"

        )


        plt.plot(

            history.history["val_loss"],

            linestyle="--",

            label="Validation Loss"

        )


        plt.xlabel(
            "Epoch"
        )

        plt.ylabel(
            "Total Loss"
        )

        plt.title(
            "BiLSTM Training and Validation Loss"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_training_validation_loss.png"

            ),

            dpi=300

        )


        plt.show()


    # ========================================================
    # DST PREDICTION + 95% INTERVAL
    # ========================================================

    @staticmethod
    def plot_dst_prediction_interval(

        actual,
        mean_prediction,
        lower,
        upper,
        sample_size=2000

    ):

        n = min(
            sample_size,
            len(actual)
        )


        x = np.arange(n)


        plt.figure(
            figsize=(16, 7)
        )


        plt.plot(

            x,

            actual[:n],

            label="Actual Dst",

            linewidth=1.2

        )


        plt.plot(

            x,

            mean_prediction[:n],

            label="BiLSTM Predicted Dst",

            linewidth=1.2

        )


        plt.fill_between(

            x,

            lower[:n],

            upper[:n],

            alpha=0.25,

            label="95% Prediction Interval"

        )


        plt.axhline(

            -30,

            linestyle="--",

            label="Minor: -30 nT"

        )


        plt.axhline(

            -50,

            linestyle="--",

            label="Moderate: -50 nT"

        )


        plt.axhline(

            -100,

            linestyle="--",

            label="Intense: -100 nT"

        )


        plt.axhline(

            -200,

            linestyle="--",

            label="Super Storm: -200 nT"

        )


        plt.xlabel(
            "Test Time Step"
        )

        plt.ylabel(
            "Dst (nT)"
        )

        plt.title(

            "BiLSTM Dst Prediction with 95% "
            "Monte Carlo Prediction Interval"

        )


        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_95CI_prediction.png"

            ),

            dpi=300

        )


        plt.show()


    # ========================================================
    # KP PREDICTION + 95% INTERVAL
    # ========================================================

    @staticmethod
    def plot_kp_prediction_interval(

        actual,
        mean_prediction,
        lower,
        upper,
        sample_size=2000

    ):

        n = min(

            sample_size,
            len(actual)

        )


        x = np.arange(n)


        plt.figure(
            figsize=(16, 7)
        )


        plt.plot(

            x,

            actual[:n],

            label="Actual Kp × 10",

            linewidth=1.2

        )


        plt.plot(

            x,

            mean_prediction[:n],

            label="BiLSTM Predicted Kp × 10",

            linewidth=1.2

        )


        plt.fill_between(

            x,

            lower[:n],

            upper[:n],

            alpha=0.25,

            label="95% Prediction Interval"

        )


        # Kp thresholds

        plt.axhline(

            50,

            linestyle="--",

            label="Kp = 5 (G1)"

        )


        plt.axhline(

            60,

            linestyle="--",

            label="Kp = 6 (G2)"

        )


        plt.axhline(

            70,

            linestyle="--",

            label="Kp = 7 (G3)"

        )


        plt.axhline(

            80,

            linestyle="--",

            label="Kp = 8 (G4)"

        )


        plt.axhline(

            90,

            linestyle="--",

            label="Kp = 9 (G5)"

        )


        plt.xlabel(
            "Test Time Step"
        )

        plt.ylabel(
            "Kp × 10"
        )

        plt.title(

            "BiLSTM Kp × 10 Prediction with "
            "95% Monte Carlo Prediction Interval"

        )


        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Kp10_95CI_prediction.png"

            ),

            dpi=300

        )


        plt.show()


    # ========================================================
    # UNCERTAINTY WIDTH
    # ========================================================

    @staticmethod
    def plot_uncertainty_width(

        kp_lower,
        kp_upper,
        dst_lower,
        dst_upper,
        sample_size=2000

    ):

        # ----------------------------------------------------
        # Dst uncertainty
        # ----------------------------------------------------

        n = min(

            sample_size,
            len(dst_lower)

        )


        x = np.arange(n)


        plt.figure(
            figsize=(15, 6)
        )


        plt.plot(

            x,

            dst_upper[:n]
            -
            dst_lower[:n],

            label="Dst 95% Interval Width"

        )


        plt.xlabel(
            "Test Time Step"
        )

        plt.ylabel(
            "Interval Width (nT)"
        )

        plt.title(
            "BiLSTM Dst Prediction Uncertainty"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_uncertainty_width.png"

            ),

            dpi=300

        )


        plt.show()


        # ----------------------------------------------------
        # Kp uncertainty
        # ----------------------------------------------------

        n = min(

            sample_size,
            len(kp_lower)

        )


        x = np.arange(n)


        plt.figure(
            figsize=(15, 6)
        )


        plt.plot(

            x,

            kp_upper[:n]
            -
            kp_lower[:n],

            label="Kp × 10 95% Interval Width"

        )


        plt.xlabel(
            "Test Time Step"
        )

        plt.ylabel(
            "Interval Width"
        )

        plt.title(
            "BiLSTM Kp × 10 Prediction Uncertainty"
        )

        plt.legend()

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Kp10_uncertainty_width.png"

            ),

            dpi=300

        )


        plt.show()


    # ========================================================
    # DST SCATTER
    # ========================================================

    @staticmethod
    def plot_dst_scatter(

        actual,
        predicted,
        r2

    ):

        plt.figure(
            figsize=(8, 8)
        )


        plt.scatter(

            actual,
            predicted,

            alpha=0.4,

            s=12

        )


        minimum = min(

            actual.min(),
            predicted.min()

        )


        maximum = max(

            actual.max(),
            predicted.max()

        )


        plt.plot(

            [minimum, maximum],

            [minimum, maximum],

            linestyle="--",

            linewidth=2

        )


        plt.xlabel(
            "Actual Dst (nT)"
        )

        plt.ylabel(
            "Predicted Dst (nT)"
        )


        plt.title(

            f"BiLSTM Dst Prediction\n"
            f"R² = {r2:.4f}"

        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_scatter.png"

            ),

            dpi=300

        )


        plt.show()


    # ========================================================
    # KP SCATTER
    # ========================================================

    @staticmethod
    def plot_kp_scatter(

        actual,
        predicted,
        r2

    ):

        plt.figure(
            figsize=(8, 8)
        )


        plt.scatter(

            actual,
            predicted,

            alpha=0.4,

            s=12

        )


        minimum = min(

            actual.min(),
            predicted.min()

        )


        maximum = max(

            actual.max(),
            predicted.max()

        )


        plt.plot(

            [minimum, maximum],

            [minimum, maximum],

            linestyle="--",

            linewidth=2

        )


        plt.xlabel(
            "Actual Kp × 10"
        )

        plt.ylabel(
            "Predicted Kp × 10"
        )


        plt.title(

            f"BiLSTM Kp × 10 Prediction\n"
            f"R² = {r2:.4f}"

        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Kp10_scatter.png"

            ),

            dpi=300

        )


        plt.show()


    # ========================================================
    # RESIDUAL
    # ========================================================

    @staticmethod
    def plot_residuals(

        actual_dst,
        predicted_dst

    ):

        residuals = (

            actual_dst
            -
            predicted_dst

        )


        plt.figure(
            figsize=(12, 6)
        )


        plt.scatter(

            predicted_dst,

            residuals,

            alpha=0.4,

            s=12

        )


        plt.axhline(

            0,

            linestyle="--",

            linewidth=2

        )


        plt.xlabel(
            "Predicted Dst (nT)"
        )

        plt.ylabel(
            "Residual (nT)"
        )


        plt.title(
            "BiLSTM Dst Residual Analysis"
        )


        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                "BiLSTM_Dst_residuals.png"

            ),

            dpi=300

        )


        plt.show()


    # ========================================================
    # CONFUSION MATRIX
    # ========================================================

    @staticmethod
    def plot_confusion_matrix(

        cm,
        labels,
        title,
        filename

    ):

        plt.figure(
            figsize=(10, 8)
        )


        sns.heatmap(

            cm,

            annot=True,

            fmt="d",

            cmap="Blues",

            xticklabels=labels,

            yticklabels=labels

        )


        plt.xlabel(
            "Predicted"
        )

        plt.ylabel(
            "Actual"
        )

        plt.title(
            title
        )


        plt.tight_layout()


        plt.savefig(

            os.path.join(

                OUTPUT_DIR,

                filename

            ),

            dpi=300

        )


        plt.show()


# ============================================================
# 10. ROC + PRECISION RECALL
# ============================================================

def create_storm_scores(

    actual_kp,
    actual_dst,
    predicted_kp,
    predicted_dst

):

    # Actual binary storm

    actual_binary = (

        (
            actual_kp
            >= 50
        )

        |

        (
            actual_dst
            <= -50
        )

    ).astype(int)


    # Continuous predicted storm score

    kp_score = (

        predicted_kp
        / 50.0

    )


    dst_score = (

        -predicted_dst
        / 50.0

    )


    predicted_score = np.maximum(

        kp_score,
        dst_score

    )


    return (

        actual_binary,
        predicted_score

    )


def plot_roc_pr(

    actual_binary,
    predicted_score

):

    # ========================================================
    # ROC
    # ========================================================

    fpr, tpr, _ = roc_curve(

        actual_binary,
        predicted_score

    )


    roc_auc = auc(

        fpr,
        tpr

    )


    plt.figure(
        figsize=(9, 7)
    )


    plt.plot(

        fpr,

        tpr,

        linewidth=2,

        label=f"BiLSTM AUC = {roc_auc:.4f}"

    )


    plt.plot(

        [0, 1],

        [0, 1],

        linestyle="--"

    )


    plt.xlabel(
        "False Positive Rate"
    )

    plt.ylabel(
        "True Positive Rate"
    )


    plt.title(
        "BiLSTM ROC Curve - Geomagnetic Storm"
    )


    plt.legend()

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()


    plt.savefig(

        os.path.join(

            OUTPUT_DIR,

            "BiLSTM_ROC_curve.png"

        ),

        dpi=300

    )


    plt.show()


    # ========================================================
    # PRECISION RECALL
    # ========================================================

    precision, recall, _ = (

        precision_recall_curve(

            actual_binary,
            predicted_score

        )

    )


    ap = average_precision_score(

        actual_binary,
        predicted_score

    )


    plt.figure(
        figsize=(9, 7)
    )


    plt.plot(

        recall,

        precision,

        linewidth=2,

        label=f"BiLSTM AP = {ap:.4f}"

    )


    plt.xlabel(
        "Recall"
    )

    plt.ylabel(
        "Precision"
    )


    plt.title(
        "BiLSTM Precision-Recall Curve"
    )


    plt.legend()

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()


    plt.savefig(

        os.path.join(

            OUTPUT_DIR,

            "BiLSTM_PR_curve.png"

        ),

        dpi=300

    )


    plt.show()


    return roc_auc, ap


# ============================================================
# 11. SAVE PREDICTIONS
# ============================================================

def save_prediction_results(

    result,
    uncertainty

):

    df_output = pd.DataFrame({

        # ----------------------------------------------------
        # Kp
        # ----------------------------------------------------

        "Actual_Kp10":
            result["actual_kp"],

        "Predicted_Kp10":
            uncertainty["kp_mean"],

        "Kp10_Lower_95CI":
            uncertainty["kp_lower"],

        "Kp10_Upper_95CI":
            uncertainty["kp_upper"],

        "Kp10_Interval_Width":
            (
                uncertainty["kp_upper"]
                -
                uncertainty["kp_lower"]
            ),

        "Kp10_Uncertainty_SD":
            uncertainty["kp_std"],


        # ----------------------------------------------------
        # Dst
        # ----------------------------------------------------

        "Actual_Dst":
            result["actual_dst"],

        "Predicted_Dst":
            uncertainty["dst_mean"],

        "Dst_Lower_95CI":
            uncertainty["dst_lower"],

        "Dst_Upper_95CI":
            uncertainty["dst_upper"],

        "Dst_Interval_Width":
            (
                uncertainty["dst_upper"]
                -
                uncertainty["dst_lower"]
            ),

        "Dst_Uncertainty_SD":
            uncertainty["dst_std"]

    })


    # ========================================================
    # CLASSIFICATION
    # ========================================================

    df_output[
        "Actual_Kp_Class"
    ] = StormClassifier.classify_kp(

        result["actual_kp"]

    )


    df_output[
        "Predicted_Kp_Class"
    ] = StormClassifier.classify_kp(

        uncertainty["kp_mean"]

    )


    df_output[
        "Actual_Dst_Class"
    ] = StormClassifier.classify_dst(

        result["actual_dst"]

    )


    df_output[
        "Predicted_Dst_Class"
    ] = StormClassifier.classify_dst(

        uncertainty["dst_mean"]

    )


    df_output[
        "Actual_Combined_Class"
    ] = StormClassifier.classify_combined(

        result["actual_kp"],

        result["actual_dst"]

    )


    df_output[
        "Predicted_Combined_Class"
    ] = StormClassifier.classify_combined(

        uncertainty["kp_mean"],

        uncertainty["dst_mean"]

    )


    # ========================================================
    # BINARY STORM
    # ========================================================

    df_output[
        "Actual_Storm"
    ] = StormClassifier.binary_storm(

        result["actual_kp"],

        result["actual_dst"]

    )


    df_output[
        "Predicted_Storm"
    ] = StormClassifier.binary_storm(

        uncertainty["kp_mean"],

        uncertainty["dst_mean"]

    )


    filename = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_predictions_with_95CI.csv"

    )


    df_output.to_csv(

        filename,

        index=False

    )


    print(
        f"\nPrediction results saved:\n"
        f"{filename}"
    )


    return df_output


# ============================================================
# 12. SAVE METRICS
# ============================================================

def save_metrics(

    result,
    interval_statistics,
    roc_auc,
    average_precision

):

    metrics_df = pd.DataFrame({

        "Metric": [

            "Kp10 RMSE",
            "Kp10 MAE",
            "Kp10 R2",
            "Kp10 Correlation",

            "Dst RMSE",
            "Dst MAE",
            "Dst R2",
            "Dst Correlation",
            "Dst Prediction Efficiency",

            "Kp10 95CI Coverage (%)",
            "Kp10 Mean Interval Width",
            "Kp10 Median Interval Width",

            "Dst 95CI Coverage (%)",
            "Dst Mean Interval Width",
            "Dst Median Interval Width",

            "ROC AUC",
            "Average Precision"

        ],


        "Value": [

            result["metrics"]["Kp10"]["RMSE"],

            result["metrics"]["Kp10"]["MAE"],

            result["metrics"]["Kp10"]["R2"],

            result["metrics"]["Kp10"]["Correlation"],


            result["metrics"]["Dst"]["RMSE"],

            result["metrics"]["Dst"]["MAE"],

            result["metrics"]["Dst"]["R2"],

            result["metrics"]["Dst"]["Correlation"],

            result["metrics"]["Dst"]["PE"],


            interval_statistics["Kp10"][
                "Coverage_Percent"
            ],

            interval_statistics["Kp10"][
                "Mean_Interval_Width"
            ],

            interval_statistics["Kp10"][
                "Median_Interval_Width"
            ],


            interval_statistics["Dst"][
                "Coverage_Percent"
            ],

            interval_statistics["Dst"][
                "Mean_Interval_Width"
            ],

            interval_statistics["Dst"][
                "Median_Interval_Width"
            ],


            roc_auc,

            average_precision

        ]

    })


    filename = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_complete_metrics.csv"

    )


    metrics_df.to_csv(

        filename,

        index=False

    )


    print(
        f"\nMetrics saved:\n{filename}"
    )


    print(
        "\n" + "=" * 90
    )

    print(
        "COMPLETE BiLSTM PERFORMANCE"
    )

    print(
        "=" * 90
    )

    print(
        metrics_df.to_string(
            index=False
        )
    )


    return metrics_df


# ============================================================
# 13. MAIN FUNCTION
# ============================================================

def main():

    print("\n")

    print("=" * 80)

    print(
        "GEOMAGNETIC STORM PREDICTION USING BiLSTM"
    )

    print(
        "Kp × 10 + Dst"
    )

    print(
        "70% Train | 15% Validation | 15% Test"
    )

    print(
        "48-Hour Lookback"
    )

    print(
        "95% Monte Carlo Prediction Interval"
    )

    print("=" * 80)


    # ========================================================
    # STEP 1
    # LOAD DATA
    # ========================================================

    preprocessor = DataPreprocessor(

        DATA_PATH

    )


    preprocessor.load_data()

    preprocessor.check_columns()

    preprocessor.handle_missing_values()

    preprocessor.clean_data()


    df = (
        preprocessor.get_processed_data()
    )


    # Save cleaned data

    cleaned_path = os.path.join(

        OUTPUT_DIR,

        "cleaned_geomagnetic_data.csv"

    )


    df.to_csv(

        cleaned_path,

        index=False

    )


    print(
        f"\nCleaned data saved:\n"
        f"{cleaned_path}"
    )


    # ========================================================
    # STEP 2
    # SEQUENCES
    # ========================================================

    seq_gen = SequenceGenerator(

        df,

        lookback=LOOKBACK

    )


    X, y = (
        seq_gen.create_sequences()
    )


    (
        X_train,
        X_val,
        X_test,

        y_train,
        y_val,
        y_test

    ) = seq_gen.split_data(

        X,

        y,

        test_size=TEST_SIZE,

        val_size=VALIDATION_SIZE

    )


    # ========================================================
    # STEP 3
    # MODEL INPUT
    # ========================================================

    input_shape = (

        X_train.shape[1],

        X_train.shape[2]

    )


    print(
        "\nBiLSTM input shape:",
        input_shape
    )


    # ========================================================
    # STEP 4
    # BUILD BiLSTM
    # ========================================================

    model = BiLSTMModel.build(

        input_shape=input_shape,

        units=64,

        dropout=0.20

    )


    # ========================================================
    # MODEL SUMMARY
    # ========================================================

    print(
        "\n" + "=" * 80
    )

    print(
        "BiLSTM MODEL ARCHITECTURE"
    )

    print(
        "=" * 80
    )


    model.summary()


    # ========================================================
    # STEP 5
    # TRAINER
    # ========================================================

    trainer = BiLSTMTrainer(

        model,

        seq_gen.scaler,

        seq_gen.all_columns

    )


    trainer.compile_model(

        learning_rate=LEARNING_RATE

    )


    # ========================================================
    # STEP 6
    # TRAIN
    # ========================================================

    history = trainer.train(

        X_train,

        y_train,

        X_val,

        y_val,

        epochs=EPOCHS,

        batch_size=BATCH_SIZE

    )


    # ========================================================
    # STEP 7
    # STANDARD EVALUATION
    # ========================================================

    result = trainer.evaluate(

        X_test,

        y_test

    )


    # ========================================================
    # STEP 8
    # MONTE CARLO DROPOUT
    # ========================================================

    uncertainty = (

        trainer.monte_carlo_predictions(

            X_test,

            n_samples=MC_SAMPLES

        )

    )


    # ========================================================
    # USE MC MEAN AS FINAL PREDICTION
    # ========================================================

    result["predicted_kp"] = (

        uncertainty["kp_mean"]

    )


    result["predicted_dst"] = (

        uncertainty["dst_mean"]

    )


    # ========================================================
    # STEP 9
    # CONFIDENCE INTERVAL STATISTICS
    # ========================================================

    kp_interval_statistics = (

        calculate_interval_statistics(

            result["actual_kp"],

            uncertainty["kp_lower"],

            uncertainty["kp_upper"]

        )

    )


    dst_interval_statistics = (

        calculate_interval_statistics(

            result["actual_dst"],

            uncertainty["dst_lower"],

            uncertainty["dst_upper"]

        )

    )


    interval_statistics = {

        "Kp10":
            kp_interval_statistics,

        "Dst":
            dst_interval_statistics

    }


    print(
        "\n" + "=" * 80
    )

    print(
        "95% CONFIDENCE / PREDICTION INTERVAL RESULTS"
    )

    print(
        "=" * 80
    )


    print(
        "\nKp × 10:"
    )

    print(

        f"95% Interval Coverage: "
        f"{kp_interval_statistics['Coverage_Percent']:.2f}%"

    )

    print(

        f"Mean Interval Width: "
        f"{kp_interval_statistics['Mean_Interval_Width']:.4f}"

    )

    print(

        f"Median Interval Width: "
        f"{kp_interval_statistics['Median_Interval_Width']:.4f}"

    )


    print(
        "\nDst:"
    )

    print(

        f"95% Interval Coverage: "
        f"{dst_interval_statistics['Coverage_Percent']:.2f}%"

    )

    print(

        f"Mean Interval Width: "
        f"{dst_interval_statistics['Mean_Interval_Width']:.4f} nT"

    )

    print(

        f"Median Interval Width: "
        f"{dst_interval_statistics['Median_Interval_Width']:.4f} nT"

    )


    # ========================================================
    # STEP 10
    # CLASSIFICATION
    # ========================================================

    classification = (

        StormClassifier.evaluate(

            result["actual_kp"],

            result["predicted_kp"],

            result["actual_dst"],

            result["predicted_dst"]

        )

    )


    # ========================================================
    # STEP 11
    # ROC / PR
    # ========================================================

    (

        actual_binary,

        predicted_score

    ) = create_storm_scores(

        result["actual_kp"],

        result["actual_dst"],

        result["predicted_kp"],

        result["predicted_dst"]

    )


    roc_auc, average_precision = (

        plot_roc_pr(

            actual_binary,

            predicted_score

        )

    )


    print(
        "\nROC AUC:",
        f"{roc_auc:.4f}"
    )


    print(
        "Average Precision:",
        f"{average_precision:.4f}"
    )


    # ========================================================
    # STEP 12
    # TRAINING PLOT
    # ========================================================

    Visualizer.plot_training_history(

        history

    )


    # ========================================================
    # STEP 13
    # DST PREDICTION WITH CI
    # ========================================================

    Visualizer.plot_dst_prediction_interval(

        result["actual_dst"],

        uncertainty["dst_mean"],

        uncertainty["dst_lower"],

        uncertainty["dst_upper"],

        sample_size=PLOT_SAMPLES

    )


    # ========================================================
    # STEP 14
    # KP PREDICTION WITH CI
    # ========================================================

    Visualizer.plot_kp_prediction_interval(

        result["actual_kp"],

        uncertainty["kp_mean"],

        uncertainty["kp_lower"],

        uncertainty["kp_upper"],

        sample_size=PLOT_SAMPLES

    )


    # ========================================================
    # STEP 15
    # UNCERTAINTY WIDTH
    # ========================================================

    Visualizer.plot_uncertainty_width(

        uncertainty["kp_lower"],

        uncertainty["kp_upper"],

        uncertainty["dst_lower"],

        uncertainty["dst_upper"],

        sample_size=PLOT_SAMPLES

    )


    # ========================================================
    # STEP 16
    # SCATTER
    # ========================================================

    Visualizer.plot_dst_scatter(

        result["actual_dst"],

        result["predicted_dst"],

        result["metrics"]["Dst"]["R2"]

    )


    Visualizer.plot_kp_scatter(

        result["actual_kp"],

        result["predicted_kp"],

        result["metrics"]["Kp10"]["R2"]

    )


    # ========================================================
    # STEP 17
    # RESIDUAL
    # ========================================================

    Visualizer.plot_residuals(

        result["actual_dst"],

        result["predicted_dst"]

    )


    # ========================================================
    # STEP 18
    # CONFUSION MATRICES
    # ========================================================

    Visualizer.plot_confusion_matrix(

        classification["kp_cm"],

        [

            "Quiet",
            "G1 Minor",
            "G2 Moderate",
            "G3 Strong",
            "G4 Severe",
            "G5 Extreme"

        ],

        "BiLSTM - Kp × 10 Storm Classification",

        "BiLSTM_Kp10_confusion_matrix.png"

    )


    Visualizer.plot_confusion_matrix(

        classification["dst_cm"],

        [

            "Quiet",
            "Minor",
            "Moderate",
            "Intense",
            "Super Storm"

        ],

        "BiLSTM - Dst Storm Classification",

        "BiLSTM_Dst_confusion_matrix.png"

    )


    Visualizer.plot_confusion_matrix(

        classification["binary_cm"],

        [

            "No Storm",
            "Storm"

        ],

        "BiLSTM - Combined Kp × 10 + Dst",

        "BiLSTM_Combined_confusion_matrix.png"

    )


    # ========================================================
    # STEP 19
    # SAVE PREDICTIONS
    # ========================================================

    prediction_df = save_prediction_results(

        result,

        uncertainty

    )


    # ========================================================
    # STEP 20
    # SAVE METRICS
    # ========================================================

    metrics_df = save_metrics(

        result,

        interval_statistics,

        roc_auc,

        average_precision

    )


    # ========================================================
    # STEP 21
    # CLASSIFICATION METRICS CSV
    # ========================================================

    classification_df = pd.DataFrame({

        "Model": [
            "BiLSTM"
        ],

        "Kp10_Accuracy": [
            classification[
                "kp_accuracy"
            ]
        ],

        "Dst_Accuracy": [
            classification[
                "dst_accuracy"
            ]
        ],

        "Combined_Accuracy": [
            classification[
                "binary_accuracy"
            ]
        ],

        "Combined_Precision": [
            classification[
                "binary_precision"
            ]
        ],

        "Combined_Recall": [
            classification[
                "binary_recall"
            ]
        ],

        "Combined_F1": [
            classification[
                "binary_f1"
            ]
        ],

        "ROC_AUC": [
            roc_auc
        ],

        "Average_Precision": [
            average_precision
        ]

    })


    classification_path = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_classification_metrics.csv"

    )


    classification_df.to_csv(

        classification_path,

        index=False

    )


    print(
        f"\nClassification metrics saved:\n"
        f"{classification_path}"
    )


    # ========================================================
    # STEP 22
    # SAVE INTERVAL STATISTICS
    # ========================================================

    interval_df = pd.DataFrame({

        "Target": [

            "Kp10",

            "Dst"

        ],

        "Confidence_Level": [

            "95%",

            "95%"

        ],

        "Coverage_Percent": [

            kp_interval_statistics[
                "Coverage_Percent"
            ],

            dst_interval_statistics[
                "Coverage_Percent"
            ]

        ],

        "Mean_Interval_Width": [

            kp_interval_statistics[
                "Mean_Interval_Width"
            ],

            dst_interval_statistics[
                "Mean_Interval_Width"
            ]

        ],

        "Median_Interval_Width": [

            kp_interval_statistics[
                "Median_Interval_Width"
            ],

            dst_interval_statistics[
                "Median_Interval_Width"
            ]

        ]

    })


    interval_path = os.path.join(

        OUTPUT_DIR,

        "BiLSTM_95CI_statistics.csv"

    )


    interval_df.to_csv(

        interval_path,

        index=False

    )


    print(
        f"\n95% CI statistics saved:\n"
        f"{interval_path}"
    )


    # ========================================================
    # STEP 23
    # FINAL SUMMARY
    # ========================================================

    print(
        "\n" + "=" * 80
    )

    print(
        "FINAL BiLSTM RESULTS"
    )

    print(
        "=" * 80
    )


    print(
        "\nKp × 10:"
    )

    print(
        f"RMSE = "
        f"{result['metrics']['Kp10']['RMSE']:.4f}"
    )

    print(
        f"MAE = "
        f"{result['metrics']['Kp10']['MAE']:.4f}"
    )

    print(
        f"R² = "
        f"{result['metrics']['Kp10']['R2']:.4f}"
    )

    print(
        f"Correlation = "
        f"{result['metrics']['Kp10']['Correlation']:.4f}"
    )


    print(
        "\nDst:"
    )

    print(
        f"RMSE = "
        f"{result['metrics']['Dst']['RMSE']:.4f} nT"
    )

    print(
        f"MAE = "
        f"{result['metrics']['Dst']['MAE']:.4f} nT"
    )

    print(
        f"R² = "
        f"{result['metrics']['Dst']['R2']:.4f}"
    )

    print(
        f"Correlation = "
        f"{result['metrics']['Dst']['Correlation']:.4f}"
    )

    print(
        f"Prediction Efficiency = "
        f"{result['metrics']['Dst']['PE']:.4f}"
    )


    print(
        "\n95% Prediction Interval:"
    )

    print(

        f"Kp × 10 Coverage = "
        f"{kp_interval_statistics['Coverage_Percent']:.2f}%"

    )

    print(

        f"Dst Coverage = "
        f"{dst_interval_statistics['Coverage_Percent']:.2f}%"

    )


    print(
        "\nCombined Storm Classification:"
    )

    print(

        f"Accuracy = "
        f"{classification['binary_accuracy']:.4f}"

    )

    print(

        f"Precision = "
        f"{classification['binary_precision']:.4f}"

    )

    print(

        f"Recall = "
        f"{classification['binary_recall']:.4f}"

    )

    print(

        f"F1 = "
        f"{classification['binary_f1']:.4f}"

    )


    print(
        "\nROC AUC = "
        f"{roc_auc:.4f}"
    )


    print(
        "Average Precision = "
        f"{average_precision:.4f}"
    )


    print(
        "\nAll results saved in:"
    )

    print(
        OUTPUT_DIR
    )


    print(
        "\n" + "=" * 80
    )

    print(
        "BiLSTM ANALYSIS COMPLETED"
    )

    print(
        "=" * 80
    )


    return {

        "model": model,

        "history": history,

        "result": result,

        "uncertainty": uncertainty,

        "classification": classification,

        "metrics_df": metrics_df,

        "classification_df":
            classification_df,

        "prediction_df":
            prediction_df

    }


# ============================================================
# 14. RUN PROGRAM
# ============================================================

if __name__ == "__main__":

    results = main()
```
